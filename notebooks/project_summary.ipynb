## Current scientific position

> **October 1, 2026 override note.** The six numbered points below preserve the September-16 scientific position. Sections BO–BW in Part III supersede the status of the universal reference-free contrast-scale estimator, the Li2020 benchmark program, and the M03 C-prime numerical backend. In particular, the project now has a seam-safe exact-when-feasible C-prime implementation and has moved the biological-null benchmark to a fresh 20M-support-cap run. The final 20M benchmark is still in progress.

> **October 3, 2026 override note.** Part IV supersedes the October-1 status of the ordinary DEB precision layer. The
> exact-normalizer fallback, support-feasible null geometry, WEB-DP curvature layer, first-order ordinary branch, and
> seam-safe 20M C-prime backend remain supported. What has been reopened is **how strongly the ordinary full/null
> precision estimates should be shrunk toward one global DEB prior mean**. The completed 30-split Pickrell null audit
> shows that the current empirical-Bayes hyperparameter fit usually implies essentially total shrinkage,
> \(s=\alpha_0/(a+\alpha_0)\approx1\), while the raw precision \(a/R\) contains a strong, reproducible coverage-dependent
> shape. Tempering that shrinkage produces a broad region of dramatically improved null calibration. The archived MAQC
> results remain valid results of the archived method and are not retroactively rescored. The adjusted-DEB branch is now
> the leading precision-layer development candidate and must be frozen and externally validated on Li2020 before it can
> replace the archived DEB rule.

> **October 3, 2026, 9:11 PM EDT override note.** Part V supersedes the Part-IV conclusion that the immediate production
> question is simply whether to cap the fitted DEB concentration. Direct capped-DEB transfer to the available Li2020
> 40v40 splits did not by itself solve the biological-null problem, and cross-dataset audits showed that the raw
> precision-coverage shape is strongly reproducible within a dataset but differs materially in amplitude between Pickrell
> and Li2020. The decisive advance was instead to model the **conditional null law of the final clipped statistic**
> \(Q=\max(T_{\rm raw},0)\) as coverage-dependent. A five-fold cross-fitted hurdle-Gamma model
> \(\pi_0(x)\delta_0+[1-\pi_0(x)]\Gamma(k(x),\theta_\Gamma(x))\) nearly removed the coverage hump in Pickrell and all three
> available Li2020 40v40 datasets and improved Li2020 benchmark FPR in every one of the 49 selected splits. After that
> correction, DEB shrinkage strength still mattered: a joint search that re-fit the hurdle law separately at every
> universal shrinkage fraction produced a broad moderate-shrinkage basin, with \(s=.60\) selected by both equal-dataset
> mean calibration and minimax calibration. With \(s=.60\) frozen, replacing the positive Gamma component by a
> one-extra-parameter generalized Gamma further improved the Li2020 \(10^{-4}\)–\(10^{-3}\) tail and benchmark FPR, while
> Pickrell estimated \(c\approx1\) and therefore remained essentially Gamma. The current unresolved defect is now much
> narrower: the **ultra-extreme first-BH tail, roughly \(10^{-6}\)–\(10^{-5}\), remains too heavy**, and the next diagnostic
> must distinguish a continuously heavier tail from a small contaminating/outlier component. These null-benchmark
> empirical-law fits are development evidence; a production estimator for datasets containing real alternatives remains
> to be specified and externally validated.


> **October 5, 2026, 10:12 AM EDT override note.** Part VI supersedes the Part-V implication that the next immediate
> production step is simply to add more positive-tail flexibility to the coverage-only hurdle law. The ultra-extreme-tail
> audit showed that the remaining first-BH excess is concentrated in a tiny recurrent gene subset and is strongly associated
> with sample-dominance structure in the Li2020 splits. Adding sample dominance as a second conditioning coordinate can
> repair that structure, but an unpooled two-dimensional empirical null is too variable, especially in Pickrell. The safer
> current development architecture partially pools a coverage-only hurdle-Gamma with a coverage-plus-dominance hurdle-Gamma,
> with the pooling weight learned inside nested cross-validation and the training contribution of likely alternatives reduced
> by Stage-03 ASH posterior-null weights. This branch was frozen before RNAontheBENCH NanoString truth was opened. On that
> benchmark it preserved the exact same raw-\(p<.01\) operating point as current M03 (18 TP / 11 FP) and slightly reduced AUC
> (0.9398 versus 0.9431), while preserving all high-dominance true-positive calls. RNAontheBENCH was then explicitly converted
> to development data for a universal-\(s\) scan. That scan showed broad discrete operating-point plateaus rather than a unique
> optimum and exposed a deeper issue: heavy DEB shrinkage can make low-raw-precision genes more significant while suppressing
> high-raw-precision genes. The immediate scientific problem is therefore no longer “pick the best universal \(s\),” but
> “estimate how trustworthy each gene's raw precision information is without using the tested signal or truth labels.”


> **October 5, 2026, 6:10 PM EDT override note.** Part VII supersedes the Part-VI implication that the next precision-layer
> step is simply to construct a gene-specific shrinkage fraction from a reliability score. The new LOO audits show that
> gene-specific raw-precision reliability is empirically real: unstable genes are enriched among Pickrell null extremes,
> while RNAontheBENCH genes harmed by global shrinkage are, on average, more stable. However, two direct attempts to turn
> that information into a new precision estimator did not supply a final method. A mean-coverage-dependent EB target failed
> to improve calibration, and the parameter-free rule \(s_g=1-\) reliability percentile transferred in the desired
> direction but only modestly and has no statistical derivation tying a percentile rank to an exact amount of shrinkage.
> A target-standardized \(Z\) score successfully predicts the direction and magnitude of shrinkage's effect but performs
> worse than reliability alone as a shrinkage-need diagnostic. The percentile rule is therefore archived as an experiment,
> not promoted into M03. The pre-detour fitted-DEB precision layer is reopened as a final method because its empirical-Bayes
> concentration can collapse toward nearly total pooling. The next principled precision-layer target is to model
> gene-specific uncertainty in \(\log\theta_g\) explicitly—e.g. through a heteroskedastic measurement-error EB model—or
> to integrate \(\theta_g\) uncertainty directly in the full/null likelihood. The conditional-null-law work remains a
> separate layer and is not invalidated by this precision-model decision.


### 1. The underlying M03 architecture remains valuable; its universal contrast-scale estimator is not settled

The established count-model backbone remains:

\[
\text{WEB-DP }r_g
\; +\; \text{ordinary first-order }C'(1)\text{ / DEB precision}
\; +\; \text{targeted exact normalized-DNB precision fallback}
\; +\; \text{genome-wide null-support feasibility}.
\]

The exact-normalizer and support-floor repairs retain their separate mathematical motivations. The completed MAQC results remain important benchmark-specific successes. The later Mixology experiments do **not** erase those numerical results.

What Mixology reopened is the assumption that the ASH-weighted observed contrast increment always supplies a dependable \(\kappa_c\). In the dense-signal mixture setting, the original estimates tracked biological contrast strength and strongly distorted cross-contrast significance behavior. Therefore the older phrase “finalized M03 production method” now means the archived MAQC-era implementation, **not a fully validated universal solution to contrast-scale estimation**. [S01, S06–S08]

### 2. The leading operational comparison is fixed \(\lambda_c=1\) versus residual-based Strategy A

Three branches now need distinct names:

| Branch | Meaning | Current position |
|---|---|---|
| **Original M03** | MAQC-era M03 with ASH-weighted observed \(\kappa_c\), \(\kappa_F=1\), and \(\lambda_c=\kappa_c\) | Retain as historical baseline; dense-signal Mixology failure documented. |
| **M03, fixed \(\lambda=1\)** | Same fitted count-model backbone, with directional rescaling disabled | Strong simple baseline; best M03 branch on the displayed overall Mixology rank-monotonicity measures. |
| **M03, Strategy-A residual \(\lambda_c\)** | Cross-fitted residual path, robustification, NB-based finite-sample reference calibration, then an A-basin choice | Active empirical candidate. Directly tested; large improvement over original M03, but no overall win over fixed one established. |

The direct Mixology all-gene median signal/significance Spearman values were **0.564** for original M03, **0.782** for A-trim, **0.790** for A-winsor, and **0.806** for fixed one. The comparator values were **0.818** for edgeR and **0.830** for DESeq2. These are reference-monotonicity diagnostics, not measured FDR or a universal ranking of methods. [S16]

### 3. The residual direction produced real diagnostic progress, but several intermediate successes were overstated

The experiments now distinguish four different objects:

\[
\text{same-fit residual normalization},\qquad
\text{out-of-fit prediction error},\qquad
\text{a robust residual scale functional},\qquad
\text{the inferentially optimal directional }\lambda_c.
\]

They are not automatically interchangeable.

The same-fit HC2 residual average near one was partly forced by the precision normalization already performed by DEB. Strict sample cross-fitting removed that immediate self-fit identity but exposed zero-training-count predictions with effectively zero plug-in variance. A small number of these observations made ordinary squared-residual means enormous. The boundary audit identified the responsible counts directly.

Robust trimming, winsorization, and median functionals stabilized the calculation. NB unit-scale simulations showed that the real-data robust values were not fully explained by that particular simulated reference experiment. Nevertheless, the completed calibration used **ordinary NB with \(\theta=1\)**, not the fitted normalized-DNB distribution at each gene's original \(\theta_g\). It supports a departure from that NB reference, not a finished proof of a unique true contrast-noise multiplier. [S09–S13]

### 4. A is the current optimization candidate; B is diagnostic, C is demoted

Two signal-dependence constructions were tested: across contrasts using observed fitted effect strength, and within each contrast using gene-level residual scores versus observed absolute fold change. Both trimming and winsorization were carried through the tests.

The first-flat-Spearman rule selected the first permitted tail percentage almost everywhere and did not identify a convincing transition. Smoothing-spline objectives were then tested:

\[
J_A=\rho^2+(\rho')^2,
\qquad
K_B=\frac{|\rho''|}{(1+(\rho')^2)^{3/2}},
\qquad
J_C=\rho^2+(\rho')^2+(\rho'')^2.
\]

The latest decision is to retain **A** as the primary candidate, retain **B** only as an elbow diagnostic, and demote **C** because its second-derivative term added instability without a demonstrated benefit. This supersedes the earlier temporary enthusiasm for B. Most winsorized A solutions were search-boundary solutions; this remains a limitation, not evidence of a well-identified interior optimum. [S14–S16]

### 5. Bottomly is out of the active evidence program

The project tried historical verification splits, full-21 references, multiple proxy-truth definitions, shared/unique rank balancing, exact-precision diagnostics, and a no-TMM M03 comparison. The inferred comparison depended substantially on how proxy truth was defined. The no-TMM check changed M03 results only modestly and did not provide a normalization rescue.

Sean's decision was to exclude Bottomly because its information was too difficult to interpret. The record is retained for transparency; the older roadmap's recommendation to use Bottomly as a decisive next validation is superseded. [S02, S04]

### 6. The current oracle search is pending, and its objective must be named precisely

The new search asks, for each Mixology contrast, which \(\lambda_c\) maximizes the **across-gene** Spearman relationship between absolute fitmixture reference effect and significance. It preserves upstream M03 stages and reruns the downstream \(\lambda\)-dependent calculation.

No completed oracle curve or oracle optimum has been reported yet. One candidate took approximately **94 seconds**. The earlier **19–24 hour** runtime discussion was a rough extrapolation, not a measured completion time. The code searches adaptively and can do more or fewer evaluations depending on overlap, expansion, refinement, caching, and fallback costs. [S17]

This oracle is an **evaluation-only, reference-rank oracle within a finite search range**. It is not automatically the global optimum for cross-contrast monotonicity, FDR, power, or the full DGE procedure. A newly identified static code-review issue in the supplied oracle runner is recorded in Sections BJ and BN; the code was not modified as part of this summary.

# 1. WEB-Seq / DEB-Seq: progression from the original method to the current framework

> **Current status — September 2026.**
> The project now has a working general-`X` architecture for DEU/DTU and a substantially revised DGE architecture. For DEU/DTU, the fixed-full-model family formulation remains: ASH supplies event-level null membership, one shared family full scale is estimated from pooled null information, each contrast receives one directional `\kappa_c`, and only the added null-constraint increment is corrected. For DGE, the current primary branch is the former M03 architecture: WEB-DP estimates `r_g`, Gamma-Gamma DEB estimates the ordinary residual precision, `\kappa_F=1`, ASH estimates the contrast-direction scale `\kappa_c`, and `\lambda_c=\kappa_c`. Two independent admissibility corrections are now part of the DGE procedure. First, if the first-order full residual coordinate `R_F^{(1)}=S_F-C'_F(1)` is nonpositive, the local `C'(1)` approximation is abandoned for that gene and the normalized double-NB precision profile is solved exactly in `\theta` using repeated `C'(\theta)` / `\log C(\theta)` evaluation. Second, for every gene, the geometry-corrected null residual statistic is checked against the minimum half-deviance permitted by the discrete null support; if the proposed `S_0^\dagger` lies below that floor, only the shrinkable residual increment above the support floor is divided by `\lambda_c`. The final support-feasible M03 rerun produced a finite p-value for every analyzed MAQC2 and MAQC3 gene while retaining the strong qPCR-reference performance described below. The earlier 12-method grid remains part of the development history, but DEB `\theta` is now the leading precision branch and the exact-normalizer / support-feasibility corrections are treated as mathematical repairs rather than truth-label tuning.

## A. Original WEB-Seq

WEB-Seq began with a double-binomial model for differential usage in the simple treatment-control setting:

```
Y_{gi}\sim DB(m_{gi},p_{gi},\theta_g).
```

The familiar working moments are

```
E(Y_{gi})\approx m_{gi}p_{gi},
\qquad
\operatorname{Var}(Y_{gi})
\approx
\frac{m_{gi}p_{gi}(1-p_{gi})}{\theta_g}.
```

These are nominal/approximate moments of the double-exponential construction; the exact normalized distribution should be distinguished from the working moment approximation.

With half-deviance

```
S_g=\frac12\sum_iD_{gi},
\qquad
a_g=\frac{d_g}{2},
```

the residual-precision working likelihood is proportional to

```
\theta_g^{a_g}e^{-\theta_gS_g}.
```

WEB shrinks the event-specific residual statistic toward a population target:

```
\widehat\theta_g^{WEB}
=
\frac{a_g}
{(1-w)S_g+wS_{\mathrm{target},g}}.
```

The related DEB Gamma-Gamma empirical-Bayes formulation uses

```
\widehat\theta_g^{DEB}
=
\frac{a_g+\widehat\alpha_0}
{S_g+\widehat\beta_0}.
```

The original two-group implementation worked well in the reported DEU experiments. That empirical success remains part of the project; it does not establish that every approximation in the original construction is exact.

### Parallel finite-sample work

Earlier notebook work examined small-sample mean fitting, approximate leave-one-out, exact leave-one-out checks, information-based alternatives to heuristic effective sample sizes, and empirical residual-moment surfaces. Interior and non-interior cases behaved differently.

Those questions were not automatically resolved by the later contrast-geometry work and remain part of the benchmark program.

---

## B. General `X`: arbitrary contrasts exposed a missing information-scale problem

The mean model was generalized to

```
\eta_g=X\beta_g,
\qquad
H_0:c^\top\beta_g=0.
```

The constrained fit was constructed through the contrast’s coefficient-space null hyperplane rather than assuming every hypothesis can be implemented by deleting a particular design column.

In the complex Tasic design, different contrasts within the same full model exhibited very different apparent null scales—approximately `2,4,1.2,4` for the contrasts emphasized in the earlier experiments.

One event-level residual precision therefore did not explain all contrast-to-contrast behavior.

This motivated separating

```
\theta_g
=
\text{event-specific residual precision}
```

from

```
\kappa_c
=
\text{information scale associated with the tested contrast direction}.
```

---

## C. First practical success: external signed-`Z` empirical-null scaling

A nuisance-adjusted signed contrast statistic was constructed across events. Empirical-null fits then estimated a contrast-specific scale, producing

```
T_{gc}^{external}
=
\frac{T_{gc}^{WEB}}
{\widehat\kappa_c}.
```

The original Tasic-derived 20-vs-20 truth benchmarks showed substantial calibration improvements under this correction.

The exact benchmark truth construction must remain attached to those results; natural biological differences alone do not constitute complete gene- or transcript-level truth.

### Current empirical-null policy

ASH is now the primary empirical-null/null-membership estimator for the flagship model.

For contrast `c` and gene/event `g`, write

```
w_gc
=
P(H0_gc | data)
```

for the ASH posterior null weight and

```
pi_0c
```

for the ASH-estimated contrast-level null fraction.

The earlier 2-Normal estimator is no longer an equal operational alternative. In the Pickrell null work it repeatedly collapsed to the imposed

```
sigma_0 = 0.05
```

lower bound and could produce severely pathological calibration.

It is retained only as a historical comparator, stress test, or negative-control diagnostic.

The external signed-`Z` empirical-null scale was crucial for discovering the missing contrast-direction scale, but it is also no longer the flagship correction. The scale is now estimated and applied inside the nested residual geometry.

---

## D. Absorbing the contrast scale into precision shrinkage was not sufficient

Experiments tried modifications such as

```
a\longrightarrow \frac a\kappa,
\qquad
S\longrightarrow\frac S\kappa.
```

Simply pushing the contrast scale into the empirical-Bayes precision layer did not reproduce the successful external correction.

Operationally,

```
\boxed{\theta_g\ne\kappa_c}
```

in the sense that they play distinct roles and cannot generally substitute for one another.

This does not require them to be statistically independent; it means their inferential functions differ.

---

## E. Residual-space geometry supplied a structural interpretation

In a common quadratic residual coordinate system, a one-constraint comparison has

```
P_{0,c}
=
P_F+u_cu_c^\top.
```

Therefore

```
Q_0
=
e^\top P_{0,c}e
=
Q_F+(u_c^\top e)^2.
```

Write

```
E(Q_F)=d_F\kappa_F,
\qquad
E(Q_0)=d_0\kappa_0.
```

For one added residual direction,

```
\boxed{
\kappa_c
=
d_0\kappa_0-d_F\kappa_F.
}
```

The projector decomposition is an identity in the common quadratic geometry.

Transferring this identity to nonlinear fitted deviances and estimated information is an additional modeling/approximation step. Likewise, matching an expected quadratic scale does not itself prove a full chi-square reference distribution.

---

## F. The geometry independently tracked signed-`Z` scale

An important development-stage result was that the first residual-geometry estimate closely tracked the signed-`Z` empirical-null scale in the Tasic contrasts.

Reported examples from that earlier implementation were approximately:

| ContrastSigned-`Z` scaleEarly geometry `\kappa_c`<br><br> |      |      |
| --------------------------------------------------------- | ---- | ---- |
| C1                                                        | 1.93 | 2.03 |
| C6                                                        | 4.25 | 4.20 |
| C11                                                       | 1.16 | 1.16 |
| C14                                                       | 3.97 | 3.91 |

This was strong evidence that the signed-`Z` phenomenon was not merely an artifact of the external empirical-null fit: a separate residual-space calculation found essentially the same directional distortion.

These values are now **historical development estimates**, not the current flagship `\kappa_c` estimates.

The later fixed-full-model/ASH implementation changed the null weighting, the exact geometry estimator, and the way the common full scale is constructed. The current Tasic flagship results are summarized below.

---

## G. Failed approach: separately rescaling shared full/null residual spaces

The transformation

```
S_F^*
=
\frac{S_F}{\kappa_F},
\qquad
S_0^*
=
\frac{S_0}{\kappa_0}
```

failed, including with oracle scales.

The reason follows directly from

```
Q_0=Q_F+Q_c.
```

Separately rescaling gives

```
\frac{Q_0}{\kappa_0}
-
\frac{Q_F}{\kappa_F}
=
Q_F
\left(
\frac1{\kappa_0}
-
\frac1{\kappa_F}
\right)
+
\frac{Q_c}{\kappa_0}.
```

The shared residual component `Q_F` receives inconsistent scaling.

This was the key realization leading to the nested/contrast-increment construction.

---

# H. Current general-`X` flagship for DEU/DTU: one family full scale plus contrast-specific directional scales

The contrast-increment correction remains the central geometry:

```
S_F^\dagger=S_F
```

and

```
\boxed{
S_{0,c}^\dagger
=
S_F
+
\frac{S_{0,c}-S_F}{\lambda_c}.
}
```

The current flagship definition of `\lambda_c` is now more specific.

For a fixed full model define, for each gene/event,

```
Q_Fg
=
\text{full-model residual quadratic/score contribution},
```

and for each tested contrast,

```
Q_cg
=
\text{added contrast-direction contribution}.
```

In the large-`n` DEU implementation,

```
Q_Fg
=
2\theta_FgS_Fg,
```

```
Q_cg
=
2\theta_Fg(S_0cg-S_Fg).
```

This same nested contrast-increment idea was then carried into DGE after normalizer correction, but the later DGE `kappa_F` audit changed how the shared full scale is handled. Section W supersedes the DGE denominator rule while preserving the same structural principle that the added contrast increment remains distinct.

Let

```
w_gc
=
P(H0_gc | data)
```

be the ASH posterior null weight and let `q_c` be the prespecified scientific probability/importance assigned to contrast `c` in the planned contrast family.

The common full-model scale is estimated once across the whole contrast family:

```
\boxed{
\widehat\kappa_F^{family}
=
\frac{
\sum_c q_c\sum_g w_{gc}Q_{Fg}
}{
\sum_c q_c\sum_g w_{gc}d_{Fg}
}.
}
```

The contrast-specific added-direction scale is

```
\boxed{
\widehat\kappa_c
=
\frac{
\sum_g w_{gc}Q_{cg}
}{
\sum_g w_{gc}d_{cg}
}.
}
```

Then

```
\boxed{
\widehat\lambda_c
=
\frac{\widehat\kappa_c}
{\widehat\kappa_F^{family}}.
}
```

This is the current flagship general-`X` geometry for DEU/DTU. The later DGE specialization is given in Section W.

### Why the family full scale is preferred

For a fixed full model, `Q_Fg` contains no contrast-specific direction.

Therefore the full residual geometry is one common object.

Different contrasts provide different amounts of null information about which genes/events should contribute to estimating that same object.

The family estimator pools **posterior null mass**, not a literal estimate of a boundary probability.

Its absolute scale does not need to equal one.

If all ASH null weights are multiplied by a common factor, that factor cancels between numerator and denominator.

This resolves the boundary problem that appeared when the arithmetic mean of bounded ASH weights was interpreted as an estimator of a gene-level probability that could equal one.

### Role of `pi_0c`

Within each contrast,

```
mean_g(w_gc)
\approx
pi_0c.
```

Therefore a contrast with a larger null fraction automatically contributes more total posterior-null information to `\widehat\kappa_F^{family}`.

In the current Tasic run this happened almost exactly.

An additional factor of `pi_0c` was tested:

```
q_c w_gc
\longrightarrow
q_c pi_0c w_gc.
```

Operationally this approximately gives the global null fraction a second weight and behaves roughly like `pi_0c^2` weighting.

It changed the Tasic full scale negligibly and is **not** part of the flagship estimator.

### Status of `\kappa_{F,c}`

The contrast-specific estimator

```
\widehat\kappa_{F,c}
=
\frac{
\sum_g w_{gc}Q_{Fg}
}{
\sum_g w_{gc}d_{Fg}
}
```

is still calculated, but only as a diagnostic/audit.

If the full model is truly shared, the different `\widehat\kappa_{F,c}` values should agree up to weighting noise.

Large disagreement is now treated as evidence to investigate the contrast family, null weighting, or model specification rather than as a reason to give each contrast its own full-model denominator.

### Status of gene-specific `\lambda_{cg}`

Gene-specific interpolation between `\kappa_{F,c}` and `\kappa_F^{family}` was explicitly tested using both arithmetic-`\lambda` and Gamma-rate constructions.

It added essentially no inferential information in Tasic because the full-model scales were already nearly identical.

Combined with the earlier conditional/local-geometry experiments, there is currently no empirical reason to carry an operational `\lambda_{cg}`.

The flagship model therefore uses

```
\boxed{
\lambda_c,
\quad
\text{not }\lambda_{cg}.
}
```

ASH supplies the gene-wise null-membership calibration; the remaining residual geometry is modeled at the contrast level.

The central principle remains:

```
\boxed{
\text{shared residual information stays shared;}
\quad
\text{only the added contrast direction is recalibrated.}
}
```

# K. Important interpretation corrections and closed questions

**Scope clarification after the later DGE audit:** the family-`\kappa_F` conclusions in this section remain the current DEU/DTU general-`X` formulation. The DGE-specific role of `\kappa_F` was revised later and is summarized in Section W: first-order `C'(1)` DGE uses `\kappa_F\equiv1`, while theta-shared DGE uses the simple observed shared-coordinate estimator.

A balanced design does **not** automatically imply

```
\lambda_c=1.
```

The relevant question is whether the tested contrast direction and the common residual geometry have the same effective scale.

Likewise, the success of the original two-group WEB method is not established to have been “luck.”

It may reflect a genuine special case, an approximation that is particularly accurate there, or compensating effects.

Boundary/discreteness effects are distinct from contrast geometry.

A scalar `\lambda_c` cannot automatically eliminate every weak-information, separation, or finite-sample distortion.

Different fitted nuisance precisions do not make two statistical model families nonnested by themselves.

Nesting concerns parameter spaces, not whether fitted nuisance values happen to differ.

Several previously open architecture questions are now closed for the current flagship model:

1. **The full residual geometry is shared across contrasts from one fixed full model.**
   `\kappa_{F,c}` is diagnostic, not an operational denominator.
2. **The planned contrast family should estimate one** **`\kappa_F^{family}`****\*\*\*\*.**
   Contrasts contribute according to posterior null information.
3. **ASH is the operational gene/event-level null-membership calibration.**
   Its weights need not average to one for the family `\kappa_F` ratio to be valid as a relative-information weighting scheme.
4. **An extra** **`pi_0c`** **factor is unnecessary.**
   `pi_0c` is already reflected in the total ASH null mass contributed by each contrast.
5. **A gene-specific** **`\lambda_{cg}`** **is not currently warranted.**
   The tested local/interpolation schemes added essentially no useful information once ASH weighting and contrast-level `\kappa_c` were in place.
6. **The boundary-aware spike-plus-Beta family hierarchy is not needed for the flagship estimator.**
   It addressed the earlier interpretation of a pooled bounded mean as a literal boundary probability; posterior-null-mass weighting inside the `\kappa_F` ratio avoids that problem more directly.
7. **External signed-\*\*\*\*****`Z`** **rescaling is no longer the operational correction.**
   It remains historically important because it exposed the contrast-scale problem and independently corroborated the geometry.

# Current unified flagship architecture for DEU/DTU

For a prespecified family of contrasts through one fixed full model:

```
1. Fit the full mean model once.
2. Estimate event/gene residual precision theta_F.
3. For each contrast c, fit the constrained/null model.
4. Use ASH to obtain w_gc = P(H0_gc | data).
5. Estimate one family full scale kappa_F^{family}.
6. Estimate one directional scale kappa_c for each contrast.
7. Set lambda_c = kappa_c / kappa_F^{family}.
8. Preserve S_F and divide only the added null increment by lambda_c.
9. Refit the null precision on S_0,c^\dagger and evaluate the WEB/DEB test.
```

The operative geometry is

```
\boxed{
\widehat\kappa_F^{family}
=
\frac{
\sum_c q_c\sum_g w_{gc}Q_{Fg}
}{
\sum_c q_c\sum_g w_{gc}d_{Fg}
},
}
```

```
\boxed{
\widehat\kappa_c
=
\frac{
\sum_g w_{gc}Q_{cg}
}{
\sum_g w_{gc}d_{cg}
},
}
```

```
\boxed{
\lambda_c
=
\frac{\widehat\kappa_c}
{\widehat\kappa_F^{family}},
}
```

and

```
\boxed{
S_{0,c}^\dagger
=
S_F
+
\frac{S_{0,c}-S_F}{\lambda_c}.
}
```

This is the current working general-`X` model for DEU/DTU.

# Expansion into differential gene expression

## L. Double-Poisson starting point and the negative-binomial bridge

The double-Poisson idea gives a linear working variance,

```
V(\mu)
\approx
\frac{\mu}{\theta}.
```

The negative binomial introduces curvature:

```
V_{NB}(\mu)
=
\mu+\frac{\mu^2}{r_g}.
```

For fixed NB size `r`, define

```
p
=
\frac{\mu}{\mu+r},
\qquad
m^{eff}
=
y+r.
```

The mean-dependent likelihood kernel is

```
p^y(1-p)^r,
```

with

```
\operatorname{logit}(p)
=
\log\mu-\log r.
```

Thus, with sequencing offset `L_i`,

```
\log\mu_{gi}
=
\log L_i+x_i^\top\beta_g
```

becomes

```
\operatorname{logit}(p_{gi})
=
x_i^\top\beta_g+\log L_i-\log r_g.
```

For fixed `r`, the NB deviance agrees with the corresponding effective-binomial deviance.

### Important scope

This is a likelihood/deviance representation.

Because

```
m^{eff}=y+r
```

depends on the realized outcome and need not be an integer, it does **not** mean that NB sampling literally becomes ordinary fixed-denominator binomial sampling.

# M. Proposed enlarged DGE variance structure

> **Later revision:** the variance decomposition below remains useful for distinguishing `r_g` and `\theta_g`, but the extra DGE full-scale term was subsequently simplified. Section W supersedes the operational `\kappa_F` rule: `\kappa_F=1` in the first-order coordinate and a simple observed `\kappa_F^\theta` in the theta-shared coordinate.

The proposed double-NB construction has working variance

```
\operatorname{Var}(Y_{gi})
\approx
\frac{
\mu_{gi}+\mu_{gi}^2/r_g
}{
\theta_g
}.
```

The intended roles are:

```
r_g
=
\text{NB curvature},
```

```
\theta_g
=
\text{multiplicative residual precision},
```

```
\kappa_F^{family}
=
\text{shared full-residual geometry},
```

```
\kappa_c
=
\text{contrast-direction geometry},
```

```
\lambda_c
=
\kappa_c/\kappa_F^{family}.
```

Ordinary NB corresponds to

```
\theta_g=1.
```

Away from `\theta=1`, exact moments depend on the normalizer, so the displayed variance remains a working expression rather than an exact universal moment identity.

# N. Candidate `r_g` estimator: double-Poisson information plus WEB shrinkage

The current candidate pipeline is

```
Y
\rightarrow
\widetilde\mu_F^{Poisson}
\rightarrow
S_g^{DP}
\rightarrow
\widehat\phi_g
\rightarrow
\widehat r_g.
```

Use

```
S_g^{DP}
=
\frac12
\sum_i
D_P(Y_{gi},\widetilde\mu_{F,gi}),
```

and

```
\phi_g^{raw}
=
\frac{S_g^{DP}}{a_F}.
```

Fit a working mean-Fano trend,

```
\phi_{trend}(\bar\mu_g)
=
1+b+\alpha_\infty\bar\mu_g,
```

shrink the observed residual statistic toward its trend target, and calculate

```
\widehat\alpha_g
=
\frac{\widehat\phi_g-1}{\bar\mu_g},
\qquad
\widehat r_g
=
\frac1{\widehat\alpha_g}.
```

The initial compcodeR known-truth experiment gave:

| EstimatorSpearman with true `\alpha=1/r`Median `\hat\alpha/\alpha`Within 2-fold<br><br><br> |       |       |       |
| ------------------------------------------------------------------------------------------- | ----- | ----- | ----- |
| Raw DP                                                                                      | 0.880 | 0.852 | 0.773 |
| Pearson                                                                                     | 0.844 | 0.825 | 0.700 |
| Trend only                                                                                  | 0.548 | 1.588 | 0.463 |
| DP + WEB                                                                                    | 0.856 | 1.108 | 0.804 |

DP + WEB improved calibration and the within-two-fold proportion but did not beat raw DP in rank correlation.

It remains a candidate estimator, not a fully validated final solution.

# P. Exact double-NB normalizer and completed score audit

Write the normalized double-NB model as

```
f_\theta(y;\mu,r)
=
C(\mu,r,\theta)
\sqrt\theta
f_{NB}(y;\mu,r)^\theta
f_{sat}(y;r)^{1-\theta}.
```

Let

```
s(y;\mu,r)
=
\frac12D_{NB}(y;\mu,r).
```

Then

```
\boxed{
\log C(\mu,r,\theta)
=
-\frac12\log\theta
-
\log
\sum_{y=0}^\infty
f_{sat}(y;r)e^{-\theta s(y;\mu,r)}.
}
```

Differentiating gives the general identity

```
\boxed{
\frac{\partial\log C}{\partial\theta}
=
-\frac1{2\theta}
+
E_\theta[D/2].
}
```

At the NB reference point,

```
C(\mu,r,1)=1,
```

and

```
\boxed{
c_1(\mu,r)
:=
\left.
\partial_\theta\log C
\right|_{\theta=1}
=
E_{NB}[D/2]
-
\frac12.
}
```

### Numerical audit

Across 20 `(\mu,r)` combinations:

```
log C(mu,r,1) discrepancy              ~ numerical zero
five-point derivative vs identity      2.92e-12 max discrepancy
expected-deviance identity residual    0
expected-score residual                0
```

The estimated derivative ranged roughly from

```
-0.135
\quad\text{to}\quad
0.121.
```

Thus the correction is not uniformly positive.

# Q. First-order residual-score correction

Conditional on fixed `\mu` and `r`,

```
\ell_M(\theta)
=
\frac n2\log\theta
-
\theta S_M
+
\sum_i\log C(\mu_{Mi},r,\theta)
+
\text{common saturated term}.
```

At `\theta=1`, define

```
c_M(1)
=
\sum_i c_1(\mu_{Mi},r).
```

Using the first-order approximation

```
\sum_i\log C_i(\theta)
\approx
c_M(1)(\theta-1),
```

and the residual-degree-of-freedom working adjustment

```
a_M
=
\frac{d_M}{2},
```

the score is

```
\frac{a_M}{\theta}
-
S_M
+
c_M(1).
```

Hence

```
\boxed{
\widehat\theta_M^{(1)}
=
\frac{a_M}
{S_M-c_M(1)}
}
```

whenever the corrected denominator is positive.

The pooled version is

```
\widehat\theta
=
\frac{
\sum_g a_{Mg}
}{
\sum_g\{S_{Mg}-c_{Mg}(1)\}
}.
```

### Recovery of NB truth

With true means:

```
\hat\theta_{naive}
\approx
0.9427,
```

whereas

```
\boxed{
\hat\theta_{corrected}
\approx
1.0166.
}
```

With fitted two-group means:

```
0.9219
\longrightarrow
\boxed{1.0093}.
```

Thus the normalizer correction strongly reduced precision bias in the known ordinary-NB experiment.

# R. First-order normalized likelihood statistic

At specified precision estimates,

```
\boxed{
T^{(1)}
=
2\{
\widehat\theta_0(S_0-c_0)
-
\widehat\theta_F(S_F-c_F)
\}
+
n\log
\frac{\widehat\theta_F}{\widehat\theta_0}
+
2(c_0-c_F).
}
```

The final

```
2(c_0-c_F)
```

term is required by the Taylor expansion.

The explicit formula matched direct evaluation of the same Taylor-expanded likelihood difference to approximately

```
5\times10^{-15}
```

maximum numerical error.

For null genes with fitted `\theta` near one, the first-order statistic tracked the exact-normalizer statistic well:

```
median absolute Taylor-only error
\approx
0.019,
```

```
95% error
\approx
0.164,
```

```
cor
\approx
0.993.
```

Errors became much larger when fitted `\theta` moved far from one, especially among alternatives.

Thus:

```
\boxed{
C'(1)
\text{ is genuinely a local approximation around }
\theta=1.
}
```

# V. `\theta_F`-aware normalizer geometry: correction to the previous strategy

This section supersedes the older version of the summary.

The first `C'(1)` approach evaluates the normalizer derivative at the ordinary-NB reference point:

```
c_M(1)
=
\sum_i
\left.
\partial_\theta\log C(\mu_{Mi},r,\theta)
\right|_{\theta=1}.
```

The exact derivative at arbitrary precision is instead

```
c_M(\theta)
=
\sum_i
\frac{\partial\log C(\mu_{Mi},r,\theta)}
{\partial\theta}.
```

That suggested testing a `\theta_F`-aware geometry.

---

## V.1 Retired two-score `\theta`-aware construction

The first attempted version used

```
Q_F^\theta
=
2\theta_F
\{S_F-C'_F(\theta_F)\}
```

and independently

```
Q_0^\theta
=
2\theta_F
\{S_0-C'_0(\theta_F)\}.
```

This looked superficially natural because the same full-model `\theta_F` was used as the precision reference on both sides.

However, the first two Pickrell replicates showed that this construction was materially worse than the standard `C'(1)`-normalized geometry.

### All-null oracle

| ReplicateStandard `C'(1)`Old two-score `\theta`-aware<br><br> |        |        |
| ------------------------------------------------------------- | ------ | ------ |
| 1                                                             | 0.0122 | 0.0215 |
| 2                                                             | 0.0081 | 0.0169 |

Averages:

```
\boxed{
0.01015
\quad\text{versus}\quad
0.0192.
}
```

### ASH

| ReplicateStandard `C'(1)`Old two-score `\theta`-aware<br><br> |        |        |
| ------------------------------------------------------------- | ------ | ------ |
| 1                                                             | 0.0146 | 0.0266 |
| 2                                                             | 0.0109 | 0.0236 |

Averages:

```
\boxed{
0.01275
\quad\text{versus}\quad
0.0251.
}
```

Thus the old `\theta`-aware construction approximately doubled the FPR relative to the ordinary first-order normalized geometry in these two development splits.

### Why it failed

The standard normalized boundary `\lambda_c` estimates were approximately

```
1.493,
\qquad
1.568,
```

whereas the old two-score `\theta`-aware calculation reduced them to approximately

```
1.013,
\qquad
1.097.
```

The full-residual scales changed little.

The problem was therefore concentrated in the contrast increment.

The expression

```
Q_0^\theta
=
2\theta_F
\{S_0-C'_0(\theta_F)\}
```

implies

```
Q_0^\theta-Q_F^\theta
=
2\theta_F
[
(S_0-S_F)
-
\{C'_0(\theta_F)-C'_F(\theta_F)\}
].
```

Thus the model-specific normalizer difference

```
C'_0(\theta_F)-C'_F(\theta_F)
```

was being inserted directly into the tested contrast direction.

That violates the structural principle established earlier:

```
\boxed{
\text{the shared full residual space and the added contrast direction must remain distinct.}
}
```

The old two-score `\theta`-aware geometry is therefore retired.

---

# V.2 Shared-score `\theta_F`-aware strategy

The corrected construction makes only the **shared full residual score** `\theta`-aware:

```
\boxed{
Q_F^\theta
=
2\theta_F
[
S_F-C'_F(\theta_F)
].
}
```

The added nested contrast direction remains exactly

```
S_0-S_F.
```

Therefore

```
\boxed{
Q_0^\theta
=
Q_F^\theta
+
2\theta_F(S_0-S_F).
}
```

and

```
\boxed{
Q_c^\theta
=
Q_0^\theta-Q_F^\theta
=
2\theta_F(S_0-S_F).
}
```

This identity is central.

The `\theta_F`-aware normalizer can change the coordinate used for the **shared residual geometry**, but it does not alter the tested contrast increment.

WEB shrinkage and the final first-order `C'(1)` normalized likelihood statistic remain unchanged in this construction.

Only the geometry used to estimate the shared full scale changes.

---

# V.3 Higher-order approximation and exact fallback for `C'(\theta_F)`

Let

```
\delta=\theta-1.
```

The local expansion used for fast evaluation is

```
\boxed{
C'(\theta)
\approx
C'(1)
+
C''(1)\delta
+
\frac12C'''(1)\delta^2
+
\frac16C''''(1)\delta^3.
}
```

With

```
s=D/2,
```

the fourth derivative at the NB reference point is

```
\boxed{
C''''(1)
=
3-\kappa_4(s),
}
```

where `\kappa_4(s)` is the fourth cumulant of the half-deviance under the NB reference distribution.

The implementation no longer merely loosens the Taylor tolerance when the approximation is poor.

If the audit exceeds the required tolerance, it falls back to exact adaptive support summation for that evaluation.

---

# V.4 Completed 30-split Pickrell comparison: `C'(1)` versus shared-score `C'(\theta_F)`

The corrected shared-score construction was subsequently run over the full 30 frozen Pickrell randomizations.

The average complete-null `p<.01` rates were approximately:

| GeometryOracleASH<br><br>            |         |         |
| ------------------------------------ | ------- | ------- |
| First-order `C'(1)` geometry         | 0.01110 | 0.01310 |
| Shared-score `C'(\theta_F)` geometry | 0.01102 | 0.01299 |

The two methods are therefore close in overall mean calibration.

The more informative splitwise pattern was directional:

- when the `C'(1)` oracle FPR was already well controlled, the two methods were nearly identical;
- in randomized splits with more false structure / larger excess FPR, the shared-score `\theta_F` version generally reduced that excess;
- small reversals occurred, so this is not a universal per-split dominance claim.

This is exactly the type of behavior the `\theta_F`-aware shared score was intended to capture:

> higher-order residual structure can matter most in the problematic splits without requiring the contrast increment itself to be modified.

### Geometry identity audit

With the same weights and same `\theta_F`, the corrected implementation gave the same contrast-direction scale under both constructions, as required:

```
kappa_c^{C'(1)}
=
kappa_c^{theta-shared}.
```

Representative all-gene values were approximately:

```
C'(1):

    kappa_F = 1.000000
    kappa_c = 1.279103
    lambda  = 1.279103
```

```
theta-shared:

    kappa_F = 0.998395
    kappa_c = 1.279103
    lambda  = 1.281028
```

The boundary/interior stratification told the same story:

```
kappa_c
```

was unchanged by construction; only the shared full coordinate moved slightly.

This is now an implementation invariant, not merely an empirical observation.

### Current status after later MAQC work

At the time of the completed Pickrell comparison, both `C'(1)` and shared-score `C'(\theta_F)` remained live flagship coordinates.

The later MAQC development superseded that model-selection question for the current production branch. DEB `theta` with the first-order `C'(1)` ordinary branch became the leading configuration, with exact normalized-DNB profiling used only when the first-order full residual coordinate becomes inadmissible.

The shared-score `C'(\theta_F)` construction remains valuable development evidence because it established that the shared full coordinate can be modified without changing the contrast increment. It is no longer the primary production branch.

The structural identity remains unchanged:

```text
Q_c = 2 theta_F (S_0-S_F).
```

# V.5 Conditional/local geometry experiments and closure of the operational `\lambda_{cg}` question

A series of Pickrell diagnostics asked whether the scalar contrast correction should be replaced by a gene-specific quantity.

The state variables studied included:

- null `max |logit(p_0)|`;
- null contrast information `I_{\Delta,g}^{(0)}`;
- group-information balance;
- absolute fitted contrast effect as a diagnostic;
- combinations of boundary state and information.

The main findings were:

1. `\kappa_F` and `\kappa_c` can change strongly across information/boundary strata.
2. Much of that movement is shared, so the ratio `\lambda` changes much less.
3. The conditional `\lambda` surface is nonmonotone rather than a simple boundary threshold.
4. Cross-fitted local-`\lambda` corrections did **not** flatten conditional FPR.
5. The 2-Normal local correction could become dramatically anticonservative in some low-scale regions.
6. Absolute fitted effect organized the geometry strongly but is test-dependent and therefore unsuitable as an operational pre-test conditioning variable.

The resulting conclusion is:

```
\boxed{
\text{there is currently no evidence that an operational }\lambda_{cg}
\text{ improves the flagship model.}
}
```

ASH already supplies the gene-wise calibration of null membership for the tested direction.

At the time of this local-geometry experiment, the remaining geometry was summarized as one shared `kappa_F` plus one `kappa_c` per contrast. The later DGE `kappa_F` audit simplified that statement further:

```text
C'(1):
    kappa_F = 1,
    lambda_c = kappa_c;

theta-shared:
    kappa_F^theta = simple observed shared-coordinate scale,
    lambda_c = kappa_c / kappa_F^theta.
```

The conclusion about local geometry is unchanged: a gene-specific `\lambda_{cg}` should be reopened only if future data show systematic residual null miscalibration by a prespecified gene-state variable after the contrast-level correction has been applied.

# V.6 Current DGE theory levels

The original hierarchy of DGE implementations has now changed because the MAQC admissibility audits showed exactly where the first-order normalizer approximation fails and where exact normalized fitting is actually needed.

## Level 1 — first-order `C'(1)` working branch

For genes whose first-order corrected full residual coordinate is admissible,

```text
R_F^(1)
=
S_F - C'_F(1)
>
0,
```

the fast first-order branch remains operationally effective.

With DEB precision shrinkage,

```text
theta_F^(1)
=
(a_F + alpha_F)
/
(R_F^(1) + beta_F),
```

and the first-order geometry uses

```text
kappa_F = 1,
lambda_c = kappa_c.
```

The MAQC Top-500 exact-refinement experiment supplied an important negative result: exact-normalizer refitting of hundreds of already-admissible, highly ranked genes produced essentially no change in TP/FP calls and almost no change in AUC. Therefore **high rank alone is not a reason to abandon the first-order branch**.

The first-order branch should now be understood as a computational approximation to the normalized DNB model, not as a different statistical model.

## Level 2 — targeted exact normalized-DNB precision fallback

If

```text
R_F^(1) <= 0,
```

the first-order approximation has explicitly failed. The MAQC2 audit showed that this does **not** mean the gene lacks a finite DNB precision.

For these genes, hold the fitted full-model mean and NB curvature fixed and solve the normalized-DNB precision profile directly:

```text
U_F(theta)
=
a_F / theta
-
S_F
+
sum_i C'_Fi(theta)
=
0.
```

The exact profile is

```text
ell_F(theta) - ell_F(1)
=
a_F log(theta)
-
(theta - 1) S_F
+
sum_i log C_Fi(theta).
```

The implementation brackets and solves the one-dimensional score root and checks that the root is a local profile maximum.

This is not a DEB posterior update. It is a **conditional profile-MLE fallback for `theta` under the normalized DNB distribution**, with `mu_F` and `r_g` held fixed.

On MAQC2 M03,

```text
4,109 / 4,109
```

genes with `R_F^(1)<=0` had finite exact full-model precision roots. Thus the nonpositive first-order residual coordinate was a failure of the local `C'(1)` approximation, not evidence that the DNB precision itself was undefined.

## Level 3 — genome-wide support-feasible null geometry

The next failure mode is independent of `R_F`.

The ordinary nested geometry proposes

```text
S_0^dagger
=
S_F
+
(S_0 - S_F) / lambda_c.
```

For discrete counts, however, the fitted null DNB model has a minimum attainable half-deviance

```text
S_0,min
=
sum_i min_y D_NB(y; mu_0i, r_g) / 2.
```

Because the closest attainable count is an integer, `S_0,min` can be strictly positive.

The MAQC audits found genes for which

```text
S_0^dagger < S_0,min.
```

For such genes, the geometry asks the null distribution to attain an impossible residual statistic. Exact null profiling then drives the score toward an upper-precision boundary rather than producing a finite interior root.

The support-feasible correction is therefore

```text
B_g
=
max(S_Fg, S_0,min,g),
```

```text
S_0,g^(dagger,feasible)
=
B_g
+
max(S_0g - B_g, 0) / lambda_c.
```

This preserves the irreducible discrete-support contribution and shrinks only the residual increment that actually lies above the support floor.

The support-floor check must be performed **genome-wide**, because the pathology can be silent even when `R_F^(1)>0`.

## Level 4 — fully normalized joint DNB fitting

Because `C(mu,r,theta)` depends on `mu` whenever `theta != 1`, a completely joint normalized-DNB fit would couple

```text
beta
<-> r
<-> theta
<-> C
<-> geometry.
```

The current method does not do this. It fits `mu` and `r` first and then conditions on them while fitting/refining `theta`.

Fully joint fitting remains a possible deeper model, but it is not required to explain or repair the current MAQC admissibility failures.

## Current operational interpretation

The current production method is therefore adaptive:

```text
ordinary admissible gene:
    fast C'(1) + DEB precision

R_F^(1) <= 0:
    exact normalized-DNB full precision fallback

all genes:
    check null support feasibility

S_0^dagger < S_0,min:
    use support-feasible null geometry
```

This architecture arose from distinct mathematical failure modes and should not be described as one generic "exact fallback."

# W. Current unified flagship architecture for DGE

The DGE architecture is now substantially more specific than the earlier 12-branch development grid.

The current leading production branch is the former **M03** configuration:

```text
r_g:
    WEB-DP

theta_g:
    Gamma-Gamma DEB on the ordinary first-order branch

shared residual coordinate:
    first-order C'(1)

full-scale geometry:
    kappa_F = 1

contrast geometry:
    ASH-weighted kappa_c

lambda_c:
    lambda_c = kappa_c
```

Two independent admissibility corrections are now built into the procedure:

```text
1. exact normalized-DNB theta fallback when R_F^(1) <= 0;
2. genome-wide support-feasible null geometry when S_0^dagger < S_0,min.
```

These are not benchmark-label adjustments. They repair mathematical failures of the approximation/geometry itself.

## Distributional layer

For gene `g`, the working DNB variance remains

```text
Var(Y_gi)
approx
[mu_gi + mu_gi^2 / r_g] / theta_g.
```

The parameters retain distinct interpretations:

```text
r_g
=
NB mean-variance curvature;

theta_g
=
multiplicative residual precision left after the NB mean/curvature fit;

kappa_c
=
contrast-direction geometry after the shared residual scale has been calibrated.
```

The normalized DNB density is

```text
f_theta(y; mu, r)
=
C(mu,r,theta)
sqrt(theta)
f_NB(y;mu,r)^theta
f_sat(y;r)^(1-theta).
```

The exact normalizer satisfies

```text
log C(mu,r,theta)
=
-0.5 log(theta)
-
log sum_y f_sat(y;r) exp{-theta D_NB(y;mu,r)/2}.
```

The exact precision score uses

```text
C'(theta)
=
d log C / d theta.
```

`C'(1)` is therefore a computational shortcut around the ordinary-NB reference point, not a separate probability model.

## W.1 Fixed full model versus tested contrast

Let the full design matrix be fixed at `X`. In the quadratic idealization,

```text
P_F = projector onto col(X),
M_F = I - P_F,
Q_F = Y' M_F Y.
```

A one-degree-of-freedom contrast is

```text
H0: c' beta = 0.
```

The constrained model is a subspace of the same full model, so

```text
Q_0,c = Q_F + Q_c.
```

Changing `c` while holding `X` fixed does not change the full residual space. It changes only the constrained/tested direction.

A weighted contrast such as

```text
(2 beta_1 + 3 beta_2)/5 - (5 beta_3 + 20 beta_4)/25
```

is simply a contrast vector through the same fixed design.

This remains fundamentally different from raw-label permutation, which changes `X`, `P_F`, and `M_F` and therefore targets a different full-model geometry.

## W.2 Why alternatives contaminate `Q_c` but not inherently `Q_F`

If

```text
Y_g = X beta_g + epsilon_g,
```

and the full model contains the true tested effect, then

```text
M_F X beta_g = 0.
```

The alternative mean signal is absorbed before the full residual coordinate is formed.

By contrast, the constrained model deliberately removes the tested direction, so the added increment

```text
Q_c = Q_0,c - Q_F
```

is noncentral under a true alternative.

This is why ASH/null weighting belongs naturally in `kappa_c`, but not in the DGE `kappa_F` denominator.

## W.3 `theta_F` already calibrates the first-order shared residual scale

The NB layer supplies

```text
V_NB(mu)
=
mu + mu^2/r.
```

The DNB layer rescales that remaining variation through `theta_F`.

In the first-order coordinate,

```text
R_F^(1)
=
S_F - C'_F(1),
```

```text
Q_F^(1)
=
2 theta_F R_F^(1).
```

The precision estimate itself is learned from this shared residual coordinate, so introducing a second population-level full-scale correction after `theta_F` is redundant.

The Baik audit showed:

```text
simple observed kappa_F
~ 1,
NB-bootstrap kappa_F
~ 1,
residual-resampling kappa_F
~ 1
```

in the first-order coordinate.

## W.4 DEB gives an exact first-order `kappa_F=1` scale identity

For DEB, let

```text
a = d_F/2,
theta_g = (a + alpha) / (R_g + beta).
```

The marginal-likelihood score equation for the Gamma prior rate implies

```text
sum_g theta_g R_g
=
a G.
```

Because

```text
Q_Fg = 2 theta_g R_g
```

and `d_F=2a`,

```text
\boxed{
sum_g Q_Fg / (d_F G) = 1.
}
```

Thus, for DEB in the same coordinate used to fit `theta`, first-order

```text
kappa_F = 1
```

is a scale-calibration identity rather than an empirical coincidence.

Therefore the primary DGE geometry is

```text
lambda_c = kappa_c.
```

## W.5 The older theta-shared coordinate remains useful as a development diagnostic, not the current primary branch

The theta-shared coordinate used

```text
Q_F^theta
=
2 theta_F [S_F - C'_F(theta_F)]
```

while preserving

```text
Q_c
=
2 theta_F(S_0-S_F).
```

That construction remains mathematically coherent and was useful in the Pickrell development work.

However, the MAQC experiments now favor the simpler M03 production architecture with DEB `theta`, first-order `C'(1)` for ordinary admissible genes, and exact normalized fitting only where the first-order approximation explicitly fails.

The theta-shared branch should therefore be retained as a historical/development comparator rather than treated as the current production default.

## W.6 `G_usable` and the MAQC coverage problem led directly to the exact-normalizer fallback

On Baik,

```text
G_usable = G_total = 17,961
```

for all methods.

MAQC2 was different. Under the ordinary first-order M03 branch,

```text
G_total   = 38,833
G_usable  ~ 34,724
```

and the exclusions were overwhelmingly genes with

```text
R_F^(1) <= 0.
```

The qPCR truth audit showed that this did not manufacture the low false-positive rate:

```text
qPCR-positive usability ~ 92.7%
qPCR-negative usability ~ 93.5%
```

for the ordinary M03 branch, with essentially no truth-status preference.

Rescoring only usable qPCR truth genes left qPCR-reference FDR unchanged.

Thus the coverage failure was real but did not explain the favorable qPCR error rate.

The exact-normalizer audit then showed:

```text
4,109 / 4,109
```

MAQC2 M03 genes with `R_F^(1)<=0` had finite exact full-model `theta_F` roots.

Therefore

```text
R_F^(1) <= 0
```

was reinterpreted as a failure of the local `C'(1)` approximation rather than a genuinely undefined DNB precision.

## W.7 Exact `C(theta)` fallback: what it actually does

For an `R_F^(1)<=0` gene, the current method holds

```text
mu_F,
mu_0,
r_g
```

fixed and re-estimates `theta` by the exact normalized-DNB precision profile.

For the full model it solves

```text
U_F(theta)
=
a_F/theta
-
S_F
+
sum_i C'_Fi(theta)
=
0.
```

The corresponding profile objective relative to `theta=1` is

```text
ell_F(theta)-ell_F(1)
=
a_F log(theta)
-
(theta-1)S_F
+
sum_i log C_Fi(theta).
```

A bracketed root solver is used, and the resulting root is checked as a local profile maximum.

The null side is solved analogously using the support-feasible corrected null statistic described below.

If both exact roots exist, the exact-profile raw statistic is

```text
T_raw,exact
=
2[theta_0 S_0^(dagger,feasible) - theta_F S_F]
+
n log(theta_F/theta_0)
+
2[log C_F(theta_F) - log C_0(theta_0)].
```

The current implementation then uses

```text
T_exact = max(T_raw,exact, 0),
p_exact = P(chi^2_1 >= T_exact).
```

Important interpretation:

```text
ordinary branch:
    DEB shrinkage estimator for theta

exact fallback:
    conditional normalized-DNB profile MLE for theta
    with mu and r fixed
```

The exact fallback is therefore not simply "DEB with a different constant."

## W.8 Double-Poisson fallback was informative but is no longer the primary rescue

Because many problematic genes had very large `r_g`, a DNB-to-double-Poisson fallback was tested.

Two trigger rules were compared:

```text
A. switch to DP when r_g hits the Poisson boundary;
B. switch to DP when R_F^(1) <= 0.
```

The `r_g`-boundary trigger was too broad. On MAQC2 M03,

```text
n(r_g at boundary) = 16,670
n(R_F<=0)          = 4,109
n(both)            = 3,830.
```

Thus most `R_F<=0` genes were near the Poisson curvature limit, but most Poisson-boundary genes had perfectly admissible first-order full scores.

The `R_F<=0` DP fallback improved MAQC2 M03 from roughly

```text
352 TP / 22 FP
```

to

```text
363 TP / 22 FP,
```

but rescued only a small fraction of the unusable genes.

A subsequent exact-DNB audit showed that the deeper problem was the first-order normalizer approximation, not the count-family choice.

The DP fallback is therefore retained as development history, not the production rescue.

## W.9 Exact full-model roots and the remaining null-boundary problem

Among the 4,109 MAQC2 M03 genes with `R_F^(1)<=0`:

```text
4,109 / 4,109
```

had finite exact full-model precision roots.

Before the support-feasibility correction, a large majority also obtained finite exact null roots, but a residual canonical set of

```text
573
```

genes remained with

```text
null_root_status = root_above_search_max.
```

For these genes the exact null score stayed positive throughout the upper search range.

A count-discrepancy audit established that **573** is the correct canonical unresolved `R_F<=0` count. Larger intermediate counts such as 645/625/620 occurred because a diagnostic sample of ordinary `R_F>0` reference genes had accidentally been included in the broad null-boundary cohort.

## W.10 The discrete support floor explains the null-root degeneracy

For the null mean of gene `g`, define

```text
S_0,min,g
=
sum_i min_y D_NB(y; mu_0gi, r_g)/2.
```

Because counts are discrete, the fitted null mean need not itself be attainable by an integer observation. Even infinite precision cannot drive the residual half-deviance below this floor.

The ordinary geometry proposes

```text
S_0^dagger
=
S_F
+
(S_0-S_F)/lambda_c.
```

For the degenerate-null genes, the audit showed that this could produce

```text
S_0^dagger < S_0,min.
```

The null score then has no finite root because the geometry is asking the normalized DNB model to attain an impossible residual statistic.

This is a geometry/support problem, not a normalizer-approximation problem.

## W.11 Support-feasible null geometry

The corrected construction separates the irreducible support cost from the contrast increment that is allowed to shrink.

Define

```text
B_g
=
max(S_Fg, S_0,min,g).
```

Then

```text
\boxed{
S_0,g^(dagger,feasible)
=
B_g
+
max(S_0g-B_g,0)/lambda_c.
}
```

Interpretation:

```text
S_0,min - S_F
=
residual cost required by discrete support;
do not shrink it.

S_0 - max(S_F,S_0,min)
=
remaining contrast-related excess;
shrink this part by lambda_c.
```

If the old geometry is already support-feasible, nothing changes.

The direct support-floor check must be applied **for every gene**, not only genes with `R_F<=0`.

## W.12 Genome-wide support-floor audit

The genome-wide audit found that support infeasibility was substantially broader than the original 573-gene exact-null boundary cohort.

For M03 on MAQC2:

```text
G analyzed                         = 38,833
support-floor violations          = 2,941  (~7.58%)
R_F > 0 support violations        = 2,206
R_F <= 0 support violations       =   735
```

Thus roughly

```text
6.35%
```

of the apparently ordinary `R_F>0` genes had a **silent support-infeasible null geometry**.

For M03 on MAQC3:

```text
G analyzed                         = 41,767
support-floor violations          = 835  (~2.00%)
R_F > 0 support violations        = 835
R_F <= 0 support violations       = 0.
```

M03 and M07 identified identical violation sets on MAQC2 and MAQC3. M11 was also identical on MAQC2 and identified a slightly smaller but highly overlapping set on MAQC3.

Most importantly,

```text
n_after_feasible_still_below_floor = 0
```

for every audited dataset/method.

The feasible-support formula therefore repaired every algebraically identified null-support violation.

## W.13 Final M03 fitting procedure

The current M03 production procedure is:

```text
1. Estimate r_g with WEB-DP.

2. Fit the full and constrained NB means under the fixed observed design.

3. Evaluate C'_F(1) and C'_0(1).

4. Form
       R_F^(1) = S_F - C'_F(1).

5. Fit the full DEB precision on the ordinary admissible R_F^(1)>0 population.

6. Using the ordinary-admissible full estimates, fit ASH to the contrast effect/se
   distribution and estimate kappa_c.

7. Set
       kappa_F = 1,
       lambda_c = kappa_c.

8. Construct the ordinary nested null statistic
       S_0^dagger = S_F + (S_0-S_F)/lambda_c.

9. For EVERY gene, compute the null support floor S_0,min.

10. If S_0^dagger < S_0,min, replace the null geometry by
       B = max(S_F,S_0,min),
       S_0^(dagger,feasible)
           = B + max(S_0-B,0)/lambda_c.

11. Fit the first-order null DEB precision on the genome-wide
    support-feasible null statistic.

12. For genes with R_F^(1)>0:
       retain the first-order DEB full/null test.

13. For genes with R_F^(1)<=0:
       abandon the first-order full precision;
       solve the exact normalized-DNB full theta profile;
       solve the exact normalized-DNB null theta profile using
       the already support-feasible null statistic;
       evaluate log C at the fitted theta values;
       calculate the exact-profile test statistic.

14. The current implementation clips negative raw statistics at zero:
       T = max(T_raw,0),
       p = chi-square_1 survival(T).

15. Recombine the ordinary and exact-fallback gene-level p-values.

16. Apply BH over the full analyzed finite-p universe.
```

This procedure deliberately separates:

```text
normalizer approximation failure
from
support-feasibility failure.
```

`R_F<=0` triggers exact precision fitting.

`S_0^dagger<S_0,min` triggers support-feasible geometry.

Neither trigger depends on benchmark truth labels.

## W.14 Why the procedure has this form

The final architecture is not an accumulation of arbitrary patches.

Each branch was forced by a distinct audit:

**Why keep the cheap first-order branch?**

The Top-500 exact-refinement experiment exact-evaluated the final Top 500 genes on both MAQC datasets. After closure:

```text
500 / 500
```

top-ranked genes had been evaluated by the exact-normalizer machinery.

Yet MAQC2 TP/FP and AUC were unchanged, and MAQC3 TP/FP were unchanged with only a negligible AUC shift.

Therefore ordinary high-ranking `R_F>0` genes do not generally benefit from expensive exact profiling.

**Why exact-fit `R_F<=0` genes?**

Because all 4,109 MAQC2 M03 first-order-invalid genes had finite exact full roots, proving that the local approximation—not the DNB model—was failing.

**Why check the support floor genome-wide?**

Because thousands of `R_F>0` genes had silent support-infeasible null geometries.

**Why not shrink below the support floor?**

Because no normalized discrete DNB distribution, even at infinite precision, can attain a residual half-deviance below its nearest-support-point minimum.

## W.15 Final admissibility result

After integrating both corrections into the full M03 rerun:

```text
MAQC2:
    G_total    = 38,833
    G_finite_p = 38,833

MAQC3:
    G_total    = 41,767
    G_finite_p = 41,767
```

Thus all analyzed genes received finite p-values under the finalized fitting rules on both MAQC datasets.

This closes the **testability/admissibility** problem observed in the earlier first-order implementation.

It does not automatically close the separate negative-`T_raw` question.

## W.16 Negative test statistics remain a distinct unresolved issue

The earlier DEB-theta audits showed that a nontrivial fraction of genes can have

```text
T_raw < 0
```

even when both full and null precision fits exist.

Exact normalizer fitting reduces one source of pathology but does not guarantee nonnegative geometry-adjusted test statistics because the geometry-adjusted null objective is not simply the ordinary nested likelihood.

The current production implementation uses

```text
T = max(T_raw,0),
```

so negative raw statistics receive `p=1`.

This issue is **not an admissibility failure**. It is a separate question about the statistical reference/test construction and remains open.

## W.17 Retired `kappa_F` estimators

The following remain retired for DGE production use:

- raw-label permutation `kappa_F`;
- ordinary-NB bootstrap `kappa_F`;
- effect-removed gene-wise residual-resampling `kappa_F`;
- pooling bootstrap and residual-resampling `kappa_F`;
- ASH-weighted `kappa_F`;
- a separate first-order `kappa_F` correction after `theta_F`.

Their diagnostic roles remain as described earlier.

## W.18 Current method status

The former 12-method grid remains valuable development history, but current evidence strongly favors:

```text
WEB-DP r_g
+
DEB theta
+
first-order C'(1) ordinary branch
+
exact normalized-DNB theta fallback for R_F<=0
+
genome-wide support-feasible null geometry
+
ASH kappa_c
+
kappa_F=1.
```

WEB-theta remains a comparator, and the matched-DEB-DP `r_g` estimator remains an important robustness result because it repeatedly tracked WEB-DP closely.

The flexible `r_g` branch has not shown a consistent operating-point advantage and is no longer the leading configuration.

# What we have learned about the model

The accumulated development work now supports several broader conclusions.

## 1. Residual precision and contrast geometry are distinct objects

The repeated failure of attempts to absorb `kappa_c` into `theta_g` established that they play different inferential roles.

`theta_g` describes event/gene-level residual dispersion/precision.

`kappa_c` describes how the particular tested direction is represented after the shared full residual variation has been accounted for.

The geometry layer is therefore not merely another dispersion estimator.

## 2. The full residual space is genuinely shared across contrasts from one fixed full model

The Tasic fixed-full-model experiment gave exact invariance of

```text
S_F,
theta_F,
nu_F,
Q_F
```

across the tested contrasts.

The linear-algebra reason is even more basic: with fixed `X`, the full residual maker `M_F=I-P_X` does not depend on the contrast vector `c`. Different contrasts alter only the constrained subspace and its added direction.

This is why raw-label permutation is not a substitute for varying contrasts: permutation changes `X`; a new contrast through fixed `X` does not.

## 3. Most meaningful contrast variation belongs in `kappa_c`

The earlier Tasic family experiments already suggested that `kappa_F` was extremely stable while `kappa_c` moved by contrast.

The later DGE work sharpened this result. In the first-order DGE coordinate, the full precision fit itself calibrates the shared scale, so `kappa_F=1`; in the theta-shared coordinate, the remaining full-scale adjustment is only the simple observed coordinate change. The nontrivial geometry still resides in the tested direction.

Thus the conceptual decomposition is increasingly:

```text
shared full residual calibration through theta_F
+
contrast-specific directional geometry through kappa_c.
```

## 4. ASH is doing the gene-wise null-membership job where null membership is actually needed

ASH is not estimating gene-specific geometry. Its role is to answer:

```text
how strongly should gene/event g contribute to the null geometry of contrast c?
```

That role is clearly appropriate for `kappa_c`, because true alternatives inflate the constrained contrast increment.

For DGE, it is **not** appropriate to use those contrast-derived weights to redefine the shared full-model scale. The Baik audit therefore retires ASH-weighted `kappa_F` for DGE while retaining ASH for `kappa_c`.

For the DEU/DTU family formulation, the earlier pooled-null-information construction remains the current branch until it is separately re-audited under the same precision-scale identity.

## 5. The former ASH-weight boundary problem is no longer central to DGE `kappa_F`

The earlier family estimator used ASH weights in a ratio, so common multiplicative depression of all weights canceled. That remains relevant to the DEU/DTU family formulation.

For DGE, the stronger result is now available: first-order `kappa_F` does not need ASH at all, and theta-shared `kappa_F^theta` is estimated directly from the observed shared coordinate.

This removes the pathological possibility that an almost-zero ASH null mass can determine the shared DGE full-scale denominator.

## 6. `pi_0c` remains useful for understanding contrast composition, but it is not a second DGE full-scale weight

In the earlier family experiment,

```text
mean_g(w_gc) approx pi_0c,
```

so multiplying by `pi_0c` again roughly double-weighted contrast purity and was unnecessary.

That conclusion remains. In current DGE, neither `pi_0c` nor ASH weights enter the simple shared `kappa_F` estimator.

## 7. The normalizer correction and contrast geometry solve different problems

The DGE work repeatedly supports the separation

```text
normalizer correction
->
residual-law / precision coordinate
```

versus

```text
kappa_c correction
->
added contrast-direction geometry.
```

The failed two-score `C'(theta_F)` experiment was especially important because it showed what happens when those roles are mixed.

## 8. The contrast increment must remain structurally intact

The corrected shared-score identity

```text
Q_c
=
2 theta_F (S_0-S_F)
```

survived the `C'(1)` versus `C'(theta_F)` comparison exactly.

This is now one of the strongest structural constraints in the model:

```text
normalizer refinements may change the shared residual coordinate,
but not the identity of the tested contrast increment.
```

## 9. Correcting the scale externally is not equivalent to correcting the likelihood geometry

Pickrell made this explicit.

An externally estimated scale—even an oracle one—did little when applied only after the statistic had already been constructed.

Applying the conceptual correction to the nested residual increment substantially repaired calibration.

Therefore:

```text
where the correction enters the model is part of the model.
```

## 10. A scalar contrast geometry is currently sufficient operationally

Local state variables such as boundary position and information clearly affect absolute geometry.

But they mostly moved shared and directional scales together, and local `lambda` corrections did not improve conditional calibration.

Thus the current evidence favors

```text
lambda_c
```

over

```text
lambda_cg.
```

That is an empirical operational conclusion, not a theorem that gene-specific geometry can never exist.

## 11. The NB bootstrap, residual resampling, and raw-label permutation answer three different questions

They should no longer be pooled or described as interchangeable ways to estimate one `kappa_F`.

```text
ordinary-NB bootstrap
=
calibration under the fitted nominal model;

gene-wise residual resampling
=
semi-empirical robustness after destroying much sample-wide structure;

raw-label permutation
=
stress test over altered full-model designs.
```

Each can still be useful, but none is the current production estimator of DGE `kappa_F`.

## 12. `G_usable` is eligibility bookkeeping, not an effective sample size

The Baik audit showed complete usability for all 17,961 genes across all 12 methods. No `n_eff`-style correction is warranted.

The artificial trimming experiment nevertheless demonstrated that selection on the shared residual score can bias the simple average upward. Future datasets must therefore report and audit `G_usable` rather than silently conditioning on it.

## 13. `R_F^(1)<=0` is a failure of the local normalizer approximation, not proof of an undefined precision

The MAQC2 exact-normalizer audit showed that all

```text
4,109 / 4,109
```

M03 genes with nonpositive first-order full residual coordinates had finite exact normalized-DNB full-model precision roots.

Therefore the state

```text
S_F - C'_F(1) <= 0
```

should be interpreted as evidence that the `theta=1` local approximation is inadequate for that gene.

It should not be interpreted as evidence that the DNB model itself has no finite precision solution.

## 14. Exact `C(theta)` fitting is targeted profile likelihood, not a second shrinkage model

For an exact-fallback gene, the fitted mean and `r_g` are held fixed and the normalized DNB likelihood is profiled only over `theta`.

The procedure repeatedly evaluates

```text
C'(theta)
```

and

```text
log C(theta)
```

until the one-dimensional precision score is solved.

Thus the exact fallback is statistically simple:

```text
profile MLE of theta under the normalized DNB family,
conditional on the already fitted mean and r_g.
```

The novelty is not the optimizer itself. The substantive architecture lies in the DNB parameterization, the separation of `r_g` and `theta_g`, and the contrast-geometry correction.

## 15. Exact normalizer fitting does not need to be used indiscriminately

The final Top-500 refinement experiment exact-evaluated every gene in the genome-wide Top 500 on both MAQC2 and MAQC3.

Refitting already-admissible top-ranked genes with exact `C(theta)` produced no material change in TP/FP calls and essentially no improvement in AUC.

Therefore the current evidence favors **mathematical-failure-triggered exact fitting**, not rank-triggered exact fitting.

The practical trigger is:

```text
R_F^(1) <= 0
```

rather than

```text
gene is highly ranked.
```

## 16. Null support feasibility is independent of full-score admissibility

A gene can have

```text
R_F^(1) > 0
```

and therefore receive an apparently ordinary first-order full precision, while the geometry-corrected null residual statistic still violates the support of the discrete null DNB model.

The genome-wide MAQC audits found thousands of such silent cases.

Thus:

```text
full-model admissibility
!=
null support feasibility.
```

The support-floor check must be performed for every gene.

## 17. The discrete support floor is a distinct geometric object

For a fitted null mean `mu_0`, the nearest attainable integer count imposes a minimum half-deviance.

This irreducible residual contribution is not contrast inflation and must not be divided by `lambda_c`.

The correct geometry therefore separates

```text
support-imposed residual cost
```

from

```text
shrinkable contrast excess.
```

This is conceptually parallel to the earlier lesson that shared residual information and contrast-direction information must not be mixed.

## 18. The final M03 method is fully testable genome-wide on MAQC2 and MAQC3

After adding the exact full fallback and genome-wide support-feasible null geometry, the final M03 rerun produced

```text
38,833 / 38,833 finite p-values on MAQC2
41,767 / 41,767 finite p-values on MAQC3.
```

The earlier `G_usable` coverage failure is therefore closed for these datasets under the finalized rules.

## 19. DEB theta is now the leading precision branch

Across Baik and MAQC, the second-stage DEB precision estimator repeatedly produced a much stronger error/power tradeoff than WEB theta.

On MAQC2, the final M03 architecture retained very high sensitivity while sharply reducing qPCR-reference false discoveries.

On MAQC3, it achieved strong calibration without requiring the `R_F<=0` exact fallback at all.

WEB theta remains valuable as a comparator, but DEB theta is now the primary production choice.

## 20. MAQC demonstrates strong controlled-reference performance, not yet general biological-population dominance

MAQC compares biologically distinct reference RNA materials but uses highly standardized reference-material replication rather than ordinary donor-to-donor biological replication.

The contrast is biologically meaningful; much of the replication noise is technical.

Therefore the MAQC results are a strong test of count-modeling, technical-noise control, ranking, and external qPCR agreement, but they do not by themselves establish superiority under realistic population heterogeneity.

The next important external tests must include genuinely heterogeneous biological samples, real-data nulls, and historical competitor-owned benchmarks.

## 21. All iterative computational workflows are now required to be resumable and timestamped

Project implementation policy now requires:

```text
durable checkpointing during execution,
resume/restart logic,
skipping completed work,
preservation of successful partial results,
retry controls for failed items,
atomic writes where practical,
local start time,
local end time.
```

This requirement applies even to apparently trivial repeated computations.

# X. DGE validation entered a new phase; Baik served both validation and development roles

The procedural rule remains:

```text
freeze a statistical method before a genuinely new benchmark outcome is observed;
fix genuine implementation/mathematical errors when discovered;
do not tune parameters to benchmark truth labels merely because an outcome is disappointing.
```

That rule was followed for the first Baik run: the then-current Pickrell-developed WEB-DNB architecture was carried into Baik without benchmark-specific tuning.

Baik was later used explicitly as development data for:

- the three-way `r_g` comparison;
- WEB versus DEB `theta`;
- the first-order versus theta-shared coordinate comparison;
- the `kappa_F` audit;
- numerical stabilization of the theta-aware normalizer machinery.

The 12-method grid therefore remains important development history.

However, later MAQC mathematical audits have now narrowed the current production candidate substantially.

The leading DGE method is the finalized M03 architecture:

```text
WEB-DP r_g
+
DEB theta
+
kappa_F = 1
+
ASH kappa_c
+
first-order C'(1) ordinary branch
+
exact normalized-DNB theta fallback for R_F<=0
+
genome-wide support-feasible null geometry.
```

The exact fallback and support-floor correction were triggered by internal mathematical failures (`R_F<=0` and support-infeasible null residual geometry), not by qPCR truth labels.

Accordingly, the next genuinely independent benchmark should freeze this **finalized M03 production procedure** in advance.

WEB-theta, matched-DEB-DP `r_g`, flexible `r_g`, and theta-shared formulations remain useful diagnostic/ablation branches, but the project no longer needs to carry the entire original 12-method grid as co-equal production candidates into every new benchmark.

# Y. First external real-data benchmark: Baik / GSE49712, followed by a Baik development grid

The Baik/GSE49712 reconstruction is exact:

```text
17,961 retained genes
63 ERCC controls
48 true alternatives
15 true nulls.
```

This benchmark supplied two distinct stages of evidence and they should not be conflated.

## Y.1 First frozen external Baik run

The authoritative ranking results from the first frozen comparison were approximately:

| Method | ERCC AUC |
| --- | ---: |
| historical-logic edgeR exact | **0.877778** |
| historical-logic edgeR QL | **0.877778** |
| current edgeR QL | **0.868056** |
| frozen WEB-DNB `C'(1)` + ASH | **0.856944** |
| frozen WEB-DNB theta-shared + ASH | **0.856944** |
| historical-logic DESeq2 | 0.818056 |
| current DESeq2 | 0.803472 |

The earlier `~0.8806` current-edgeR number is superseded by the frozen combined output above.

At adjusted `p/q < .10`, the correct current edgeR QL ERCC counts are:

```text
48 true positives
15 false positives.
```

All 15 null ERCCs were called. The corresponding empirical ERCC false-discovery proportion is

```text
15 / (48 + 15) = 23.81%.
```

The earlier internal statement that edgeR had 13 false positives is retired; the correct count is 15.

The first frozen WEB implementation was more conservative at the operating point but ranked slightly below edgeR. That run established that a very young WEB-DNB DGE implementation was already competitive enough to justify deeper development.

## Y.2 Twelve-method Baik development experiment

Baik was then used explicitly as **development data** to compare three `r_g` estimators, two `theta` shrinkage methods, and the two shared-residual coordinates.

The null-resampling experiment completed:

```text
480 / 480 requested null fits successfully.
```

The first observed-data pass scored 9 of the 12 methods. The three missing observed methods were not statistical evidence against those methods:

```text
M08:
    local Taylor C'(theta_F) audit failure;

M10 and M12:
    Windows checkpoint PermissionError.
```

The numerical lesson was to preserve the frozen `0.001` audit and fall back from a failing local Taylor approximation to direct/audited `C'(theta_F)` evaluation rather than aborting. The checkpoint failures were I/O problems rather than fit failures.

Among the methods with observed Baik scores, the important `q<.10` operating points were:

| Family | TP | FP | empirical ERCC FDP | sensitivity |
| --- | ---: | ---: | ---: | ---: |
| WEB-DP / matched `r_g` + WEB `theta` | 46 | 11 | 19.30% | 95.83% |
| WEB-DP / matched `r_g` + DEB `theta` | **44** | **7** | **13.73%** | **91.67%** |
| flexible `r_g` + WEB `theta` | 46 | 14 | 23.33% | 95.83% |
| flexible `r_g` + DEB `theta` (M11) | 43 | 9 | 17.31% | 89.58% |

The visible AUCs were approximately:

```text
M11 flexible-r + DEB theta + C'(1)        0.862500
M03/M04/M07 ordinary/matched + DEB theta  0.861111
M01/M02/M05/M06 ordinary/matched + WEB    0.856944
M09 flexible-r + WEB                      ~0.8561
current edgeR QL                          0.868056.
```

The main scientific pattern was not the tiny AUC ordering. It was the threshold behavior of the second precision layer.

Holding the ordinary/matched `r_g` family essentially fixed,

```text
WEB theta:
    46 TP / 11 FP

DEB theta:
    44 TP / 7 FP.
```

So DEB `theta` removed four false calls at the cost of two true calls relative to WEB `theta`.

Relative to current edgeR QL,

```text
edgeR:
    48 TP / 15 FP

ordinary/matched + DEB theta:
    44 TP / 7 FP.
```

That is an 8-of-15, or approximately 53.3%, reduction in false-positive ERCC calls at the cost of 4-of-48, or 8.3 percentage points, of sensitivity.

This is a striking operating-point tradeoff, but it is **not** perfect FDR control: `7/(44+7)=13.73%`, still above the nominal 10% threshold.

Two other Baik development conclusions were unusually clean:

```text
matched trend-DEB-DP r_g
~
WEB-DP r_g
```

for practical inference, while

```text
C'(1)
~
theta-shared C'(theta_F)
```

produced essentially identical calls in the successful paired comparisons.

The flexible first-stage `r_g` estimator changed rankings slightly but produced worse threshold calls than the simpler ordinary/matched branch.

The Baik development results remain important evidence for why DEB `theta`, WEB-DP/matched curvature estimation, and the first-order coordinate were retained. They no longer define the current production grid: later MAQC admissibility work narrowed the leading method to finalized M03. Baik must not be reused as untouched external validation for that selection.

# Z. MAQC: truth provenance, mathematical audits, and the finalized M03 result

MAQC2 and MAQC3 became the most informative DGE datasets in the project because they simultaneously exposed:

- the practical importance of DEB `theta`;
- the limits of the first-order `C'(1)` normalizer approximation;
- the existence of silent support-infeasible null geometries;
- the distinction between ranking, threshold calibration, and matched-error power;
- the need to separate reference-material benchmarking from genuine biological-population heterogeneity.

The current reconstructed qPCR truth universes remain:

```text
MAQC2:
399 positive
107 negative

MAQC3:
403 positive
119 negative.
```

These do **not** match the historical benchmark universe usually described as approximately

```text
388 positive
143 negative.
```

That historical mapping discrepancy remains unresolved and must be stated whenever the current qPCR universe is used.

## Z.1 The historical reproduction discrepancy points to identifier mapping, not competitor failure

The historical-logic competitor audit showed that DESeq2, edgeR exact, and edgeR QL all shifted in nearly the same direction relative to the historical reported benchmark counts.

That strongly suggests the principal mismatch lies in the truth-scoring universe:

```text
qPCR measurements
-> qPCR positive/negative rules
-> historical gene identifiers
-> historical annotation/mapping
-> RNA-seq analysis universe.
```

Once the historical mapping is recovered, the already-saved genome-wide method outputs can be rescored without rerunning the statistical fits.

## Z.2 MAQC “actual FDR” terminology remains retired

The quantity

```text
FP / (TP + FP)
```

against the qPCR-negative reference subset should be called

```text
qPCR-reference FDR
```

or

```text
qPCR disagreement rate among scored discoveries.
```

It is not literal frequentist FDR over all endogenous genes because:

- truth is available only for a selected subset;
- intermediate qPCR genes are omitted;
- qPCR itself is an external assay rather than omniscient biological truth;
- mapping/annotation determines the scored universe.

The benchmark is nevertheless extremely valuable because it supplies external, orthogonal signal labels.

## Z.3 Initial 12-method MAQC experiment

The first explicit 12-method MAQC experiment compared:

```text
3 r_g estimators
x 2 theta estimators
x 2 shared coordinates.
```

The clearest early pattern was that the second-stage precision estimator mattered substantially more than the first-stage matched WEB-DP versus matched-DEB-DP choice.

For MAQC2, the ordinary WEB-DP branch gave approximately:

```text
WEB theta / C'(1):
    370 TP / 96 FP

DEB theta / C'(1):
    352 TP / 22 FP.
```

The theta-shared DEB branch gave approximately:

```text
351 TP / 18 FP.
```

For MAQC3:

```text
WEB theta / C'(1):
    376 TP / 24 FP

DEB theta / C'(1):
    375 TP / 16 FP.
```

Thus DEB `theta` dramatically reduced false qPCR-reference calls while preserving most sensitivity.

The matched trend-DEB-DP `r_g` branch again closely reproduced WEB-DP. The flexible `r_g` branch did not provide a consistently better threshold tradeoff.

This was the first strong evidence that DEB `theta` should become the leading precision branch.

## Z.4 MAQC2 `G_usable` audit: the favorable FDR was not caused by dropping qPCR negatives

The first-order M03 branch on MAQC2 had only about

```text
34,724 / 38,833
```

usable genes because roughly 4,109 genes had nonpositive corrected full residual scores.

The qPCR truth audit showed:

```text
positive qPCR genes usable: 370 / 399  ~ 92.7%
negative qPCR genes usable: 100 / 107 ~ 93.5%.
```

There was no evidence that qPCR negatives were preferentially made untestable.

When the qPCR metrics were recomputed using only truth genes that the method could actually test, the qPCR-reference FDR did not improve artificially; it remained the same because the TP and FP calls among testable genes were unchanged.

The original scoring actually penalized sensitivity by counting untestable positives as false negatives while slightly flattering specificity by counting untestable negatives as true negatives.

Therefore the low MAQC2 false-positive rate was **not** a consequence of selective exclusion.

## Z.5 The `R_F<=0` population and the double-Poisson detour

The dominant MAQC2 exclusion reason was

```text
R_F^(1)
=
S_F - C'_F(1)
<= 0.
```

Because many of these genes had extremely large fitted `r_g`, the project tested whether the DNB model should fall back to its double-Poisson limit.

Two trigger rules were compared.

### Poisson-curvature boundary trigger

For M03:

```text
r_g at upper/Poisson boundary: 16,670 genes
R_F<=0:                       4,109 genes
both:                         3,830 genes.
```

Thus

```text
P(r boundary | R_F<=0) ~ 0.932
```

but

```text
P(R_F<=0 | r boundary) ~ 0.230.
```

The Poisson-boundary trigger was therefore far too broad. MAQC3 strengthened this conclusion: thousands of genes could hit the `r_g` boundary while **zero** genes had `R_F<=0`.

The broad DP trigger did not improve inference and could yield very large negative-statistic fractions.

### `R_F<=0` DP trigger

Switching only the `R_F<=0` genes to a true double-Poisson branch improved MAQC2 M03 approximately from

```text
352 TP / 22 FP
```

to

```text
363 TP / 22 FP.
```

This showed that the excluded genes could contain substantial real signal, but the DP branch rescued only a small fraction of the unusable population.

The later exact-normalizer audit showed why: the true problem was the local `C'(1)` precision approximation, not NB curvature itself.

The DP fallback is therefore development history, not the finalized rescue.

## Z.6 Exact-normalizer audit: every MAQC2 `R_F<=0` gene had a finite exact full precision root

The exact normalized-DNB audit solved

```text
U_F(theta)
=
a_F/theta
-
S_F
+
sum_i C'_Fi(theta)
=
0
```

for the problematic genes.

For M03:

```text
4,109 / 4,109
```

`R_F<=0` genes had finite exact full-model precision roots.

The DP-rescued subset had extremely shallow first-order deficits and often very large exact precisions. The previously DP-unrescued genes typically had deeper first-order deficits but still had ordinary finite exact roots.

This result changed the interpretation of the first-order failure:

```text
R_F^(1)<=0
```

does not imply that the DNB model lacks a finite precision.

It implies that the local normalizer approximation evaluated at `theta=1` is inadequate for that gene.

## Z.7 Exact normalized-DNB rescue substantially improved MAQC2

Replacing first-order-invalid M03 genes with exact normalized-DNB full/null profile inference changed the MAQC2 operating point from approximately

```text
baseline first-order M03:
    352 TP / 22 FP / 85 TN / 47 FN
```

to

```text
exact-normalizer rescue:
    381 TP / 24 FP / 83 TN / 18 FN.
```

Thus the exact rescue recovered:

```text
+29 TP
+2 FP
```

relative to the baseline first-order M03 branch.

The qPCR-reference FDR remained near 6%, while sensitivity rose to approximately 95.5%.

This was a much larger gain than the DP fallback and established exact normalized-DNB profiling as the appropriate `R_F<=0` rescue.

## Z.8 Exact-refitting the Top 500 ordinary genes did not improve inference

A separate experiment tested whether exact normalized-DNB precision should be used more broadly simply because a gene is highly ranked.

For each dataset, the final genome-wide Top 500 M03 genes were iteratively refined until every Top-500 gene had been exact-evaluated at least once.

Closure was achieved:

```text
MAQC2:
    500 / 500 Top-500 genes exact-evaluated

MAQC3:
    500 / 500 Top-500 genes exact-evaluated.
```

Yet:

```text
MAQC2:
    TP/FP unchanged
    AUC essentially unchanged

MAQC3:
    TP/FP unchanged
    AUC change ~ -0.00015.
```

Therefore **high rank is not an adequate trigger for exact profiling**.

The exact fallback is valuable when the first-order approximation demonstrates its own failure, not simply because the gene is scientifically interesting or highly significant.

This negative result is important for computational design.

## Z.9 The remaining exact-null failures exposed a discrete-support geometry problem

After exact full fitting, a residual canonical set of

```text
573
```

M03 `R_F<=0` genes still had

```text
null_root_status = root_above_search_max.
```

Their exact full-model roots were finite, but the exact null score remained positive over the entire upper search range.

A dedicated count-discrepancy audit confirmed that 573 is the correct canonical unresolved `R_F<=0` count.

Larger intermediate counts such as 645, 625, and 620 arose because the first feasible-support audit accidentally included sampled `R_F>0` reference genes with the same null-boundary status.

For the canonical 573-gene group, the null residual statistic had been shrunk below the minimum half-deviance permitted by the fitted discrete null DNB support.

## Z.10 Support-feasible geometry and the genome-wide silent violation population

For each gene, define

```text
S_0,min
=
sum_i min_y D_NB(y; mu_0i, r_g)/2.
```

The ordinary geometry

```text
S_0^dagger
=
S_F
+
(S_0-S_F)/lambda_c
```

can produce

```text
S_0^dagger < S_0,min.
```

The support-feasible correction uses

```text
B = max(S_F,S_0,min)
```

and

```text
S_0^(dagger,feasible)
=
B
+
max(S_0-B,0)/lambda_c.
```

The direct genome-wide audit showed that this was not a rare edge case.

For MAQC2 M03:

```text
38,833 analyzed genes
2,941 support-floor violations  (~7.58%)

2,206 violations with R_F > 0
  735 violations with R_F <= 0.
```

Thus approximately 6.35% of the apparently ordinary `R_F>0` population had a **silent support-infeasible null geometry**.

For MAQC3 M03:

```text
41,767 analyzed genes
835 support-floor violations (~2.00%)
835 with R_F > 0
0 with R_F <= 0.
```

The correction restored

```text
S_0^(dagger,feasible) >= S_0,min
```

for every identified violating gene in every audited branch.

This established that the support-floor check must be genome-wide and independent of the `R_F<=0` exact-fallback trigger.

## Z.11 Final support-feasible M03 production procedure

The finalized M03 procedure is:

```text
1. WEB-DP estimates r_g.

2. Fit full and constrained NB means.

3. Evaluate C'_F(1), C'_0(1).

4. Form R_F^(1)=S_F-C'_F(1).

5. Fit ordinary full DEB theta on R_F^(1)>0 genes.

6. Estimate ASH null weights and kappa_c from the established
   first-order-admissible population.

7. Set kappa_F=1 and lambda_c=kappa_c.

8. Construct the ordinary nested null statistic.

9. For every gene, calculate S_0,min.

10. If the ordinary nested null statistic falls below S_0,min,
    replace it with the support-feasible geometry.

11. Fit the first-order null DEB precision on the support-feasible
    null statistic.

12. For R_F^(1)>0 genes, retain the ordinary first-order DEB test.

13. For R_F^(1)<=0 genes, solve exact normalized-DNB full and null
    theta profiles using repeated C'(theta)/log C(theta) evaluation,
    with the support-feasible null statistic.

14. Combine the p-values and apply genome-wide BH.
```

The exact fallback is a one-dimensional conditional profile-MLE calculation for `theta` with the fitted means and `r_g` held fixed.

The support correction is a separate geometry feasibility constraint.

## Z.16 MAQC is a controlled reference-material benchmark, not a realistic biological-population benchmark

A crucial interpretation correction followed from understanding the experimental design.

MAQC compares biologically different RNA reference materials, but the replication is largely repeated measurement of standardized reference material rather than independent organisms drawn from heterogeneous biological populations.

Thus the dataset is closer to

```text
strong fixed biological contrast
+
predominantly technical replication/noise
```

than to

```text
two biological populations
+
substantial donor-to-donor heterogeneity
+
technical noise.
```

This makes MAQC an excellent sanity check for:

- count modeling;
- technical-noise control;
- qPCR agreement;
- ranking;
- calibration;
- behavior at very small nominal replicate counts.

It does **not** by itself establish superiority under ordinary biological heterogeneity.

The surprisingly strong MAQC2 performance at 2-vs-2 replication is therefore impressive but must be interpreted in the context of standardized reference materials.

## Z.17 MAQC sample-size follow-up and the need for real biological heterogeneity

The historical benchmark paper also used larger cancer datasets and repeated sample-size subsampling.

A valuable next experiment is to reproduce those biological-data sample-size analyses with frozen subsets and current edgeR/DESeq2 comparators.

Particularly useful is the within-group null design, where samples from the same cancer population are randomly split into two groups. That creates a real-data null containing genuine biological heterogeneity rather than standardized-reference replication.

The primary mechanistic question is whether, as biological replication decreases:

```text
R_F<=0 frequency rises,
support-floor violations rise,
DEB theta becomes liberal or unstable,
WEB theta becomes comparatively conservative,
or the finalized M03 architecture remains calibrated.
```

This experiment is now higher priority than further MAQC-specific tuning.

# AB. Real-data evidence should be organized by strength of truth, not by dataset prestige

The external-validation program now has a more coherent structure.

No single real-data benchmark supplies perfect biological ground truth. Different datasets trade truth certainty against biological realism.

The useful hierarchy is:

## Tier 1 — exact synthetic truth embedded in real sequencing libraries

Example:

```text
Baik / ERCC spike-ins.

```

Advantages:

```text
true direction/status is experimentally specified;
known concentration ratios;
clean ranking target.

```

Limitations:

```text
few features;
artificial molecules;
mostly technical rather than ordinary biological variation.

```

Baik belongs here.

## Tier 2 — orthogonal truth for endogenous genes

Examples include:

```text
MAQC / SEQC qRT-PCR
RNAontheBENCH NanoString

```

Advantages:

```text
real endogenous genes;
independent measurement technology;
substantially greater biological relevance than ERCC alone.

```

Limitations:

```text
truth exists for only a subset;
assay thresholds define the positive/negative sets;
identifier mapping matters;
intermediate genes may be excluded;
the resulting "reference FDR" is not identical to frequentist FDR.

```

MAQC belongs here.

## Tier 3 — modern reference-material and mixture benchmarks

The strongest eventual endpoint is the modern Quartet/MAQC multicenter framework, which combines:

```text
reference RNAs,
known mixtures,
orthogonal assays,
ERCC controls,
many laboratories,
many upstream processing pipelines.

```

This provides much stronger evidence about robustness across laboratories and data-processing choices.

Its disadvantage is practical complexity: the exact processed count matrices are not currently as immediately available as the Baik/MAQC objects already assembled for the project.

Therefore the Quartet/Wang benchmark is now best viewed as a **later heavyweight external validation**, not the immediate next experiment.

The general principle is:

```text
\boxed{
\text{do not let ERCC, qPCR, or a reference-mixture benchmark stand alone.}
}

```

A persuasive DGE validation should converge across multiple forms of truth.

The external real-data benchmark candidates and their complementary truth sources are summarized in the supplied benchmark review.

---

# AC. Historical benchmark reproduction becomes a second, independent validation axis

The strongest benchmark strategy is no longer to invent a very large custom simulation factorial and then ask whether WEB wins it.

Instead:

```text
\boxed{
\text{reproduce the benchmark arguments used to establish DESeq2 and edgeR,
verify those results without WEB,
then insert frozen WEB-DNB into the same machinery.}
}

```

This solves an important credibility problem.

If we design the distributions, effect-size grids, thresholds, and performance metrics ourselves after developing WEB, even a fair design can look method-favorable.

If WEB performs well in the experiments its competitors themselves used to demonstrate their strengths, the interpretation is much cleaner.

The resulting DGE validation now has **two orthogonal axes**:

```text
External-truth real data

```

and

```text
Competitor-owned historical benchmark reproduction.

```

A convincing result should eventually survive both.

# AD. DESeq2 home-field benchmark suite

## AD.1 Love et al. Figure 6: sensitivity subject to actual FDR control

The first DESeq2-owned simulation benchmark should replace the previously proposed generic ordinary-NB factorial as the primary ordinary-NB challenge.

Its essential design is approximately:

| ComponentHistorical setup |                                        |
| ------------------------- | -------------------------------------- |
| genes                     | 10,000                                 |
| truth                     | ordinary negative binomial             |
| mean/dispersion source    | empirical Pickrell joint distribution  |
| total sample sizes        | 6, 8, 10, 20                           |
| group sizes               | 3v3, 4v4, 5v5, 10v10                   |
| null fraction             | 80%                                    |
| DE fraction               | 20%                                    |
| fold changes              | 2, 3, 4                                |
| direction                 | randomized                             |
| multiple testing          | BH                                     |
| nominal calling level     | adjusted `p<.10`                       |
| primary criterion         | sensitivity versus achieved/actual FDR |

This is especially valuable because the generating model is **ordinary NB truth**:

```text
\boxed{
theta_g=1.
}

```

Thus WEB receives no benefit from generating the data under its enlarged double-NB family.

The primary question is not:

```text
which method makes the most discoveries?

```

It is:

```text
\boxed{
\text{among methods controlling actual FDR at the required level,
which has the greatest sensitivity?}
}

```

For the next genuinely new truth-known benchmark, the **finalized M03 procedure must be frozen in advance** and run alongside:

```text
current DESeq2
current edgeR QL.
```

The former WEB-theta, matched-DEB-DP, flexible-`r_g`, theta-shared, and geometry-ablation branches may be carried as diagnostic secondary analyses, but they should not be selected opportunistically after seeing the benchmark outcome.

Historical competitor configurations can be added in the reproduction stage where useful.

This exact benchmark now supersedes the earlier proposal to begin with a newly designed grid of generic NB simulations. The broader custom grid can still be useful later, but there is no reason to invent a home-field challenge when DESeq2 already supplied one.

## AD.2 Exact historical Pickrell specificity benchmark

The project's current 29-donor, 30-split Pickrell analysis remains valuable as an independent development and validation dataset.

It is **not**, however, the exact historical DESeq2 specificity benchmark.

The historical reproduction used:

```text
26 male Nigerian LCL samples
46-bp reads
10 samples drawn without replacement
random 5-vs-5 groups
30 randomizations
genes with zero total count excluded
fraction of raw p-values below .01.

```

Accordingly, two Pickrell analyses should coexist:

```text
29-donor project Pickrell
=
independent development / validation benchmark

```

and

```text
26-sample historical Pickrell
=
literal DESeq2 benchmark reproduction.

```

These should never be conflated.

## AD.3 Bottomly real-data reproducibility benchmark

The Bottomly benchmark supplies a different type of real-data evidence.

The historical experiment repeatedly divided the original 10-vs-11 mouse dataset into approximately:

```text
3v3 evaluation experiment

```

and

```text
7v8 verification experiment,

```

while respecting experimental-batch balance.

Rather than pretending the larger subset gives perfect truth, the benchmark used the higher-powered verification experiment as a proxy and could rotate the method defining verification truth.

Especially important is the matched-precision analysis:

```text
precision = 0.90
<=> approximate FDR = 0.10

```

followed by comparison of sensitivity.

This asks essentially the same question the WEB project now considers scientifically important:

```text
\boxed{
\text{at the same empirically achieved error budget,
which method retrieves more reproducible signal?}
}

```

A WEB win there would be particularly persuasive because the evaluation criterion is inherited from DESeq2 rather than designed for WEB.

The supplied benchmark review establishes these DESeq2 home-field benchmarks as the first historical-reproduction family.

# AE. edgeR home-field robustness benchmark

The edgeR-affiliated Zhou, Lindsay & Robinson benchmark provides the complementary challenge.

Its representative simulation used approximately:

| ComponentSetup         |                                 |
| ---------------------- | ------------------------------- |
| genes                  | 10,000                          |
| base model             | ordinary NB                     |
| mean/dispersion source | empirical Pickrell distribution |
| extreme dispersions    | upper 10% removed               |
| groups                 | 5v5                             |
| DE fraction            | 10%                             |
| DE direction           | half up / half down             |
| fold change            | 3                               |
| contaminated genes     | 10%                             |
| contamination          | one randomly positioned outlier |
| outlier multiplier     | approximately 1.5–10            |

The same underlying simulated dataset could be evaluated both:

```text
without outliers

```

and

```text
with outliers.

```

That pairing is especially informative because it separates ordinary-NB performance from robustness.

The benchmark also varied difficulty and sample size, including roughly:

```text
hard effects:   FC 2-2.2
medium effects: FC 3-3.3
easy effects:   FC 6-6.6

```

with nominal FDR thresholds around:

```text
.02
.05
.10

```

and comparisons at 3v3, 5v5, and 10v10.

The key WEB challenge is therefore:

```text
\boxed{
\text{At the same achieved FDR,
does WEB retain more power under both clean NB and outlier-contaminated NB?}
}

```

This is a substantially stronger robustness claim than merely showing WEB behaves well under double-NB-generated data.

The Zhou benchmarking framework was designed in a form amenable to adding new methods, making it particularly suitable for direct WEB insertion.

# AF. FDR-calibration benchmark tradition: Burden / QL-style evaluation

A third historical family targets the error estimate itself rather than only power.

The Burden-style benchmark used tens of thousands of Pickrell-derived transcript parameter combinations, ordinary NB simulation, multiple sample sizes, multiple DE prevalences, and compared the **reported FDR with the known true FDR along the ranked discovery list**.

It also included Poisson-inverse-Gaussian simulations, deliberately violating the NB sampling assumption.

The relevant sample sizes included approximately:

```text
2, 3, 4, 6, 10 per condition

```

with true-DE proportions around:

```text
1%, 5%, 10%, 15%.

```

This gives WEB two separate tests:

```text
NB truth:
does reported error track actual error?

```

and

```text
non-NB truth:
does calibration remain useful under distributional misspecification?

```

The evaluation should emphasize the leading edge of the ranked list, where actual scientific discoveries are made, rather than only global AUC.

The benchmark question is:

```text
\boxed{
\text{Does WEB's nominal/reported error track the true error at least as well as
edgeR QL and DESeq2 while retaining comparable or greater discovery power?}
}

```

This benchmark complements Love Figure 6 and Zhou Figures 4–5 rather than duplicating them.

# AG. Historical reproduction must precede WEB insertion

Every historical benchmark should be run in two stages.

## Stage 1 — reproduction

Use the original benchmark code, data objects, simulation logic, preprocessing, and where practical the historical package behavior.

The purpose is to establish:

```text
\boxed{
\text{we can reproduce the published qualitative/quantitative benchmark
before WEB is present.}
}

```

If the competitor ordering does not reproduce, WEB should not yet be interpreted.

## Stage 2 — modern replay with frozen data

Freeze the exact same simulated datasets, real-data splits, and truth labels and run the **finalized M03 procedure** together with:

```text
current edgeR QL
current DESeq2.
```

Selected historical WEB/DEB branches may be added as diagnostic ablations, but the primary comparison should not choose among them after seeing the benchmark outcome.

This guards against the obvious objection:

> WEB was compared with a 2014 version of a competitor that has since improved.

Both the historical result and the modern replay are scientifically useful, but they answer different questions.

# AH. The next independent benchmark must freeze the finalized M03 procedure

The earlier development plan required carrying all 12 WEB-DNB branches into the next benchmark. That was appropriate before the MAQC audits resolved several of the open architecture questions.

The current production candidate is now sufficiently specific to freeze directly:

```text
r_g:
    WEB-DP

theta:
    DEB on the ordinary first-order branch

shared coordinate:
    C'(1) for ordinary admissible genes

kappa_F:
    1

kappa_c:
    ASH-weighted observed contrast geometry

exact fallback:
    exact normalized-DNB theta profile if R_F^(1)<=0

null feasibility:
    genome-wide support-floor check and support-feasible geometry

multiple testing:
    declared BH procedure.
```

The exact fallback and support-floor correction are part of the prespecified method because they repair internally detectable mathematical failures.

For future benchmark reproducibility, the frozen configuration must also include:

- normalizer tolerances;
- exact root-search bounds/tolerances;
- support-floor tolerance;
- ASH settings;
- `r_g` bounds;
- DEB fitting settings;
- negative-`T_raw` handling;
- multiple-testing universe;
- checkpoint/resume behavior.

The former alternative branches should be retained as **diagnostic ablations**, not selected opportunistically within each benchmark cell:

```text
WEB theta,
matched trend-DEB-DP r_g,
flexible trend-DEB-DP r_g,
theta-shared C'(theta_F),
lambda_c = 1 geometry ablation.
```

If a future benchmark exposes a reproducible failure of the frozen M03 procedure, that failure should be diagnosed mechanistically rather than repaired by selecting whichever historical branch happens to win that dataset.

# AI. `r_g`, DEB `theta`, and the current interpretation of where performance comes from

The project initially treated `r_g` as the least mature layer, and that remains partly true: WEB-DP is young relative to the highly engineered dispersion systems in edgeR and DESeq2.

However, the evidence now separates the contributions of the two DGE dispersion/precision layers more clearly.

## `r_g`

The first-stage candidates were:

```text
WEB-DP,
matched trend-DEB-DP,
flexible trend-DEB-DP.
```

WEB-DP and matched trend-DEB-DP repeatedly produced nearly indistinguishable practical inference on Baik and MAQC.

The flexible trend-DEB-DP branch changed curvature estimates more substantially but did not produce a consistent operating-point advantage.

Thus WEB-DP remains the leading first-stage estimator, with matched-DEB-DP retained as a strong robustness/parameterization check.

## `theta`

The second-stage precision estimator now appears substantially more consequential.

Across Baik and MAQC, DEB `theta` produced a much stronger false-positive/power tradeoff than WEB `theta`.

The MAQC2 result became especially decisive after the exact-normalizer admissibility correction: the final M03 DEB branch achieved high sensitivity, much lower qPCR-reference FDR, higher MCC, and higher AUC than the current edgeR and DESeq2 comparators.

Thus the current working interpretation is:

```text
WEB-DP provides a useful estimate of NB curvature;
DEB provides the leading residual-precision shrinkage;
exact normalized-DNB profiling is reserved for genes where the C'(1)
approximation explicitly fails.
```

## Exact profiling is not a replacement for DEB everywhere

The Top-500 exact-refinement experiment showed that exact profiling of already-admissible highly ranked genes did not improve the final calls.

Therefore the current method is not:

```text
fit exact theta for every important gene.
```

It is:

```text
use DEB theta where the fast approximation is admissible;
use exact normalized-DNB theta where the approximation is not.
```

## Remaining identifiability question

The larger theoretical question remains how reliably `r_g` and `theta_g` can be distinguished under genuine biological heterogeneity.

MAQC's standardized reference-material replication may make technical residual precision unusually identifiable.

This is one reason real biological null/subsampling benchmarks are now especially important.

# AJ. The first real-data benchmark runner remains part of the audit trail, but it is no longer the whole DGE development infrastructure

The original real-data wrapper deliberately reused the flagship Pickrell function rather than rewriting WEB-DNB for Baik or MAQC.

For each of:

```text
Baik_GSE49712
MAQC2
MAQC3
```

it built an adapter directory and passed the standardized count matrix and real experimental labels through the same then-frozen fitting machinery.

That first-run infrastructure remains important because it documents the genuine out-of-sample Baik test of the original architecture.

Afterward, a separate explicit 12-method experiment was built to study:

```text
3 r_g estimators
x 2 theta estimators
x 2 shared coordinates,
```

with resumable/checkpointed null fitting and numerical audits.

The current DGE method should therefore distinguish:

```text
first frozen external Baik run
```

from

```text
later Baik development experiments.
```

External truth labels remain joined only after genome-wide fitting and are not used to fit `r_g`, `theta`, ASH, `kappa_c`, or the multiple-testing universe.

# AK. Revised interpretation of the Baik/MAQC stage

## Baik

Baik has now done its job twice:

1. it supplied the first genuinely frozen real-signal test of the Pickrell-developed WEB-DNB architecture;
2. it then became development data for the expanded `r_g` / `theta` / shared-coordinate grid and the `kappa_F` audit.

The first frozen result showed WEB ranking close to current edgeR and well above DESeq2 on the 63 ERCCs.

The later development grid showed that DEB `theta` can sharply improve the threshold false-positive tradeoff, that matched DEB-DP behaves almost identically to WEB-DP, and that the first-order versus theta-shared coordinate produces little change in Baik calls.

Those later findings are scientifically useful but must not be advertised as untouched external validation.

## MAQC

MAQC should be described more carefully than Baik.

The qPCR truth mapping remains historically unresolved, so the current scores are explicitly tied to the current reconstructed qPCR universe.

At the same time, the major post-MAQC method changes were **not chosen by optimizing against qPCR labels**.

They were triggered by internal mathematical failures visible before truth scoring:

```text
R_F^(1) <= 0
```

showed that the first-order normalizer approximation could become inadmissible, and

```text
S_0^dagger < S_0,min
```

showed that the geometry-corrected null statistic could leave the support of the fitted discrete distribution.

The exact-normalizer fallback and support-feasible null geometry therefore have a different provenance from ordinary benchmark tuning. They are data-exposed but truth-blind mathematical repairs.

Their effect on qPCR performance was evaluated only after the internal defects were identified.

The MAQC results nevertheless should not be generalized directly to ordinary heterogeneous biological experiments. MAQC uses standardized reference RNA materials and therefore contains much less independent biological replicate heterogeneity than a typical human or animal DGE study.

Thus the strongest current interpretation is:

```text
MAQC demonstrates excellent controlled-reference performance and external qPCR agreement;
the next question is whether the same architecture remains calibrated and powerful
under genuine biological heterogeneity.
```

The historical identifier mapping should still be recovered before presenting MAQC as an exact reproduction of the original benchmark.

# AL. The validation program now has a two-dimensional framework

The complete DGE benchmarking strategy can be viewed as a matrix.

One dimension is **truth type**:

```text
exact spike-in truth
orthogonal endogenous truth
proxy/high-replicate truth
known simulation truth
model-misspecified truth
modern reference-material truth.

```

The second dimension is **benchmark provenance**:

```text
WEB-developed benchmark
DESeq2-owned historical benchmark
edgeR-owned historical benchmark
independent third-party benchmark
modern external reference-material benchmark.

```

A result is more persuasive when WEB performs well across cells of this matrix rather than only within one favored benchmark family.

For example:

```text
Baik
=
exact synthetic truth + independent real-data benchmark

Love Figure 6
=
known ordinary-NB truth + DESeq2 home field

Zhou
=
known NB/outlier truth + edgeR home field

Bottomly
=
real-data proxy truth + DESeq2 home field

MAQC
=
orthogonal endogenous truth + independent real-data benchmark

Burden/PIG
=
known truth under model misspecification + independent error-control benchmark

Quartet
=
modern reference-material truth + large multicenter independent benchmark.

```

This is much stronger than asking whether one aggregate benchmark declares WEB the winner.

# AP. Revised interpretation of success and failure

The project has moved beyond the question of whether WEB-DNB can produce reasonable DGE results.

The current questions are now:

```text
1. Does finalized M03 retain its calibration/power advantage under genuine biological heterogeneity?

2. Does the DEB-theta advantage survive smaller biological sample sizes,
   or does WEB theta become preferable in low-information regimes?

3. How often do R_F<=0 and support-floor violations occur outside
   standardized reference-material datasets?

4. Does the current exact-fallback/support-feasible procedure preserve
   null calibration in competitor-owned simulations and real-data nulls?

5. How much remaining error is attributable to r_g estimation versus
   the residual-precision/test-statistic layer?

6. What is the correct theoretical treatment of negative T_raw values?
```

The project should continue separating:

```text
ranking quality,
operating-point calibration,
power at matched achieved error,
coverage/testability,
parameter-estimation error,
numerical failure,
mathematical admissibility failure.
```

A method can rank well but calibrate poorly, calibrate well but lose power, or appear conservative because genes are untestable.

The recent MAQC audits showed why these distinctions matter: the early first-order M03 result had a genuine coverage defect, yet the favorable qPCR-reference FDR was not caused by selective exclusion. Repairing the approximation recovered large amounts of signal without destroying calibration.

Future failures should therefore be diagnosed at the layer where they arise rather than summarized by a single discovery count or AUC.

# AQ. Strongest eventual validation claim

The strongest eventual claim would require convergence across independent benchmark families, not simply dominance on MAQC reference materials.

A persuasive result would look like:

```text
1. Finalized M03 remains calibrated on independent complete-null data.

2. The exact-fallback and support-feasible geometry preserve calibration rather
   than merely improving reference-material signal benchmarks.

3. M03 retains favorable power/error tradeoffs in biological real-data null
   subsampling with genuine within-group heterogeneity.

4. M03 achieves greater or comparable power at equal achieved FDR in the
   DESeq2 Love Figure-6 ordinary-NB benchmark.

5. M03 remains robust in the edgeR/Zhou clean-versus-outlier benchmark.

6. Reported/nominal error tracks known true error in Burden-style calibration
   experiments, including model misspecification.

7. The method remains competitive in Bottomly reproducibility and independent
   external truth benchmarks such as RNAontheBENCH/SEQC.

8. The result survives comparison with current edgeR QL and current DESeq2,
   not only historical package versions.

9. The same architecture extends coherently to general-X DGE and remains
   closely connected to the established DEU/DTU framework.
```

Only after that convergence would a broad dominance claim over modern NB DGE methods be justified.

The current MAQC result is nevertheless unusually strong on its own terms: on the current qPCR-reference universe, finalized M03 combines much lower reference-FDR, very high sensitivity, higher MCC, and higher AUC than current edgeR QL and DESeq2 Wald on both MAQC datasets.

The appropriate current claim is therefore benchmark-specific:

```text
\boxed{
On the current reconstructed MAQC qPCR-reference benchmarks, finalized M03
substantially improves the error/power tradeoff relative to current edgeR QL
and DESeq2 Wald while also improving ranking and MCC.
}
```

The next independent benchmarks determine how broadly that statement generalizes.

# AS. Changes to the previous benchmark and geometry roadmap

The preceding summary remains valid except where explicitly superseded above.

The principal updates are now:

```text
- The old 12-method grid remains development history, but finalized M03 is now the
  leading DGE production candidate.

- Finalized M03 uses:
      WEB-DP r_g
      + DEB theta
      + kappa_F=1
      + ASH kappa_c
      + first-order C'(1) for ordinary admissible genes
      + exact normalized-DNB theta fallback for R_F<=0
      + genome-wide support-feasible null geometry.

- R_F<=0 is now understood as a failure of the local C'(1) precision
  approximation, not proof that the DNB precision is undefined.

- On MAQC2 M03, all 4,109 R_F<=0 genes had finite exact full theta roots.

- The exact fallback is a conditional one-dimensional normalized-DNB profile-MLE
  fit for theta with mu and r held fixed; it is not simply another DEB posterior
  formula.

- Double-Poisson fallback was informative but is no longer the primary rescue.
  The Poisson-boundary trigger was too broad, and the R_F<=0 DP branch rescued far
  fewer genes than exact DNB profiling.

- The exact Top-500 refinement experiment showed that exact profiling of already
  admissible high-ranking genes produced essentially no inferential improvement.
  Exact fitting should therefore be triggered by mathematical approximation failure,
  not by rank alone.

- A second, independent pathology was discovered: the nested geometry can produce
  S_0^dagger below the minimum half-deviance permitted by the discrete null DNB
  support.

- The support-feasible correction preserves the irreducible support cost and shrinks
  only the residual increment above the support floor.

- The support-floor check must be genome-wide. On final-audit M03:
      MAQC2: 2,941 / 38,833 support-floor violations,
             including 2,206 silent R_F>0 violations.
      MAQC3: 835 / 41,767 support-floor violations,
             all R_F>0.

- The feasible-support construction restored the support inequality for every
  identified violating gene.

- The finalized M03 rerun produced finite p-values for every analyzed gene:
      MAQC2: 38,833 / 38,833
      MAQC3: 41,767 / 41,767.

- Final canonical MAQC2 M03:
      381 TP / 24 FP / 83 TN / 18 FN
      sensitivity 0.954887
      qPCR-reference FDR 0.059259
      MCC 0.746385
      AUC 0.948985.

- Final canonical MAQC3 M03:
      375 TP / 16 FP / 103 TN / 28 FN
      sensitivity 0.930521
      qPCR-reference FDR 0.040921
      MCC 0.770286
      AUC 0.950122.

- Current edgeR QL and DESeq2 Wald have substantially higher qPCR-reference FDR
  and lower MCC on both current reconstructed MAQC truth universes; final M03 also
  has higher AUC on both.

- Near-final matched-error frontier analyses showed M03 recovering more true
  positives with fewer false positives than edgeR/DESeq2 at their achieved
  qPCR-reference error budgets. These descriptive frontier calculations should be
  refreshed from the final support-feasible p-values before manuscript use.

- MAQC is now interpreted explicitly as a controlled biological-reference-material
  benchmark with predominantly technical replication, not as a complete surrogate
  for ordinary heterogeneous biological populations.

- The next high-priority benchmark is genuine biological heterogeneity, including
  historical cancer sample-size / within-group-null subsampling.

- Negative T_raw remains a separate unresolved test-statistic issue. It is not an
  admissibility failure and is currently handled by T=max(T_raw,0).

- The C'(theta_DEB) one/two-step recentering experiment remains an efficiency audit,
  not part of the finalized production method until results justify it.

- All future iterative computation must be resumable/checkpointed by default, and
  top-level execution functions must print local start and end times.
```

This is the current boundary between **development history**, **retired approaches**, the **finalized M03 production candidate**, and the **next independent validation program**.

# Potential improvements to WEB / DEB / WEB-DNB

## DEU/DTU

The flagship improvement is now established rather than merely proposed:

```
ASH gene/event null weights
+
one posterior-null-mass kappa_F^{family}
+
one kappa_c per contrast
+
nested increment correction.
```

Continue finite-sample ALO/exact-LOO and information-based correction work as a separate refinement layer.

Compare empirical

```
E(q)
```

and

```
Var(q)
```

surfaces where small-sample behavior warrants it.

Develop the multiple-df extension using a tested-subspace matrix/operator rather than forcing a scalar one-df correction.

Reuse the single full fit, information matrices, and family null information efficiently across many planned contrasts.

Do not add gene-specific `\lambda_{cg}`, hierarchical family-null weights, or extra `pi_0c` weighting unless future residual diagnostics show that the scalar family/contrast architecture is insufficient.

---

## DGE

The current leading DGE production architecture is:

```text
WEB-DP r_g
+
DEB theta on ordinary admissible genes
+
ASH kappa_c
+
kappa_F = 1
+
genome-wide support-feasible null geometry
+
exact normalized-DNB theta fallback when R_F^(1)<=0.
```

The immediate improvement targets are now narrower and more mechanistic.

### 1. Make exact fallback cheaper without changing its target

The exact fallback repeatedly evaluates `C'(theta)` and `log C(theta)` during one-dimensional root solving.

A separate audit is testing whether

```text
C'(theta_DEB)
```

followed by one or two fixed-point/recentering updates can approximate the exact normalized root well enough to reduce computation.

This is an efficiency question only. The exact normalized score remains the target.

### 2. Do not exact-refit genes merely because they rank highly

The completed Top-500 experiment found essentially no inferential gain from exact-refitting ordinary admissible top-ranked genes.

Therefore a generic Top-K exact-refinement rule is not currently justified.

### 3. Resolve negative `T_raw`

Some genes have valid full and null fits but a negative geometry-adjusted raw statistic.

The current production rule is

```text
T=max(T_raw,0).
```

This remains a theoretical issue distinct from admissibility.

Potential directions include deriving a test statistic appropriate to the geometry-adjusted profile objective or reformulating the penalized/full-null comparison so that the desired nesting/nonnegativity property is explicit.

### 4. Quantify behavior under genuine biological heterogeneity

MAQC's standardized reference-material replication is unusually controlled.

The next important test is whether the separation

```text
r_g
=
NB curvature

theta_g
=
residual precision
```

remains identifiable and calibrated under real between-subject variation.

### 5. Quantify sample-size behavior

Track, as biological replication decreases:

```text
R_F<=0 fraction,
support-floor violation fraction,
exact-fallback rate,
negative-T rate,
DEB theta behavior,
WEB theta behavior,
r_g boundary rate.
```

This will determine whether DEB theta remains the preferred precision estimator in low-information biological settings.

### 6. Generalize support-feasible geometry to general `X`

The support-floor concept is not inherently two-group.

For each constrained model, the corrected residual target must remain inside the attainable support of the fitted discrete distribution.

The general-`X` extension should preserve that principle contrast by contrast.

### 7. Keep fully joint normalized-DNB fitting as a later escalation path

A fully joint fit of

```text
beta,
r,
theta,
C
```

is mathematically possible but is not currently required to repair the observed failures.

It should be pursued only if independent benchmarks reveal a systematic limitation of the current conditional architecture.

## Single-cell

Develop biological-replicate-aware residual geometry.

Distinguish donor covariance from contrast-scale calibration.

Investigate partial borrowing of precision or geometry across related cell types.

Distinguish UMI, full-length transcript, exon/junction, and nuclear count regimes.

| DESeq2 exposure-only Wald | 18 | 11 | 115 | 6 | 0.615198 | 0.928571 |

M03 ranked slightly better by the displayed AUC, but edgeR had fewer NanoString-reference false calls at the declared threshold and the same 18 true calls. That is not an operating-point M03 win. M03's reference false-discovery proportion at that point was \(11/29\approx37.93\%\), versus \(8/26\approx30.77\%\) for current edgeR. These are disagreement fractions within the small externally scored panel, not literal genome-wide frequentist FDR. [S03]

## AT.2 Exact precision confirmation was informative, but screening and full rescoring are different procedures

Exact-precision confirmation of the current M03 p < 0.01 discoveries changed the displayed operating point from **18 TP / 11 FP** to **17 TP / 4 FP**. The latter gives 21 calls, reference FDP approximately 19.05%, and MCC approximately 0.714842. This is a **screen-then-confirm rule**, not proof that exact profiling improves the whole ranking.

A separate exact calculation on the complete 150-gene NanoString panel produced **146 finite exact results**, **18 TP / 9 FP**, MCC approximately 0.647515, and AUC approximately 0.935516. The transition table showed 21 genes called under both procedures, eight current-only calls lost after exact fitting, and six exact-only calls gained. The exact-all-panel result therefore did not reproduce the 17/4 screened operating point. [S03]

Exact-called confirmation also reduced reference false calls in the displayed M01/M09/M11 comparisons. M01 went from 18/11 to 17/4; M09 and M11 went from 18/14 to 17/7. These are development ablations and must retain their actual method labels rather than being merged into one canonical M03 result.

## AT.3 Matched-error comparison did not establish superiority over edgeR

At or below the M03 screened reference-FDP budget, edgeR could recover **17 TP / 2 FP**; at a threshold matching the M03 reference-FDP value, edgeR recovered **17 TP / 4 FP**. In the reverse comparison, allowing the M03 screen threshold to change while retaining exact confirmation could reach **18 TP / 8 FP**, matching the current edgeR threshold result. [S03]

**Conclusion:** RNAontheBENCH supplied a useful external challenge. It showed competitive ranking, an informative exact-confirmation effect, and no demonstrated matched-error advantage over edgeR in the displayed experiment. The strong MAQC operating-point result therefore did not automatically generalize to this panel. The exact-called and exact-all-panel experiments should remain separate from the mathematical-failure-triggered exact fallback used in the archived M03 specification.

<a id="au-biological-sample-size-work"></a>
# AU. Biological sample-size work

## AU.1 The Li2020 reproduction program moved from planning into execution

The local preparation code froze group sizes **5, 10, 15, 20, 25, 30, 35, 40** and **50 within-null repetitions**. Its own comments distinguish the authors' stated repetition counts in other arms from the project's transparent choice of 50 repeats for the within-population null arm. The distinction between literal historical reproduction and project extension must be retained. [S18]

The sample-origin audit also found that labels could not safely be inferred from file names alone. The supplied OC metadata supported a cancer-only interpretation; some AdLC and TNBC identifiers were opaque or did not support the initially assumed cancer-versus-normal interpretation. Analyses must use the provenance actually established by metadata, and unresolved origin should remain unresolved rather than being filled in from expectations. This is especially important when describing a split as “within-cancer null.” [S18]

## AU.2 Current runtime evidence

Sean reported that one **40v40, 50-replicate, 750-fit** run lasted from:

- September 14, 2026, 4:33:55 PM local time;
- to September 16, 2026, 7:15:23 PM local time.

The elapsed time is **50 hours, 41 minutes, 28 seconds**, or approximately **243.3 seconds per fit averaged over that workload**. This is a direct calculation from the reported timestamps, not a forecast for the remaining sample sizes. [S02, S18]

Sean described **three 750-fit types run simultaneously at each sample size**, with 35v35 down through 5v5 still in the program. Seven remaining sizes would correspond to **15,750 fits** if all three types and all 750 fits per type remain required. Wall-clock time cannot be obtained by multiplying a single isolated-fit estimate without accounting for concurrent resource sharing, sample-size-dependent fitting costs, and already completed checkpoints.

No complete final FPR/power summary for that entire stress-test program is present in the latest record. Completion of a timed batch is not a completed statistical validation claim.

<a id="av-bottomly"></a>
# AV. Bottomly

## AV.1 Why Bottomly was initially useful

The project reconstructed the historical 10-versus-11 mouse setting, with repeated small 3v3 evaluation samples and larger verification subsets. It compared M03, edgeR, and DESeq2 under several ways of turning larger-data results into reproducibility references. The important limitation became increasingly clear: **verification calls are method-dependent proxy truth**.

The working record included historical split verification, top-K verification sets, pairwise zero/one/two-method agreement, and full-21 reference fits. Full-21 references reduced split-specific reference variability but overlapped the evaluation samples, so they were not independent verification truth. [S02, S04]

## AV.2 The truth-set construction was deliberately varied

The tested constructions included:

**Separate full-data method references.** Rank genes using the full-data M03, edgeR, or DESeq2 result and define top-K truth separately. Rotate which method supplies the reference rather than treating one method as omniscient.

**Half-K unions.** Take top K/2 from M03 and top K/2 from the comparator. Their union has fewer than K genes when the lists overlap. For example, nominal K=50 gave 28 genes for an M03+edgeR reference with 22 shared genes, and 31 for M03+DESeq2 with 19 shared genes. These are not exact-K references.

**Exact-K shared/unique rank construction.** Sean proposed separating shared candidates and the method-specific unique candidates, retaining rank information, and alternating between methods when selecting unique genes. The example was shared ranks `{1,6,7,8,9,10}`, edgeR-only `{2,3}`, and WEB-only `{4,5}`: selections progress as `{1}`, `{1,2}`, `{1,2,4}`, `{1,2,4,3}`, and `{1,2,4,3,5}`. This avoids letting one method supply consecutive unique picks merely because its next unique candidate has the next absolute rank.

Sean then replaced the minimum-rank treatment of shared genes with **the average of their two method ranks followed by reranking**. The resulting exact-K references included fixed K values and K values derived from full-data q < 0.10 discovery counts. These algorithmic details matter: they change the target, not merely how a table is displayed. [S02, S04]

## AV.3 Method-relative conclusions depended on the reference

In the displayed 30-split summaries, M03 often recovered more proxy-reference positives than edgeR for larger mixed truth sets but with a precision tradeoff. Relative to DESeq2, M03 frequently made fewer recovered calls and had different sensitivity/precision tradeoffs. The apparent “winner” was not stable across truth definitions, K, thresholding, and comparator pairing.

The full-21 q < 0.10 call counts used in the later summaries were approximately **962 M03**, **1,280 edgeR**, and **1,702 DESeq2**. Evaluation-set mean call counts were approximately **129.4**, **100.1**, and **156.1**, respectively. These numbers describe the displayed configuration; they should not be mixed with earlier q < 0.01 runs or a different verification construction. [S04]

Progressive exact-precision analyses and a top-2000 exact experiment were also implemented or launched. The available record is insufficient to declare a single final whole-Bottomly exact-refinement outcome. Code existence and a running progress log do not establish completion.

## AV.4 No-TMM M03 experiment

Sean asked whether TMM normalization might favor edgeR and be responsible for WEB/M03's Bottomly behavior. A library-size-only M03 diagnostic reused the relevant data and reran the full-21 and 30 evaluation fits without TMM; competitor procedures were not silently redefined.

The displayed TMM versus no-TMM comparison was modest:

| Quantity | TMM | No TMM / library-size-only |
|---|---:|---:|
| Full-21 M03 calls | 962 | 946 |
| Mean evaluation calls | 129.4 | 127.1 |

Full-data call-set Jaccard was approximately **0.967**, with p-rank correlation approximately **0.9985**. Across evaluation splits, mean call-set Jaccard was approximately **0.921** and mean p-rank correlation approximately **0.9819**. The later exact-K comparisons remained qualitatively similar. [S04]

**Conclusion:** removing TMM was a legitimate mechanism check but did not rescue the Bottomly comparison or explain the major behavior. It does not prove that normalization never matters for WEB; it limits this particular explanation on this dataset.

## AV.5 Final decision

Sean explicitly removed Bottomly from the active benchmark evidence program because the information was too difficult to judge. The methodological experiments are retained in the archive, but Bottomly is no longer a required future “win,” an active tuning target, or decisive validation evidence. This supersedes the Bottomly recommendations in the inherited roadmap. [S02]

<a id="aw-mixology-preparation-and-initial-failure"></a>
# AW. Mixology preparation and initial failure

## AW.1 Exact preparation required filtering on all 20 poly-A mRNA samples

The GSE64098 preparation used H1975/HCC827 mixtures at **0%, 25%, 50%, 75%, and 100% H1975**. There were three good poly-A mRNA replicates per mixture level, giving **15 good samples** and **10 pairwise 3v3 contrasts**. The preparation also contained one degraded R2D sample per mixture level, giving **20 poly-A mRNA samples** for the paper-matching filtering step. [S05]

The first attempt filtered only the 15 good samples. It retained **14,905** genes at CPM > 1 in at least three samples and **16,031** at CPM > 0.5 in at least three samples, short of the reported targets.

Filtering on all 20 samples, then selecting the 15 good samples for inference, reproduced:

| Preparation item | Corrected result |
|---|---:|
| All poly-A samples used in CPM filtering | 20 |
| Good samples used in 3v3 comparisons | 15 |
| Genes before filtering | 23,669 |
| CPM > 1 in at least three samples | **14,981** |
| CPM > 0.5 in at least three samples | **16,131** |
| Pairwise 3v3 contrasts | 10 |

The failed `samples` variable reference was an implementation error; the later preparation log confirmed the corrected counts. [S05]

## AW.2 Meaning of the contrast labels

`100_vs_000` compares three pure-H1975 samples with three pure-HCC827 samples. It is **not 100 samples versus zero samples**. All ten comparisons in this experiment are 3v3. Likewise, `075_vs_025` refers to mixture percentages, not sample counts.

This corrected the initial speculation that unequal group sample sizes explained the pathology. It did not settle general imbalance questions elsewhere in the project; it established that sample-count imbalance was not the explanation for these ten particular comparisons. [S02, S05]

## AW.3 Initial benchmark results exposed a serious M03 problem

The first Stage-2 run showed high M03 rank/sign agreement with the fitmixture reference in several respects, but poor Pearson/RMSE behavior and highly irregular discovery patterns across attenuated contrasts.

Across the ten contrasts, the displayed averages included:

| Method | Mean RMSE | Mean Pearson | Mean Spearman | Mean q < .05 calls |
|---|---:|---:|---:|---:|
| M03 | 1.7141 | 0.7338 | 0.9504 | 4,856.4 |
| edgeR | 0.2322 | 0.9683 | 0.9503 | 6,772.9 |
| DESeq2 | 0.2354 | 0.9642 | 0.9505 | 6,693.0 |

The mismatch between M03's high rank correlation and poor absolute error was a real reported result. The later experiments primarily modified uncertainty/testing through \(\lambda_c\), not the upstream fitted means. Therefore those later tests do **not** establish that the original fold-change estimation/RMSE issue was repaired. [S06]

The pure comparison gave 6,237 M03 q < .05 calls, whereas `075_vs_025` gave 8,931, including thousands of calls absent from the pure comparison. Such differences are not automatically false positives—contrast-specific noise and power can genuinely differ—but the magnitude and systematic behavior motivated the subsequent diagnostics.

## AW.4 Reference effects are a benchmark target, not complete binary truth

The saved `ref.csv` files provide the fitmixture-predicted log2FC reference used in the analyses. They are distinct from M03's observed fitted fold changes. The current summary preserves that distinction.

The reference effect can support directional, magnitude, and monotonicity diagnostics. It does not turn every extra attenuated call into a known false discovery. In particular, “new call with weaker reference effect” is a diagnostic event, not a direct FDR estimate. [S06, S16]

<a id="ax-dense-signal-contrast-geometry"></a>
# AX. Dense-signal contrast geometry

## AX.1 Sean identified the central problem with the original \(\kappa_c\) estimator

The original estimator used ASH null weights to isolate the contrast-noise contribution from an observed increment containing both noise and alternatives. Sean pointed out that this becomes unreliable when the comparison contains few or no useful null genes. The statement “there are no null genes” was the motivating dense-signal interpretation; the benchmark does not independently establish literal zero null genes in every contrast.

The displayed original estimates were:

| Contrast | Original \(\kappa_c\) | Reference median absolute log2FC |
|---|---:|---:|
| 100 vs 0 | 4.3193 | 0.5634 |
| 100 vs 25 | 3.0050 | 0.4263 |
| 75 vs 0 | 2.1305 | 0.4025 |
| 100 vs 50 | 2.4955 | 0.2845 |
| 75 vs 25 | 1.0235 | 0.2720 |
| 50 vs 0 | 1.8944 | 0.2726 |
| 100 vs 75 | 1.8908 | 0.1477 |
| 75 vs 50 | 1.0854 | 0.1343 |
| 50 vs 25 | 1.0766 | 0.1374 |
| 25 vs 0 | 1.6032 | 0.1416 |

The across-contrast Spearman relationship with reference effect strength was approximately **0.8545**. This alone is not proof of contamination, but the subsequent interventions supplied much stronger mechanism evidence. [S06, S07]

## AX.2 Fixed \(\kappa_c=\lambda_c=1\) was an informative falsification

Holding the upstream M03 fit fixed and setting \(\lambda_c=1\) greatly reduced the pathological contrast-strength dependence of the testing behavior.

| Diagnostic | Original M03 | Fixed \(\lambda=1\) |
|---|---:|---:|
| Median gene-level Spearman across contrasts, all genes | 0.564 | **0.806** |
| Median gene-level Spearman, strong-reference stratum | 0.685 | **0.879** |
| Fraction of negative gene-level correlations, all genes | 15.1% | **8.2%** |
| Mean material raw-p inversion fraction, all genes | 27.6% | **14.1%** |
| Mean material q < .05 call inversion fraction | 3.4% | **1.9%** |

Calls increased, sometimes substantially: for example, `100_vs_000` changed from **6,237 to 10,995**. That rise was not itself proof of liberal FDR; the experiment lacked complete binary null truth. Sean correctly insisted that a larger discovery count alone was not a satisfactory resolution. [S07]

## AX.3 Homogeneous/null-rich column subsets

Sean proposed estimating geometry within subsets of columns where most genes should be null, then pooling with sample-size weights. The implementation used the three replicates at each identical mixture level, making three 1-versus-2 pseudo-contrasts per level: **15 pseudo-contrasts in five blocks**.

The displayed ASH-weighted block estimates were approximately:

| Mixture level | Block \(\kappa\) |
|---|---:|
| 0% | 0.8038 |
| 25% | 0.8421 |
| 50% | 0.9068 |
| 75% | 0.9618 |
| 100% | 0.8948 |

The sample-size-weighted pooled estimate was **0.881868**. Since all five blocks had three samples, the nominal sample-size weights were equal in this experiment. The unweighted-\(Q_c\) pooled audit was much larger, approximately **1.560797**. Individual pseudo-contrast estimates and ASH null fractions varied markedly, particularly when an R1 sample was the singleton. [S07]

Using 0.881868 in target contrasts increased calls beyond the fixed-one result. For example, pure-comparison calls rose to approximately **11,295**. Its median gene-level monotonicity was essentially the same as fixed one, around **0.806** for all genes and **0.879** for the stronger stratum.

**Interpretation:** this supplied a global within-mixture residual reference, but it did not establish that the same scale transfers to every target contrast direction. It also moved in the more liberal direction relative to fixed one, which was the opposite of the conservative correction Sean wanted to investigate. It is retained as a diagnostic, not the selected operational procedure.

## AX.4 The remaining identification problem

The central distinction became:

\[
\text{variation within the fitted full model}
\quad\ne\quad
\text{the noncentral observed null-vs-full contrast increment}.
\]

When alternatives dominate the observed contrast, that increment alone cannot identify a null-noise scale without additional modeling, replicate information, or assumptions. The later residual experiments were attempts to obtain that additional information from within-group behavior rather than infer it from the group difference.

<a id="ay-welch-bin-and-plateau-heuristics"></a>
# AY. Welch-bin and plateau heuristics

## AY.1 The proposed rank-bin experiment

Sean proposed ranking genes by a simple absolute Welch statistic, dividing them into roughly equal-size percentile bins, estimating \(\kappa_c\) in each bin, and examining its relationship with effect magnitude and abundance. This was explicitly an empirical attempt to find a stable conservative procedure, not an assertion that lower-ranked genes are known nulls. [S02, S07]

The resulting binned estimates rose strongly with the Welch/effect ranking. The contrast-level association between binned \(\kappa\) and the Welch rank was approximately 0.81–1.00 in the displayed table. A pooled intermediate region near absolute Welch values around four had local estimates just above two, followed by extreme high-tail behavior.

The lowest-|t| cumulative-prefix estimates told a different story: the pooled medians rose from roughly **0.64** at the lowest 5% toward **0.97** at the lowest 50%, approximately **1.00** at 75%, and approximately **1.89** for the complete gene universe. Thus a local bin near \(\kappa\approx2\) was not the same object as a pooled estimate using every gene below the corresponding cutoff. [S07]

## AY.2 Plateau stress test

The stress test varied the ranking variable, bin count, starting percentile, run length, adjacent-change tolerance, within-run variation, and cutoff placement. It considered both absolute Welch and absolute mean differences in log-CPM.

A base rule declared a plateau for every contrast, but its selected cutoffs varied approximately **1.65 to 6.99 in absolute Welch units**, and its prefix \(\kappa\) values ranged roughly **0.664 to 0.994**. Some stability-ranked alternatives produced prefix estimates closer to 0.95–0.99, but the choice changed with binning and rule details.

**Conclusion:** the experiment supported the concern that observed signal changes the original \(\kappa\) calculation. It did not supply a stable conservative target-specific correction. The resulting prefix estimates were often below one and did not solve the operational problem. A locally flat segment was not enough to identify the quantity of interest.

<a id="az-revisiting-web-precision-and-the-full-scale"></a>
# AZ. Revisiting WEB precision and the full scale

Sean asked whether revisiting \(\kappa_F\), or using WEB rather than DEB for \(\theta_g\), could bring \(\lambda_c\) into balance.

The DEB full-scale identity remains specific to the coordinate and marginal-likelihood score equation used to estimate the DEB precision. It must not be attributed to every possible WEB formulation by definition. The Mixology WEB diagnostic nevertheless found **near-complete pooling**, with WEB weights approximately **0.969–1.000** and unweighted full-scale estimates approximately **0.984097–1.000000**. [S08]

The original \(\kappa_c\)-versus-reference association was approximately **0.8545**. With WEB precision and the fixed original ASH weighting, the corresponding relationship remained approximately **0.8303**. WEB \(\kappa_F\) itself had essentially no association with reference strength, approximately **0.0061**.

**Conclusion:** replacing DEB precision with WEB did not repair the dense-signal directional-scale problem on Mixology. The empirically near-one full scale did not provide a free balancing parameter. This reinforced the distinction between a shared residual scale already calibrated through \(\theta\) and a directional quantity estimated from a contrast increment contaminated by alternatives.

<a id="ba-counterfactual-null-experiment"></a>
# BA. Counterfactual-null experiment

## BA.1 Construction

The next diagnostic removed the fitted biological contrast while retaining the fitted nuisance/noise structure. It compared a projected-full null construction with a construction based on the already fitted observed null mean. The implementation described normalized-DNB reweighting from an NB proposal and reported importance-sampling effective sample sizes rather than silently equating DNB with ordinary NB. [S09]

For the one-df comparison, the target diagnostic was of the form

\[
\lambda_c^{CF}
=E_{CF}\left[2\theta_{Fg}(S_{0g}^{*}-S_{Fg}^{*})\right].
\]

The output gave a projected-full median \(\lambda\) of **1.001754**, range approximately **1.000009–1.004178**, median independent-half difference approximately **0.001981**, and median importance ESS fraction approximately **0.879**. The alternative null-mean construction was similarly close to one. [S09]

## BA.2 What it established and what it did not

Under the fitted parametric noise process, the expected added-direction scale was near one. The large observed target estimates were therefore not reproduced by that simulated fitted-noise calculation after the mean contrast was removed.

However, the experiment generated from the model whose residual calibration was under examination. It could not discover real misspecification that was absent from the fitted simulator. The result is best retained as a **model-implied reference check**, not independent empirical proof that real-data \(\lambda_c\) must be one.

This is why the discussion moved from parametric counterfactuals to the observed residual matrix.

<a id="bb-same-fit-residual-matrix-and-leverage"></a>
# BB. Same-fit residual matrix and leverage

## BB.1 Sean's residual-matrix proposal

Sean's proposal was to remove each gene's fitted structure, put residuals on a comparable scale across genes, and use genes as repeated information about each sample's residual behavior. Earlier responses initially confused this with refitting residual means or downweighting gene influence on shared hyperparameters. Those are different ideas; the actual experiment concerned the **standardized residual matrix itself**. [S02]

The implementation's primary residual was the working Pearson form

\[
z_{gj}
=\frac{\sqrt{\theta_g}(y_{gj}-\mu_{F,gj})}
{\sqrt{\mu_{F,gj}(1+\alpha_g\mu_{F,gj})}},
\qquad \alpha_g=1/r_g.
\]

For the two-group mean fit, working information and leverage were

\[
i_{gj}=\frac{\mu_{F,gj}}{1+\alpha_g\mu_{F,gj}},
\qquad
h_{gj}=\frac{i_{gj}}{\sum_{k\in h(j)}i_{gk}}.
\]

The HC2-style residual-square correction was

\[
z^2_{gj,HC2}=\frac{z^2_{gj}}{1-h_{gj}},
\qquad
\phi_j=\operatorname{mean}_g z^2_{gj,HC2}.
\]

These are working-model/linearization formulas. They are not exact normalized-DNB moment identities in every sparse or non-unit-precision regime. [S10]

## BB.2 Propagating sample scales through contrast information

For group information \(I_{gh}=\sum_{j\in h}i_{gj}\), the proposed sample-scale propagation was

\[
\lambda_{g,c}^{res}
=
\frac{
\sum_{j\in0}\phi_j i_{gj}/I_{g0}^{2}
+
\sum_{j\in1}\phi_j i_{gj}/I_{g1}^{2}
}{1/I_{g0}+1/I_{g1}}.
\]

The contrast summary averaged this ratio across the declared valid gene population. It avoided using \(S_0-S_F\) as the direct empirical noise input.

## BB.3 Result and the self-normalization catch

The output showed:

| Diagnostic | Result |
|---|---:|
| Mean leverage | 0.333333 |
| Median raw Pearson sample scale | 0.666231 |
| Median HC2 Pearson sample scale | 0.998898 |
| Median primary residual \(\lambda\) | 1.006183 |
| Primary residual \(\lambda\) range | 0.999105–1.036743 |
| Association with reference contrast strength | −0.0545 |

The individual sample scales were heterogeneous: for example, median \(\phi\) was approximately **2.055** for R1_000 and **1.510** for R1_025, with many R3 scales around **0.82–0.88**. Median repeated-sample CV across target comparisons was about **0.2134**. [S10]

The aggregate result near one was **not independent validation**. In a 3v3 fit there are four full residual degrees of freedom spread over six observations. The same DEB precision fit had already calibrated that residual coordinate; HC2 then approximately reversed the expected leverage loss. The chain from an average near 2/3 to an average near one was therefore partly structural.

The sample-level heterogeneity was more informative than the overall average. It did not, however, prove that shared R1/R2/R3 composition “cancels” all heteroskedasticity in a contrast; equality of group composition alone is not a variance-cancellation theorem.

## BB.4 The global 15-sample proposal was not adopted as the general operational answer

A proposed alternative was to fit one full five-mixture-level model over all 15 good samples, normalize the residual matrix once, and then obtain target-contrast scales from the relevant columns. Sean correctly objected that a general user may have only a target two-group dataset, not extra mixture levels or a global 15-sample fit.

That construction was demoted to a possible Mixology-specific diagnostic. The operational residual strategy had to work using the samples available in the target contrast itself.

<a id="bc-strict-sample-cross-fitting"></a>
# BC. Strict sample cross-fitting

## BC.1 Target-only out-of-fit residuals

The implemented cross-fit used all **nine one-sample-per-group holdout pairs** for each 3v3 contrast. Each fit trained on **2+2 samples**. Every physical sample was held out three times, once for each possible opposite-group holdout. The fitted mean for a held-out sample came from its own group's remaining replicates. [S11]

The FAST version actually run refit:

- WEB-DP \(r_g\);
- the full two-group NB means;
- Gamma-Gamma DEB \(\theta_g\) using **raw full NB half-deviance**.

It did **not** refit the full finalized \(C'(1)\)-corrected M03 precision procedure within each fold. A frozen-nuisance sensitivity branch kept the original target \(r_g,\theta_g\) and cross-fit only the means.

The saved six-sample normalization/exposure offsets were treated as fixed. Consequently “strict” described the held-out mean/dispersion/precision fitting, **not complete independence of the entire preprocessing pipeline from the held-out sample**.

## BC.2 Prediction uncertainty was included

The working predictive variance was

\[
V_{gj}^{pred}
=
\frac{
\mu_{gj}^{pred}(1+\alpha_g\mu_{gj}^{pred})
+(\mu_{gj}^{pred})^2/I_{g,h}^{train}
}{\theta_g^{train}}.
\]

Then

\[
z_{gj}^{2,CF}
=\frac{(y_{gj}-\mu_{gj}^{pred})^2}{V_{gj}^{pred}}.
\]

The second numerator term approximates uncertainty in the fitted group mean. With equal information and two training replicates, the familiar working inflation is about 1.5. The displayed median prediction inflation was indeed near **1.493–1.500**.

For each sample-gene combination, the implementation averaged the three held-out squared residual evaluations, requiring finite values in all three, and then pooled across genes. Those three evaluations are dependent; they must not be treated as three independent residual replicates.

## BC.3 The ordinary mean exploded

The primary mean-square residual \(\lambda\) ranged from approximately **1.403 to 2.19 billion**. Pure-endpoint contrasts dominated the failures, while interior-only comparisons remained in the vicinity of 1.4–1.7. The frozen-nuisance branch also exploded, pointing away from nuisance refitting as the sole cause.

The median-scale branch remained finite, approximately **1.04–1.38** across contrasts, but that alone was not a calibration theorem. The later audit was required to identify why the mean had failed.

<a id="bd-boundary-audit-and-robust-residual-paths"></a>
# BD. Boundary audit and robust residual paths

## BD.1 The numerical explosion had an identifiable count-level cause

The top offending sample-gene observations had **both same-group training counts equal to zero**, while the held-out count was often only **3 or 4**. Their exposure-adjusted unregularized training mean was zero; the implementation's tiny numerical mean floor produced a nearly zero plug-in predictive variance. The squared residual then became of order **\(10^{12}\)–\(10^{13}\)**. [S12]

The same gene's original six-sample fitted mean could be around one. This was not evidence of billions-fold biological noise. It was a failure of the zero-boundary plug-in prediction and ordinary squared-residual mean.

Examples in the displayed top table included gene IDs 84557, 84072, 56925, 3948, 2793, and 8406, repeatedly appearing with zero training totals and small positive held-out counts.

## BD.2 Interpreting the boundary percentages correctly

The contrast table reported approximately **100%** boundary contribution for `100_vs_000` and **50%** for many one-endpoint comparisons. These were averages of **per-sample contribution fractions**, not necessarily the fraction of the pooled six-sample total residual sum. For a one-endpoint comparison, three affected samples and three unaffected samples naturally produce a mean near 50% even if the affected samples dominate the pooled total.

The worst sample/contrast cases had approximately all of their squared-residual sum in the top 0.1% of genes. The diagnostic therefore identified a tiny set of catastrophic contributions, not merely a moderate heavy tail. [S12]

## BD.3 Robustification paths

The audit applied upper trimming and upper winsorization sample by sample after averaging each sample-gene's three cross-fit evaluations. It used tail percentages from zero through 10%, with fine attention to the very small upper tail.

The analytic reference factors used \(X\sim\chi_1^2\). If \(c=F^{-1}_{\chi_1^2}(1-p)\), they were

\[
a_{trim}(p)=\frac{F_{\chi_3^2}(c)}{1-p},
\qquad
a_{winsor}(p)=F_{\chi_3^2}(c)+cp.
\]

The median reference was \(\operatorname{median}(\chi_1^2)\).

These factors make the corresponding idealized chi-square functionals comparable; they do **not** prove the actual cross-fitted count residuals have a chi-square distribution. In particular, the average of three dependent squared cross-fit residuals is not automatically \(\chi_1^2\). The later simulation was designed to measure the combined procedure's reference response rather than assume that equivalence.

| Upper-tail percentage | Median trim \(\lambda\) | Median winsor \(\lambda\) |
|---|---:|---:|
| 0.25% | 1.5083 | 1.5588 |
| 0.50% | 1.4687 | 1.5273 |
| 1% | 1.4213 | 1.4899 |
| 2% | 1.3674 | 1.4353 |
| 5% | 1.2996 | 1.3565 |
| 10% | 1.2488 | 1.2983 |
| Median functional | 1.1542 | Not applicable |

The catastrophic tail was no longer driving the aggregate estimates by roughly **0.25%** in the displayed audit. The reference-effect correlation fell toward approximately **0.20** at stronger tail control; the median functional gave approximately **0.2121**. [S12]

## BD.4 What succeeded and what remained unresolved

The audit **succeeded** in diagnosing the boundary failure and in producing finite, interpretable residual-scale paths. It **did not** identify a unique robustification fraction: the curves continued declining smoothly from about 1.5 toward 1.25, with the median functional lower still. There was no clear universal plateau that determined the true \(\lambda_c\).

The fact that an estimate exceeded one was not itself a reason to choose it. Conservatism had to be evaluated through the downstream model and appropriate benchmark, not selected by the desired direction of the answer.

<a id="be-known-unit-scale-nb-calibration"></a>
# BE. Known-unit-scale NB calibration

## BE.1 What was simulated

The final completed calibration used **10 simulations per contrast**, reduced from 20 at Sean's request because of runtime. With ten contrasts and nine folds per simulated dataset, the requested workload was **100 simulated 3v3 datasets and 900 strict 2+2 fold fits**. [S13]

The generator was

\[
Y^*_{gj}\sim NB(\mu_{F,gj},r_g),
\]

using the saved full-model means, fitted \(r_g\), and fixed exposure configuration. It retained the fitted biological differences. It was therefore a **unit-extra-noise model-reference simulation with alternatives retained**, not an all-genes-null experiment.

The generator set the ordinary-NB reference \(\theta=1\). It did not simulate the exact normalized-DNB family at the original fitted \(\theta_g\). The estimator refit the same FAST cross-fit nuisance procedure within every simulation and applied the same robust path and fixed target information geometry used in the real-data audit.

For contrast \(c\) and functional/tail setting \(e\), the implemented comparison was

\[
b_{ce}=\operatorname{mean}_{sim}\widehat\lambda_{ce}^{*},
\qquad
\widehat\lambda_{ce}^{cal}
=\widehat\lambda_{ce}^{obs}/b_{ce}.
\]

## BE.2 Calibration reduced but did not erase the real-data reference discrepancy

## BE.3 Interpretation limits

The data supported a robust residual-scale discrepancy relative to the **specific fitted-NB reference experiment**. They did not prove that the remaining 14% median excess was entirely genuine unmodeled observation noise, nor that a scalar of that size was optimal for the finalized M03 test.

Possible distinctions still mattered: the ordinary-NB generator versus fitted DNB precision, approximate predictive variance, the FAST raw-deviance fold fit, gene/sample dependence, sparse boundary behavior, and finite-sample nuisance fitting. The later direct test was needed precisely because an empirical scale estimate can look plausible while not improve inference over fixed one.

The displayed empirical exceedance probability was **0.0909** for every median-functional comparison. With ten simulations and the implemented plus-one calculation, this is the minimum possible value \(1/(10+1)\), indicating no simulated value exceeded its observed counterpart. It is not a high-resolution tail probability or a fully calibrated composite-null significance test. The small simulated SDs describe Monte Carlo variability under that generator, not uncertainty in the generator's correctness. [S13]

<a id="bf-observed-fc-spearman-plateau-experiments"></a>
# BF. Observed-FC Spearman plateau experiments

## BF.1 Two different proposals were tested

**Sean's across-contrast proposal:** replace the unavailable reference effects in operational selection with observed fitted fold changes. At each robustification percentage \(p\), calculate

\[
\rho_{across}(p)
=\operatorname{Spearman}_c\left(
\widehat\lambda_c^{cal}(p),
\operatorname{median}_g|\widehat{\log_2FC}_{gc}|
\right).
\]

This uses multiple contrasts, not one isolated comparison.

**The assistant's within-contrast proposal:** form a gene-level weighted residual score from the robustified sample residuals and calculate

\[
\rho_c(p)
=\operatorname{Spearman}_g\left(
q_{gc}(p),|\widehat{\log_2FC}_{gc}|
\right).
\]

For trimming, the implementation retained a gene in this correlation only when it survived the sample-wise trimming in all six samples. Winsorization kept finite genes while capping residual values. Thus the two within-gene paths did not have identical gene populations. [S14]

## BF.2 The first-flat-region rule did not identify the intended transition

The rule searched from **0.25%**, required three consecutive points, and declared them flat when the correlation range was at most **0.05**. Tolerances 0.025–0.10 were also examined.

Across contrasts, observed-FC correlations remained around **0.56–0.67** in the useful region. The primary rule selected approximately **1% trimming** and **0.25% winsorization**.

Within contrasts, every comparison and both robust estimators selected **0.25%**, the first allowed point. Yet the trimming curves visibly kept decreasing through 10%, while winsorization changed the gene-level correlations much less. [S14]

**Conclusion:** the detector was mostly finding a locally smooth part of the curve at its imposed lower bound. Flatness did not establish independence or identify the desired amount of robustification.

## BF.3 A flat rank correlation is especially easy to overinterpret

The across-contrast calculation involved only ten overlapping contrasts. If their ordering does not change, Spearman can remain exactly constant while the absolute \(\lambda\) values continue moving. A flat \(\rho\) therefore need not signal that all biologically related contamination has disappeared.

Likewise, capping a tail can markedly reduce a mean scale while preserving much of the ordering of individual gene scores. The relatively flat winsorized within-gene rank-correlation curves are not, by themselves, evidence that winsorization failed to control extreme residual magnitude.

The observed fold changes and residual scores also come from related data. A nonzero correlation is not automatically biological contamination; a small one is not proof of valid inference. These are diagnostic objectives rather than ground-truth definitions of noise.

<a id="bg-spline-abc-optimization"></a>
# BG. Spline A/B/C optimization

## BG.1 Sean recalled a squared-derivative optimization idea

The conversation moved from threshold-based plateau detection to a smooth objective. The implementation fit a cubic smoothing spline using SciPy's GCV-selected smoothing and examined three selectors:

\[
\text{A: }\min J_A(u)=\rho(u)^2+\rho'(u)^2,
\]

\[
\text{B: }\max K(u)=\frac{|\rho''(u)|}{(1+\rho'(u)^2)^{3/2}},
\]

\[
\text{C: }\min J_C(u)=\rho(u)^2+\rho'(u)^2+\rho''(u)^2.
\]

The symbol \(u\) matters: derivatives in the actual code were with respect to a normalized coordinate, not automatically physical tail percentage or physical \(\lambda\). Unit derivative weights were choices of the experiment, not derived constants. [S15]

## BG.2 Tail-percentage optimization was sensitive to the coordinate

The first implementation used log10 tail percentage normalized to [0,1], with linear tail percentage as a sensitivity analysis. After excluding the pathological low-tail region, only about six observed points remained: **0.25%, 0.5%, 1%, 2%, 5%, 10%**. A dense 4,001-point evaluation grid did not create additional observed information.

The selected tail percentages changed substantially between log and linear coordinates. Trimming often produced interior A/C optima; winsorized A/C frequently chose the maximum tested robustification of 10%. The calibrated values at selected points sometimes clustered around 1.23–1.27, but individual-contrast results were more variable.

The second-derivative/curvature plots were jagged. The initial claim that B had become the strongest selector was subsequently withdrawn. Sparse knots, derivative sensitivity, and the smoothing choice were all potential contributors.

## BG.3 Why “B has zero boundary optima” was not independent evidence

The code prohibited B from searching the outer **5% of each normalized domain**. Its boundary flags, however, tested the full domain endpoints. Therefore B's reported zero fraction on the global boundary was partly guaranteed by the search restriction. It could not fairly be used as a standalone advantage over unrestricted A/C. [S15, S19]

## BG.4 Current status of the three selectors

Sean judged A the best current candidate and C an extension with no demonstrated benefit. The subsequent agreed status was:

- **A:** primary empirical candidate;
- **B:** elbow diagnostic only until smoothing/point-removal stability is demonstrated;
- **C:** demoted because the extra second-derivative term had no demonstrated payoff.

A smoothing-strength × leave-one-point-out stress test was proposed, but Sean chose to test A directly on Mixology first. That robustness experiment has **not** been completed in the supplied record.

<a id="bh-optimization-in-calibrated-lambda-space"></a>
# BH. Optimization in calibrated-lambda space

## BH.1 The experiment removed tail percentage from the nominal optimization axis

For each robustification setting, the existing data supplied both \(\widehat\lambda^{cal}(p)\) and \(\rho(p)\). The new experiment plotted \(\rho\) against the resulting calibrated \(\lambda\) and applied A/B/C to that curve. It retained both signal-dependence constructions and both robust estimators. Continuous optima were mapped to the nearest actually tested \(\lambda\) point; the associated tail percentage was reported as provenance. [S15]

Across contrasts, the horizontal coordinate was the median calibrated \(\lambda_c\) at a given tail setting. Within a single contrast, it was that contrast's calibrated \(\lambda_c\). The “median within-gene curve” paired the median \(\lambda\) and median within-gene \(\rho\) across contrasts; it was a summary path, not the result for a single fitted contrast.

## BH.2 Results

For the across-contrast curve, continuous optima were:

| Robust path | A basin | B curvature | C curvature basin |
|---|---:|---:|---:|
| Trim | 1.2442 | 1.2668 | 1.2318 |
| Winsor | 1.2294 | 1.2603 | 1.2294 |

The winsorized A/C values were at the left edge of the available \(\lambda\) range, corresponding to the highest tested winsorization.

For the median within-gene curve:

| Robust path | A continuous \(\lambda\) | B continuous \(\lambda\) | C continuous \(\lambda\) |
|---|---:|---:|---:|
| Trim | 1.2572 | 1.2299 | 1.3406 |
| Winsor | 1.2294 | 1.2603 | 1.2294 |

The individually selected A-trim values had median approximately **1.242483**, range approximately **1.048722–1.553080**, and CV approximately **0.1480**. Individual A-winsor values had median approximately **1.229353**, range approximately **1.065023–1.467867**, and CV approximately **0.1273**. About 30% of individual A-trim and 90% of individual A-winsor continuous optima were on a full-domain boundary. [S15]

## BH.3 Trim/winsor path agreement was strong at one aggregation level, not universal

On the common observed \(\lambda\) range, linear-interpolation comparisons gave:

| Signal-dependence construction | Trim/winsor \(\rho\) RMSE | Maximum absolute difference |
|---|---:|---:|
| Across contrasts | **0.009868** | **0.022051** |
| Median within-gene curve | **0.047029** | **0.077667** |

The across-contrast paths were close. The within-gene paths did not collapse nearly as well; trimming progressively changed the retained gene population and winsorization did not.

This supports a useful empirical similarity of some paths, not proof that the true noise parameter is uniquely identified or that all routes must converge to 1.25.

## BH.4 A coordinate issue still remains

The code normalized each curve's own observed \(\lambda\) interval to [0,1]. For interval width \(R\),

\[
\frac{d\rho}{du}=R\frac{d\rho}{d\lambda},
\qquad
\frac{d^2\rho}{du^2}=R^2\frac{d^2\rho}{d\lambda^2}.
\]

Consequently the actual A objective corresponds to \(\rho^2+R^2\rho_\lambda'^2\), and C additionally weights physical second derivative squared by \(R^4\). Changing the tested interval or using different trim/winsor intervals can therefore change the objective. The earlier phrase “moving to lambda space eliminates parameterization dependence” was too strong. [S15, S19]

The relevant result is narrower: A produced operational candidates worth testing, and the next experiment actually inserted those candidates into M03.

The active scientific comparison is therefore **fixed one versus residual A**, not “A has solved dense-signal calibration.” A is a credible candidate to investigate, while its benefits over the simpler baseline remain unestablished.

<a id="bj-oracle-lambda-search"></a>
# BJ. Oracle-lambda search

## BJ.1 Sean's next question

Sean proposed finding the best \(\lambda\) for each mixture comparison so that the project could see the attainable target and assess what an operational estimator should aim for.

The implemented primary objective is

\[
S_c(\lambda)
=\operatorname{Spearman}_g\left(
|\log_2FC^{ref}_{gc}|,
-\log_{10}p_{gc}(\lambda)
\right).
\]

It is calculated **across genes within one contrast**. The primary selected value maximizes the all-gene score. The strong-reference score at \(|\log_2FC^{ref}|\ge0.5\) is a secondary reported objective and tie-breaker. [S17]

## BJ.2 Implemented search and stored outputs

The supplied runner uses:

| Component | Setting |
|---|---|
| Initial grid | 0.50–2.00 by 0.05 |
| Exact comparator values | 1, that contrast's A-trim, and A-winsor |
| Boundary expansion | Blocks of 0.25, within hard bounds 0.10–4.00 |
| Fine search | Approximately ±0.10 by 0.005 |
| Fine recentering | Up to four rounds |
| Near-optimal tolerance | Score within 0.002 of the evaluated maximum |
| Reused state | Frozen original Stage 01/02 |
| Candidate state | One working downstream checkpoint per contrast, cleared/switched when \(\lambda\) changes |
| Durable result | Per-\(\lambda\) metric JSON; complete gene results at the final selected value |

The planned summary compares the selected oracle with fixed one and both A variants, including absolute \(\lambda\) error and score regret. The final oracle vector is also to be scored on cross-contrast monotonicity, but that metric is **not what the per-contrast optimization maximizes**.

## BJ.3 What “oracle” does and does not mean here

The intended oracle is a **reference-rank oracle over the tested range and implementation**, not an omniscient gene-level FDR or power oracle. Genes differ in variance, abundance, and information, so ideal significance need not perfectly rank absolute effects across genes.

Likewise, choosing each contrast to maximize its across-gene score does not mathematically maximize the median across-contrast gene trajectory correlation. The latter is a different, coupled objective. The oracle-vector diagnostics are therefore descriptive follow-up, not a proven ceiling on the cross-contrast metric.

The near-optimal minimum and maximum in the current code form the envelope of evaluated qualifying points. They need not prove every intermediate value is near-optimal or that the set is connected. No local-grid search establishes a global optimum outside its explored region.

## BJ.4 Status and runtime

The latest displayed log was for the first contrast, `100_vs_000`, with 33 coarse candidates. The \(\lambda=0.5\) evaluation ran approximately from **7:39:05 PM to 7:40:39 PM**, or **94 seconds**. [S17]

The earlier conversational estimate of roughly 740 evaluations and 19.3 hours added 33 coarse and 41 fine points without subtracting their overlap. In a simple case, about five of the 41 fine points may already have been evaluated on the coarse grid, while exporting the final oracle can require an additional fit. Boundary expansions, exact fallbacks, recentering, and cache behavior can dominate those small counting differences. The **19–24 hour estimate remains only a rough extrapolation from one candidate**, not a guarantee.

**No completed optimum, final curve, oracle-vector score, or recovery/regret table was present at the evidence cutoff.** The correct current status is “search launched; results pending.”

## BJ.5 New static code-review flags identified while assembling this summary

These are separate from the previously reported runtime results; no corrected rerun is being claimed.

The supplied oracle file includes `contrast` in every per-candidate metric payload, then later copies the resulting curve and calls `curve.insert(0, "contrast", contrast_name)` again. That would raise a duplicate-column error when the first completed contrast reaches this aggregation step. The per-candidate metrics and saved contrast artifacts preceding that step are designed to remain on disk, so the issue is a reporting/aggregation defect rather than evidence that those candidate evaluations failed. [S19]

A second resume concern is that `geometry_override_label` is assigned inside the newly computed Stage-03 branch and later used in final diagnostics. A partial resume that loads Stage 03 but has not yet reached a cached final result needs to restore this diagnostic label as well. The summary documents the issue; it does not silently amend the supplied runner.

These checks should precede interpreting a complete oracle report. Syntax checking alone did not exercise either path.

## BK.3 Promising results that have survived a direct check

**Fixed one:** on Mixology, disabling the problematic observed directional estimator substantially improved gene trajectory monotonicity and reduced inversions. It remains the strongest simple M03 baseline in the displayed direct comparison.

**Robust cross-fitted residual paths:** the procedure extracted finite, heterogeneous residual-scale information once a tiny catastrophic tail was bounded. The completed NB-reference calibration did not erase the observed differences.

**A-basin with residual scales:** A-trim and A-winsor both substantially improved original M03 when their values were inserted into the actual downstream test. The effect was not merely cosmetic agreement on a diagnostic plot. However, fixed one still had better all-gene and strong-gene rank-monotonicity scores, so an incremental benefit of the extra machinery remains unproven.

**MAQC:** the completed archived M03 procedure retained strong external qPCR-reference operating points, coverage, MCC, and AUC. Mixology's later failure narrows the generality of the scale-estimation claim; it does not retroactively change those saved benchmark results.

**RNAontheBENCH:** the method remained competitive enough to yield informative comparisons, but the displayed matched-error experiments did not establish an edgeR advantage for M03. That negative result is part of the evidence, not something to omit. [S01, S03, S07, S12, S13, S16]

<a id="bl-corrections-to-earlier-interpretations"></a>
# BL. Corrections to earlier interpretations

This section makes explicit the changes in interpretation required by the accumulated record. It does not silently replace the historical narrative.

## BL.1 “Noise scale above one” is not automatically “we have identified the correct \(\lambda_c\)”

The robust residual functional measures out-of-fit disagreement relative to a fitted working predictive variance and a specified reference calibration. The final M03 directional correction acts on a geometry-adjusted null residual statistic. A theoretical or empirical bridge between these quantities still has to earn its validity.

The direct A comparison was therefore essential: it showed substantial improvement over original M03 but no overall advantage over fixed one on the reported rank-monotonicity measures.

## BL.2 The NB reference calibration was not exact fitted-DNB validation

The completed generator used the original full means and \(r_g\), but ordinary NB sampling with \(\theta=1\). The earlier prose “data where the fitted DNB model is exactly correct” does not describe the final completed FAST calibration experiment. That difference matters because the original precision estimates could differ materially from one.

The defensible statement is that robust real-data residual scales exceeded the response of the specified NB-based simulation/estimation pipeline. The experiment did not certify a unique true excess-noise percentage under every admissible DNB nuisance configuration.

## BL.3 Small correlation with reference effect is encouraging, not an independence certificate

The diagnostic had ten overlapping mixture comparisons, and correlation is only one aspect of a joint distribution. Correlation with observed fold change also differs from correlation with the fitmixture reference. The later observed-FC across-contrast correlations around 0.56–0.67 do not contradict the smaller reference-effect correlation numerically; they are different inputs.

The optimization path itself also changes gene inclusion and residual ranking. A low or flat Spearman should not be treated as the definition of a correctly calibrated noise estimator.

## BL.4 The within-gene score has its own selection effects

The trim implementation kept only genes surviving all six sample trims for the within-gene correlation. Thus its decreasing correlation reflects both bounded residual influence and a changing analysis population. Winsorization retained more of the original ranking and gene set. The difference between their \(\rho(\lambda)\) paths is therefore not surprising and is not evidence that one path must be the unique correct decomposition.

A further source-level detail is that the within-gene trim transformation retained all values at or below its cutoff, whereas the earlier scale audit dropped an exact number of sorted values. Ties at the cutoff can therefore change exact membership. This is an implementation distinction to record, not a demonstrated explanation for the entire observed result. [S14, S19]

## BL.5 The optimization did not derive its own statistical loss weights

The derivative coefficients were fixed at one in a normalized coordinate. Moving from tail percentage to calibrated \(\lambda\) changed the analysis usefully, but did not eliminate choices about scale, range, smoothing, or derivative weights.

Maximizing curvature additionally excluded boundary regions by design. Promoting B because it had no endpoint selections was not warranted. Sean's observation of jagged B curves correctly led to demoting the apparent curvature optimum.

## BL.6 Reproducibility anomalies are not false-positive labels

A gene with smaller reference effect can be more significant in a different comparison if the effective noise/information differs. “New plus weaker” counts and cross-contrast inversions are useful diagnostics in the controlled mixture experiment, but they are not known type-I errors and do not estimate literal FDR.

The pure reference call set also changes across method variants. Aggregate new-versus-pure counts must not be explained as though the pure set were fixed across all methods.

## BL.7 Two numerical corrections

The calibrated median-functional value **1.1411** is the median of contrast-specific ratios, not exactly \(1.1542/1.0135\).

The direct A-winsor selected-scale/reference-effect correlation shown in the source table is **+0.20**, not −0.20. [S13, S16]

## BL.8 “Oracle” needs an objective qualifier

The running oracle maximizes an across-gene effect/significance rank score for each contrast separately. It is not a known FDR-power optimum, and it does not optimize the cross-contrast trajectory metric jointly. Calling its final vector a universal “ceiling” would overstate the objective actually implemented.

A near-flat oracle score curve could make a numerical \(\lambda\) error immaterial. This is why **performance regret and near-optimal regions** are more informative than insisting on precise recovery of one point estimate.

<a id="bm-current-architecture-and-next-steps"></a>
# BM. Current architecture and next steps

## BM.1 Preserve the supported backbone, explicitly reopen the scale-estimation layer

The present two-group research architecture is best expressed as:

\[
\boxed{
\text{WEB-DP curvature}
+\text{DEB ordinary precision}
+\text{normalizer admissibility handling}
+\text{support-feasible null geometry}
+\text{an explicitly declared }\lambda_c\text{ policy}.
}
\]

The \(\lambda_c\) policy is now an experimental branch, not a universally settled default:

- original ASH-weighted observed increment, retained as a baseline;
- fixed \(\lambda_c=1\), retained as the main simple comparator;
- residual A-trim/A-winsor, retained as active candidates;
- the pending reference-rank oracle, evaluation-only.

The original MAQC results used their recorded high geometry values, approximately 6.41 and 7.89. A Mixology discovery that fixed one is effective does not authorize silently replacing those values or rewriting the MAQC results. A future fixed-one/residual-A MAQC comparison would be a new ablation requiring its own saved output.

## BM.2 Immediate sequence

**First, complete the oracle experiment on its stated terms.** Address the recorded reporting/resume defects without discarding completed candidate metrics, verify that each candidate truly used the requested scale, and distinguish numerical failures from objective behavior. No final oracle conclusion should be anticipated from the start log.

**Second, inspect the curves, not just the maximizers.** Determine whether the score changes enough to matter, whether maxima lie at search bounds, whether all-gene and strong-gene objectives agree, and whether near-optimal regions are broad or disconnected. Compare fixed one and A through regret as well as \(\lambda\) distance.

**Third, compare the resulting oracle vector on the previously used trajectory diagnostics.** An improvement in the per-contrast objective may or may not improve across-contrast monotonicity. Keep those questions separate.

**Fourth, use the oracle result to decide the next development target.** A broad near-optimal region including one would weaken the case for expensive adaptive scale estimation. Reproducible, consequential contrast-specific optima would justify trying to recover them operationally. An oracle that fails to improve the relevant downstream behavior would point away from \(\lambda\) alone as the remaining bottleneck.

**Fifth, return to the proposed stability/calibration tests only where justified.** The A smoothing-strength/leave-one-point-out audit, exact finalized fold-precision sensitivity, and fitted-DNB generator comparison remain possible next tests; none should be described as already completed.

## BM.3 Independent validation remains necessary

Mixology has been repeatedly used for development: dense-signal diagnosis, residual construction, robustification, reference calibration, objective selection, and now oracle search. The fact that reference labels were loaded after individual fits prevented one kind of direct fitting leakage, but does not make this entire adaptive development history an untouched external validation.

After a method is frozen, the original validation principles still apply: complete-null behavior, matched achieved-error power, real biological heterogeneity, competitor-owned simulation frameworks, and appropriate current competitor configurations.

The previously planned Love, Zhou, Burden-style, historical Pickrell, broader SEQC, and Quartet work remains in the program where not completed. Bottomly is the explicit exception: it has been removed from the active evidence set. The biological-sample-size program is in progress and should supply its actual completed summaries before stronger conclusions are drawn.

## BM.4 DEU/DTU and single-cell scope

The fixed-full-design DEU/DTU family results and the single-cell/single-nucleus roadmap in Part I remain preserved. The current two-group Mixology tests do not demonstrate that the general-X family estimator failed on Tasic, nor that single-cell plans have been validated.

They do motivate a broader diagnostic question for any future family-scale analysis: how much usable null information is available, how much of the residual scale is already imposed by precision fitting, and whether empirical directional estimation remains identifiable in dense-signal families. This is a question to test, not a retroactive replacement of the Tasic results.

Donor-level replication and the distinction between cells, nuclei, biological samples, and technical replication remain essential. No single-cell performance conclusion is added merely because the residual matrix has many genes.

The completed calibration has **ten** replicates per contrast, regardless of earlier 20-replicate script comments. The actual selected direct-test values are the saved `selected_tested_lambda` values, not necessarily the nearby continuous spline optima.

## Final research position at this cutoff

The project has a substantive and extensively audited count-model architecture, strong specific MAQC results, an informative but non-dominant RNAontheBENCH result, a discarded Bottomly proxy-truth line, and a dense-signal Mixology challenge that exposed a failure of the original observed directional-scale estimator.

The residual program has made useful progress: it separated self-normalization from out-of-fit prediction, identified a concrete low-count boundary failure, created robust scale paths, measured their behavior under a declared NB reference, and tested the selected scale inside M03 itself. It has **not** yet established that the added adaptive machinery improves on fixed \(\lambda=1\) overall or that the optimized residual functional identifies the scientifically correct directional scale.

The next result that can materially change that conclusion is the **completed, implementation-checked oracle experiment**, interpreted under its actual reference-rank objective and compared with both the simpler baseline and the operational A selections. Until then, the oracle target, its regret curves, and the attainable improvement remain unknown.

<a id="bo-general-x-scale-identification"></a>
# BO. General-`X` scale identification after Mixology

## BO.1 The theoretical object was separated from its empirical estimator

Sean clarified the target distinction that had become blurred during the dense-signal Mixology work:

\[
T_c\mid H_0 \sim \kappa_c\chi_1^2
\]

defines a **contrast-direction null-law scale**. That object exists conceptually independently of the realized effect-size distribution in one real dataset. Estimating it from observed data is a separate identification problem.

The residual-space decomposition remained

\[
\kappa_c=d_0\kappa_0-d_F\kappa_F,
\]

but the September 17–18 work asked whether sample residuals, leverage correction, or exact leave-one-out calculations could recover the same directional scale without relying on a sufficiently rich null population.

## BO.2 Exact leave-one-out and HC2 identified the full residual scale, not the contrast direction

The near-null Mixology diagnostics gave approximately:

```text
near-null exact LOO jackknife:   0.83–0.88
near-null kappa_F:               ~0.84
HC2-style residual scale:        ~0.82
near-null kappa_c:               ~1.3–1.5
HC3:                             ~1.529
ordinary-NB / DNB controls:      ~1
```

The tempting numerical agreement between HC3 and the near-null `kappa_c` was rejected as accidental. Exact LOO and HC2 were estimating the residual space already represented by the fitted full model.

The fixed-design nonidentification argument made the limitation sharper. If the cross-fitted residual transform is built from the full-model residual/projector geometry, it can identify objects living in the full residual subspace while annihilating the tested direction. One can construct positive-definite covariance structures with the same observable full-residual behavior and the same `kappa_F≈0.84`, yet materially different directional scales, for example `kappa_c≈0.84` versus `kappa_c≈1.50`.

Thus:

\[
\boxed{
\text{reference-free residual moments do not generally identify }u_c^\top R u_c.
}
\]

This explains why making the residual estimator progressively more exact did not solve the directional-scale problem.

## BO.3 Positive-control / coefficient-noise ideas remained exploratory

Sean also explored the opposite of negative controls: using strong/top genes as “positive controls” for the fitted contrast direction, removing that direction from other genes, and asking whether the remaining residuals could identify directional noise. A related framing was

\[
\widehat b_g=b_g+\varepsilon_{cg},
\]

with the idea of treating fold change as the latent signal and the fitted-minus-reference difference as contrast noise.

These were useful conceptual probes, especially when an independent fitmixture reference was available in Mixology. They did not produce a general reference-free estimator that survived the identification argument. The project therefore did not promote them into the production architecture.

## BO.4 Practical policy for alternative-dense designs

The practical September-18 conclusion was:

```text
Use ASH directional calibration when there is enough stable null support.

When null and alternative contributions are not separable enough for ASH
to identify the directional scale reliably:
    use lambda_c = 1 as an explicit no-additional-correction fallback;
    retain the ASH estimate as a diagnostic;
    issue a warning that directional calibration is unresolved.
```

No universal “X% alternatives” threshold was accepted. The warning should be based on the stability/amount of usable null information and the experimental design rather than pretending there is a universal density cutoff.

Mixology 100-vs-0 is an unusually alternative-dense stress test. Treating that regime as a scope limitation is defensible because most DGE experiments are not designed so that nearly the entire transcriptome should shift. However, the Mixology failure remains scientifically important and must be reported rather than hidden. A design-informed option to disable observed ASH directional correction is preferable to claiming universal reliability.

This update sharpens, but does not erase, the September-16 result: fixed `lambda=1` remains the strongest simple baseline in the dense-signal Mixology comparison, while the original ASH-based directional estimator remains useful in ordinary null-rich settings and in the existing M03 benchmark lineage. [S20]

# BP. Li2020 biological-null benchmark: historical competitor reproduction

## BP.1 Frozen benchmark design

The Li2020 within-population null program is now fully specified for the project as:

```text
datasets:          AdLC, OC, TNBC
group sizes:       5, 10, 15, 20, 25, 30, 35, 40 per group
replicates:        50 frozen random within-population splits per n
gene universe:     full supplied filterNA0.5 matrix
call rule:         finite BH q <= 0.05 AND finite |log2FC| >= 1
FPR:               called genes / fixed filtered gene universe
```

The 50 repeats in this within-null arm are a **project-frozen extension**; the original paper did not state that repeat count for this specific arm.

The historical/current-package reconstruction includes `ExposureOnly`, `RLE`, `TMM`, `UQ`, and `UQ-pgQ2` normalization paths with edgeR exact, edgeR QL, and DESeq2 Wald. The primary M03 comparison uses canonical TMM.

The full historical-logic run contains

\[
3\times8\times50\times5\times3=18{,}000
\]

method/normalization cells. The benchmark runner saves cell-level results as well as the usual summaries, which made it possible to inspect the empirical distribution over the 50 frozen splits rather than relying only on means. [S21]

## BP.2 Reproduced sample-size pattern

For TMM, the endpoint mean FPRs reproduced the striking sample-size behavior:

| Dataset | Method | Mean FPR 5v5 | Mean FPR 40v40 |
|---|---|---:|---:|
| AdLC | edgeR QL | 0.000359 | 0.039941 |
| AdLC | edgeR exact | 0.004490 | 0.034713 |
| AdLC | DESeq2 Wald | 0.005110 | 0.004407 |
| OC | edgeR QL | 0.000018 | 0.014915 |
| OC | edgeR exact | 0.001890 | 0.011874 |

| TNBC | DESeq2 Wald | 0.007500 | 0.003519 |

Thus edgeR exact and QL increase strongly with sample size in these within-population splits, whereas DESeq2 Wald does not show the same large-`n` inflation.

This is an **empirical property of this benchmark construction**, not a universal edgeR characterization. Sean emphasized that edgeR had been more conservative than DESeq2 in many other project settings. That broader record makes the Li2020 behavior especially important to diagnose mechanistically rather than simply label as an intrinsic edgeR defect.

## BP.3 The empirical FPR distributions changed the interpretation

The project then extracted all 50 replicate FPRs for each dataset, sample size, and method and constructed:

```text
raw 50-replicate FPR tables;
sorted empirical CDFs;
literal empirical PMFs over FP count/FPR;
quantile/tail summaries;
paired method-by-frozen-split matrices;
pairwise method correlations by dataset and n.
```

The 40v40 distributions showed that the edgeR rise is not merely a mean driven by a few catastrophic splits. The distribution itself is broadly shifted.

Screenshot-transcribed 40v40 summaries were approximately:

| Dataset | Method | Mean | Q25 | Median | Q75 | Q97.5 |
|---|---|---:|---:|---:|---:|---:|
| AdLC | edgeR exact | 3.471% | 1.926% | 2.694% | 4.234% | 8.46% |
| AdLC | edgeR QL | 3.994% | 2.389% | 3.071% | 4.708% | 8.75% |
| AdLC | DESeq2 Wald | 0.441% | 0.109% | 0.197% | 0.476% | 2.53% |
| OC | edgeR exact | 1.187% | 1.031% | 1.202% | 1.299% | 1.65% |
| OC | edgeR QL | 1.492% | 1.339% | 1.471% | 1.657% | 2.04% |
| OC | DESeq2 Wald | 0.143% | 0.083% | 0.126% | 0.187% | 0.353% |
| TNBC | edgeR exact | 4.402% | 2.751% | 3.897% | 6.164% | 8.31% |

The OC edgeR case is especially informative: mean and median are almost identical, so the inflated mean is not a rare-tail artifact.

## BP.4 DESeq2 also revealed nonexchangeable sample structure

DESeq2 did not show edgeR's broad large-`n` shift, but its small-sample empirical distributions had substantial right tails. The worst displayed DESeq2 mean across dataset/sample-size cells was approximately **0.9311% for AdLC 5v5**, while the worst individual frozen split was approximately **7.1332% for AdLC 10v10**.

That pattern suggests that the nominal “same cancer / same population” null is not a perfectly homogeneous exchangeable null. Some random partitions can evidently create much stronger expression imbalance than others.

Sean's current interpretation is that **sample structure is primary** and methods differ in how they respond to it. Plausible mechanisms to test include latent subtype, batch, tumor purity, sex, stage, ancestry, treatment history, RNA quality, or other sample-level axes. Under that interpretation, an edgeR discovery in a random split need not be a statistical false positive if the split accidentally separates real latent biology.

This remains a hypothesis to test, not a completed sample-level causal audit.

## BP.5 Next decisive sample-structure analysis

The next direct diagnostic should preserve frozen-split identity rather than sorting the 50 FPRs independently. For every dataset and sample size:

1. correlate replicate-level FPR across M03, edgeR, and DESeq2 on the **same frozen split**;
2. identify the high-FPR splits;
3. calculate each physical sample's inclusion/assignment influence on FPR;
4. inspect available metadata for repeated sample-level drivers.

If the same samples or the same A/B compositions generate tails across methods, that directly supports the sample-heterogeneity interpretation. If a split is benign for one method and extreme for another, that isolates a method-specific response. [S21, S25]

<a id="bq-exact-when-feasible-cprime"></a>
# BQ. C-prime numerical backend: from continuum failures to exact-when-feasible

## BQ.1 The original gamma-continuum approximation failed exact audits

The initial fast C-prime backend used a gamma-continuum approximation in high-count/high-`mu/r` regions. Detailed audits showed that when it failed, the dominant error was usually the **finite-`mu` continuum approximation itself**, not the 128-node quadrature.

Representative failures included backend discrepancies around:

```text
TNBC examples: ~0.0079 and ~0.0202
AdLC n20 rep24:
    backend error       ~0.009243
    interpolation error ~0.000060
OC n5 rep21:
    backend error       ~0.002411
    interpolation error ~0.000083
```

Increasing quadrature order did not close these gaps. This ruled out “just use more gamma quadrature nodes” as the primary fix.

## BQ.2 Continuum eligibility was tightened without loosening the audit

The continuum eligibility policy was aligned to:

\[
\mu\ge 50{,}000,
\qquad
\mu/r\ge100,
\]

with the C-prime audit tolerance frozen at:

\[
0.001.
\]

A six-cell integration audit was run before resuming the full benchmark.

Under the then-10M exact-support cap, four of six cells passed and the two remaining failures were work-cap failures, not interpolation/audit failures. The observed exact support windows included approximately:

```text
AdLC n40: ~9.03M
OC   n40: ~10.30M
TNBC n40: ~12M
```

Raising the explicit exact-work cap to **15M** gave a 6/6 pass in that targeted audit.

## BQ.3 The full benchmark exposed a deeper issue: exact whenever feasible

The six-cell success did not establish that the gamma approximation was accurate at every eligible point in the full benchmark. Additional cells still showed substantial gamma-vs-exact error.

The policy was therefore changed from “use continuum in a nominal high-count region” to:

```text
if exact central support window <= cprime_support_max:
    use exact discrete C-prime

else if point is in the frozen continuum-eligible region:
    use gamma continuum

else:
    fail explicitly
```

This is the **exact-when-feasible** policy.

The key conceptual change is that the support cap is a computational work limit, not a statistical threshold.

## BQ.4 Exact-gene backend was corrected to use the same policy

A separate implementation audit found that the selective exact-gene backend still switched away from discrete summation above the small caching threshold. That was inconsistent with the new genome-wide policy.

The corrected exact-gene backend is now:

```text
support width <= 250,000:
    exact discrete, cached arrays

250,000 < support width <= cprime_support_max:
    exact discrete, chunked

support width > cprime_support_max:
    continuum only if frozen continuum eligibility holds
```

Thus `cprime_grid_exact_width_max = 250,000` is a **memory/cache threshold**, not the exact-versus-continuum threshold.

This correction matters for both the mandatory `R_F<=0` exact fallback and any later selected-gene exact refinement. [S22]

<a id="br-threadpool-fix"></a>
# BR. Windows threadpool and execution fix

The exact backend exposed a Windows concurrency failure:

```text
OSError: GetModuleFileNameEx failed
```

The problem arose because `threadpoolctl` was being invoked concurrently from the three top-level dataset threads, while nested exact-gene execution could also touch the same process-global native-library state.

The corrected benchmark execution policy is:

```text
top-level workers:
    exactly 3
    one worker each for AdLC, OC, TNBC

BLAS/OpenMP:
    limited to 1 thread process-wide
    limit established once in the main thread

inside dataset worker threads:
    threadpoolctl.threadpool_limits temporarily replaced by a no-op context

nested exact-gene workers:
    N_JOBS_EXACT = 1
```

The real process-wide BLAS limit remains active while the worker-thread `threadpoolctl` calls are disabled, so the fix avoids both Windows DLL-enumeration races and accidental nested oversubscription.

This was an implementation/runtime repair, not a statistical-method change. [S23]

<a id="bs-seam-safe-cprime"></a>
# BS. TNBC n=30 replicate 35: seam pathology and seam-safe C-prime

## BS.1 The apparent “slow split” was not simply a slow exact calculation

During the 15M exact-when-feasible run, TNBC n=30 replicate 35 stalled in Stage 2. The support-width prescan showed why blind grid refinement was dangerous.

Approximate grid work by refinement level was:

| Refinement | Grid | Exact nodes | Gamma nodes | Approx exact support work |
|---:|---:|---:|---:|---:|
| 0 | 90 × 80 | 6,922 | 278 | 4.719B points |
| 1 | 135 × 120 | 15,590 | 610 | 10.588B |
| 2 | 203 × 180 | 35,189 | 1,351 | 23.833B |
| 3 | 304 × 270 | 79,077 | 3,003 | 53.412B |
| 4 | 456 × 405 | 177,973 | 6,707 | 120.021B |

The base-grid row reconstruction alone took approximately **512.9 seconds**.

## BS.2 The audit localized the error to mixed exact/gamma cells

At refinement 0:

```text
linear interpolation:
    max audit error       = 0.00639143
    mixed-cell max error  = 0.00639143
    nonmixed max error    = 0.00003642

cubic interpolation:
    max audit error       = 0.00603233
    mixed-cell max error  = 0.00603233
    nonmixed max error    = 0.000001993

mixed exact/gamma grid cells = 50
mixed audit points           = 1 / 120
```

This was decisive. Ordinary interpolation inside a homogeneous numerical backend was extremely accurate. The large error came from interpolating **across the numerical seam** between an exact discrete target and a gamma-continuum target that were not equal at the work-cap boundary.

Refining the rectangular grid could not repair a discontinuity in the numerical target; it merely multiplied exact work.

## BS.3 Seam-safe evaluator

The fix preserves the node-level exact/gamma policy but forbids mixed-backend averaging.

For a query point:

```text
ordinary homogeneous cell:
    existing rectangular interpolation

mixed exact/gamma cell:
    classify the query point by its own production backend;
    try same-backend 2-D branch interpolation;
    if that branch cannot evaluate the query:
        direct production-backend fallback
```

The implementation uses backend-specific Delaunay/`LinearNDInterpolator` objects only for mixed-cell queries. The same hybrid evaluator is used for:

```text
the independent C-prime audit;
the final C1/C2/C3/C4 full-model matrices;
the final C1/C2/C3/C4 null-model matrices.
```

Therefore the frozen `0.001` audit tolerance remains tied to the actual production approximation.

## BS.4 Validation on the pathological split

The seam-safe implementation was isolated and tested on the same TNBC n=30 replicate 35 split.

The result was:

```text
selected grid:                 90 x 80
passed at refinement 0:        True
C-prime max audit error:       0.0003642
mixed exact/gamma cells:       50
audit mixed-cell points:       1
audit direct seam fallback:    1

production mixed-cell points:  7
production branch interpolation: 3
production direct fallback:      4

C-prime build time:            ~535.7 s
whole split time:              ~622.5 s
Li2020 FPR:                    0.002043
```

Relative to the previous cubic mixed-seam error of about `0.00603`, the maximum audit error fell by roughly a factor of 16.6 and stayed below the unchanged 0.001 tolerance.

This validates the seam-safe logic on the known pathological split. It is strong targeted evidence, not a universal proof that every future exact/gamma seam is harmless. [S24]

<a id="bt-support-cap-20m"></a>
# BT. OC n=40 replicate 13: support-cap audit and the move to 20M

## BT.1 The 15M failure was only slightly beyond the work cap

In the descending 15M seam-safe run, OC n=40 replicate 13 failed because one production grid point required:

\[
15{,}027{,}503
\]

discrete support points, compared with the frozen cap of:

\[
15{,}000{,}000.
\]

The offending point was approximately:

\[
\mu=47{,}736.5,
\qquad
r=0.07242.
\]

It was **not** continuum eligible because `mu < 50,000`, despite a very large `mu/r`.

The width exceeded the cap by only about **0.1834%**.

## BT.2 Direct exact evaluation was cheap and numerically clean

The dedicated uncapped audit evaluated the offending node exactly:

```text
support width:          15,027,503
exact runtime:          ~1.57 s
recovered NB mass:      ~1.00000000001
exact C1:               0.0495917572
```

Thus the failure was clearly a computational policy boundary, not evidence that exact evaluation at this width was impractical.

## BT.3 Gamma continuum remained badly wrong

At 8192 quadrature nodes:

```text
exact C1:               0.0495918
gamma C1:               ~0.037461
absolute C1 error:      ~0.01213
```

The error was about twelve times the entire `0.001` C-prime tolerance and changed little with increasing quadrature order.

This again identified **finite-`mu` continuum error**, not quadrature resolution.

Therefore the project explicitly rejected lowering the continuum `mu>=50,000` threshold merely to make this point run.

## BT.4 Refinement-wide work scan

Across all permitted refinement grids, the largest non-continuum-eligible support widths were approximately:

```text
refinement 0: 15,027,503
refinement 1: 15,262,771
refinement 2: none
refinement 3: none
refinement 4: 15,292,312
```

A minimal rounded cap around 15.3M would have covered this observed range. Sean preferred additional headroom.

The final decision was:

\[
\boxed{
\texttt{cprime\_support\_max = 20,000,000}
}
\]

while leaving unchanged:

```text
continuum minimum mu          = 50,000
continuum minimum mu/r        = 100
C-prime audit tolerance       = 0.001
seam-safe mixed-cell policy   = unchanged
gamma quadrature policy       = unchanged
```

The 20M value is an **exact-evaluation work ceiling**, not a new statistical tuning parameter. It intentionally favors exact discrete calculation over continuum approximation for a wider range of feasible supports. [S26]

<a id="bw-updated-position"></a>
# BW. Updated scientific position and next experiments

## BW.1 The statistical M03 architecture remains unchanged by the numerical repairs

The current two-group M03 statistical architecture remains:

```text
WEB-DP r_g
+
DEB theta on ordinary admissible genes
+
first-order C'(1) residual coordinate
+
kappa_F = 1
+
ASH-weighted kappa_c
+
support-feasible null geometry
+
mandatory exact normalized-DNB theta fallback for R_F<=0
+
BH over the fixed analyzed universe.
```

The September 25–October 1 work changed **how C-prime is evaluated numerically**, not the underlying statistical target.

The three major numerical/backend rules are now:

```text
1. exact discrete C-prime whenever support width <= 20M;

2. gamma continuum only when exact support exceeds that work cap
   and the frozen continuum eligibility criteria hold;

3. never interpolate across an exact/gamma seam.
```

## BW.2 The Li2020 benchmark is now scientifically more important than MAQC-specific tuning

The biological-null benchmark is directly testing the limitation identified in the MAQC interpretation:

```text
standardized reference-material replication
versus
real heterogeneous biological samples.
```

The historical competitor behavior shows that this dataset is not a trivial null. Sample composition can matter greatly, and different methods respond differently as `n` grows.

This makes Li2020 especially valuable for evaluating:

```text
null calibration under biological heterogeneity;
sample-size behavior;
right-tail split sensitivity;
R_F<=0 frequency;
support-floor frequency;
exact-fallback frequency;
negative-T frequency;
r_g boundary behavior;
DEB theta stability;
method-specific interaction with latent sample structure.
```

## BW.3 Current M03 evidence is promising but still interim

The superseded 15M run showed low 40v40 M03 medians and means relative to edgeR, with OC especially tight and AdLC/TNBC retaining right tails.

That is a strong reason to continue the benchmark, but the final claim must wait for the new fresh 20M run.

The canonical Li2020 result should report for every dataset and sample size:

```text
all 50 replicate FPRs;
mean / median / SD;
2.5, 5, 10, 25, 75, 90, 95, 97.5% quantiles;
zero-FP frequency;
P(FPR >= 0.1%, 0.5%, 1%, 2%, 5%);
P(R>0);
raw-p calibration;
paired frozen-split correlations across methods;
M03 numerical diagnostics and failure counts.
```

## BW.4 Sample-level influence is the highest-value next mechanism audit

Once the final 20M 40v40 block is available, preserve split identity and ask which physical samples repeatedly appear in the high-FPR tails.

A useful sample influence statistic is:

\[
\Delta_s
=
E(\mathrm{FPR}\mid s\text{ included})
-
E(\mathrm{FPR}\mid s\text{ excluded}),
\]

with separate assignment effects for being placed in group A versus B where appropriate.

This should be connected to available sample metadata rather than inferred from filenames.

## BW.5 Dense-alternative directional-scale policy remains a separate unresolved issue

The general-`X` / Mixology work and the Li2020 M03 work should not be conflated.

For the current two-group Li2020 M03 benchmark, the archived ASH `kappa_c` procedure remains the frozen production branch.

For a future general-user implementation, the project should expose a clear policy:

```text
ordinary null-rich design:
    ASH directional calibration

alternative-dense / insufficient-null-support design:
    lambda_c = 1 fallback
    + explicit diagnostic warning
```

The exact diagnostic trigger still needs to be formalized and validated outside Mixology.

## BW.6 Negative raw statistics remain open

None of the C-prime numerical repairs resolves the separate theoretical question of negative geometry-adjusted `T_raw`.

The current rule remains:

\[
T=\max(T_{\mathrm{raw}},0),
\qquad
p=P(\chi_1^2\ge T).
\]

This remains a statistical-reference issue, not a C-prime admissibility or support-window issue.

# BX. Implementation, checkpoint, and provenance updates

## BX.1 Split-ID handling

The first isolated TNBC seam validation attempted to select a hard-coded split ID with zero padding. The manifest used a different literal split-ID format, so the validation initially found zero matching rows.

The corrected validation selects by:

```text
dataset = TNBC
n_per_group = 30
replicate = 35
```

and only then uses the manifest's actual `split_id`.

This should remain the pattern for future targeted audits.

## BX.2 Old audit versus new seam-safe validation

An older auto-discover C-prime audit was accidentally rerun after the seam-safe implementation had been created. Because it resumed the old grid-row caches, it was **not** a test of the new seam-safe method.

The isolated seam-safe validation was therefore kept in its own namespace and recomputed the target split under the new code.

This distinction is important: resuming old numerical caches can silently turn a “validation of new code” into a rerun of old code.

## BX.3 Checkpoint-reuse policy evolved several times

After the seam fix, an initial runner used a completely new namespace and restarted every stage. A Stage-1-only migration runner was then constructed because the seam change begins in Stage 2.

Sean correctly objected that recomputing Stages 2–5 was still most of the expensive run and therefore did not preserve enough of the completed work to be useful.

A more selective salvage rule was discussed:

```text
completed split with only one C-prime backend on its old grid:
    provably no exact/gamma seam in that grid

completed split with both exact and gamma nodes:
    potentially seam-affected; audit before reuse
```

Rather than build a mixed-lineage final benchmark, Sean chose a new descending 40→5 run.

After the OC support-cap audit, Sean then explicitly requested that the final 20M run be **fresh with no resume at all**. That final policy now overrides the general resumability default for this benchmark only.

## BX.4 Corrupted generated runner was replaced

The first generated descending-run artifact was reported corrupted. It was regenerated cleanly from the source text and Python-compiled before delivery.

This was an artifact-generation failure, not a statistical or checkpoint failure.

The current 20M fresh runner was likewise generated from the current seam-safe source and syntax-checked.

## BX.5 Exact-gene semantics are now fixed

The project terminology remains:

```text
"exact gene refinement"
"exact-profile the called genes"
"do exact refinement afterward"
```

means selected-gene **full normalized-DNB precision profiling through**
`fit_webdnb_m03(..., refine_from=previous_result, ...)`, with genome-wide M03 quantities frozen.

It does **not** mean recomputing the genome-wide C-prime grid exactly.

Conversely:

```text
"exact C-prime grid"
```

refers to the genome-wide/base normalizer derivative numerical backend.

These two operations must remain distinct in code, discussion, and summary language.

## BX.6 Current source lineage

The relevant current chain is:

```text
canonical support-feasible M03
    ->
15M exact-when-feasible C-prime
    ->
exact-gene backend aligned to exact-when-feasible policy
    ->
Windows threadpool control fix
    ->
TNBC n30 rep35 mixed-seam diagnosis
    ->
seam-safe branch/direct evaluator
    ->
pathological-split validation at 90x80
    ->
descending Li2020 40->5 run
    ->
OC n40 rep13 15M cap failure
    ->
uncapped exact-vs-gamma support audit
    ->
20M support-cap decision
    ->
fresh nonresuming 20M final benchmark namespace
```

The current final benchmark artifact is:

```text
webdnb_m03_seam_safe_20M_DESC40to5_FULL_FRESH_v1.py
```

and the current method version string is:

```text
2026-10-01-M03-seam-safe-cprime-20m-v1
```

## Final research position at the October 1 cutoff

The project now has three distinct levels of evidence that should not be collapsed.

**First, the M03 statistical architecture is mature enough to freeze.** The MAQC work established why exact normalized-DNB fallback and support-feasible null geometry are necessary, and the later numerical work did not change that statistical target.

**Second, the C-prime numerical backend is materially stronger than it was at the September-16 cutoff.** The project has identified and repaired two separate numerical failures: finite-`mu` continuum error where exact calculation was feasible, and mixed exact/gamma interpolation across a discontinuous numerical seam. The support cap is now 20M with unchanged continuum eligibility and audit tolerance.

**Third, the biological-null benchmark is giving the first strong evidence about how the method behaves under real sample heterogeneity.** The historical competitor reproduction shows a broad edgeR large-`n` shift in this benchmark and a different DESeq2 right-tail pattern. The superseded 15M M03 run showed a much lower 40v40 center, with right-tail sensitivity rather than broad inflation. The fresh 20M run must finish before those M03 numbers become canonical.

The most important unresolved scientific questions are therefore:

```text
1. Does the final 20M M03 run preserve the encouraging 40v40 calibration
   across all sample sizes and all 50 frozen replicates?

2. Which physical samples / latent sample structures drive the high-FPR tails?

3. Does M03's residual-precision + support-feasible architecture separate
   method calibration from real latent biological imbalance better than the competitors?

4. Can a reliable design-aware warning identify when ASH directional calibration
   lacks enough null information, so lambda_c=1 should be used instead?

5. What is the correct theoretical treatment of negative T_raw?
```

No broad dominance claim is justified yet. The strongest current statement is narrower:

> The archived M03 architecture has strong controlled-reference performance, and its interim Li2020 biological-null behavior is substantially more encouraging than the edgeR large-`n` behavior in the same frozen splits. The final fresh 20M run and sample-level influence audit are the next evidence required to determine how general that advantage is.



<a id="dated-continuation-oct2-oct3"></a>
# Part IV — Dated continuation and current interpretation: October 2–3, 2026

**Update timestamp:** October 3, 2026, 1:20:00 PM EDT (UTC−04:00).

This continuation records the work after the October-1 cutoff. The most important scientific change is that the
ordinary DEB precision layer is no longer treated as settled merely because DEB outperformed WEB precision in the
earlier Baik/MAQC development grid. The Pickrell complete-null work exposed a different failure mode: the fitted DEB
Gamma prior can become so concentrated that it almost completely removes a strong coverage-dependent precision shape
already present in the raw DB/DNB residual statistic. A controlled shrinkage-strength experiment then showed that
retaining part of that raw shape can improve null calibration by an order of magnitude in the extreme tail.

The older MAQC results, exact-normalizer fallback, support-floor repair, and C-prime numerical work remain part of the
record. The new finding reopens **the strength of ordinary DEB shrinkage**, not the existence of the DNB model, the
normalizer correction, or the targeted exact fallback. [S30–S38]

<a id="by-li2020-kappa-coverage"></a>
# BY. Li2020 40v40: directional scale and false positives are strongly coverage-dependent

## BY.1 The first 40v40 audit used the completed splits available at the time

The coverage audit was launched while the fresh 20M Li2020 40v40 block was still running. The snapshot contained:

```text
AdLC: 15 completed splits
OC:   18 completed splits
TNBC: 16 completed splits
total: 49 completed 40v40 splits
```

For each split the audit retained the original global \(\kappa_c\), then estimated an ASH-weighted
\(\kappa_c(\text{coverage})\) in twelve equal-count coverage bins using the same contrast-scale quantity rather than
inventing a new test statistic. The resulting pattern was highly reproducible across the three datasets. [S30]

The local-to-global ratio was generally above one at low-to-middle coverage and below one at high coverage.
Representative dataset-median patterns were approximately:

```text
AdLC:
    low/mid coverage:  ~1.00 -> ~1.22
    high coverage:     ~0.95 -> ~0.76-0.79

OC:
    low/mid coverage:  ~0.93 -> ~1.18
    high coverage:     ~0.91 -> ~0.77

TNBC:
    low/mid coverage:  ~1.17 -> ~1.34
    high coverage:     ~0.91 -> ~0.75
```

The sign and shape were therefore not peculiar to one cancer matrix.

## BY.2 False positives concentrated in the same coverage region

False-positive genes concentrated strongly in the intermediate-coverage region where local \(\kappa_c\) exceeded the
global value. The high-coverage tail contained far fewer false positives under the original global-\(\kappa_c\) M03 fit.

This made the original scalar interpretation incomplete:

\[
\boxed{
\kappa_c\text{ is not acting like one coverage-invariant null scale in these real data.}
}
\]

The global mean can conceal opposite conditional errors: anti-conservative middle-coverage behavior and conservative
high-coverage behavior can partially cancel in a whole-genome average.

## BY.3 Split-level evidence supported a structural pattern

Across the 49 completed splits, the fraction of raw \(p<.01\) genes was strongly associated with benchmark FPR, and the
coverage-shape diagnostics moved with split difficulty. That motivated the direct counterfactual question: if
\(\kappa_c\) were allowed to vary by coverage, would the false positives disappear?

The answer was useful but ultimately negative as a production proposal. [S30, S31]

<a id="ca-local-kappa-counterfactual"></a>
# CA. Local \(\kappa_c(\text{coverage})\) is diagnostically real but is not the production fix

## CA.1 Li2020 binwise counterfactual reduced many false positives

A direct Li2020 40v40 counterfactual replaced the scalar global \(\kappa_c\) with the twelve-bin local curve while
recomputing the downstream null geometry and test. Among the 49 completed splits, **47 improved and only two worsened**
on the displayed benchmark FPR.

However, it also added false positives in high-coverage bins where the local scale fell below the global value. This was
exactly the concern raised before the counterfactual: repairing the middle can make the high-coverage tail more
permissive.

## CA.2 Pickrell full comparison showed the decisive failure mode

The completed Pickrell comparison reused saved original edgeR/DESeq2 p-values and compared:

```text
edgeR exact
edgeR QL
DESeq2 LRT
DESeq2 Wald
WEB-DNB M03 global kappa_c
WEB-DNB M03 12-bin kappa_c(coverage) diagnostic
```

Approximate complete-null summaries were:

| Method | Mean raw \(p<.01\) | Mean BH(.05) discoveries | \(P(R>0)\) |
|---|---:|---:|---:|
| edgeR QL | 0.01209 | 0.23 | 0.133 |
| edgeR exact | 0.01130 | 4.90 | 0.567 |
| DESeq2 LRT | 0.01214 | 8.10 | 0.667 |
| DESeq2 Wald | 0.01278 | 11.07 | 0.833 |
| M03 global \(\kappa_c\) | 0.01322 | 2.70 | 0.667 |
| M03 12-bin \(\kappa_c(\text{coverage})\) | 0.01278 | 6.77 | 0.900 |

The local-\(\kappa_c\) branch modestly improved the coarse \(p<.01\) average but made the BH extreme tail substantially
worse. At \(p=.001\), the global branch had roughly \(0.00244\) of genes below threshold, whereas the local branch had
roughly \(0.00280\).

Therefore:

\[
\boxed{
\kappa_c(\text{coverage})\text{ is detecting a real coverage effect, but correcting it directly does not repair the
full null distribution.}
}
\]

This was the key reason to move upstream rather than turn local directional scaling into the production method.
[S31, S33]

## CA.3 Aggregate calibration can hide conditional and extreme-tail errors

Two rules became explicit:

1. whole-genome \(p<.01\) calibration can look close to nominal while coverage strata contain large positive and negative
   errors that cancel;
2. BH can be poor even when \(p<.01\) looks acceptable, because the first BH thresholds with roughly \(10^4\) genes live
   near \(10^{-6}\)–\(10^{-5}\).

<a id="cb-precision-layer-audit"></a>
# CB. Precision-layer audit: raw curvature versus global DEB flattening

## CB.1 Final DEB \(\theta\) was almost perfectly flat over coverage

The fitted DEB values were nearly horizontal:

```text
median theta_F across coverage bins ~1.0214
median theta_0 across coverage bins ~1.0436
median theta_0/theta_F              ~1.032
```

The raw-p FPR curve changed by more than an order of magnitude over the same coverage axis.

## CB.2 The raw pre-DEB precision had the missing curvature

The crucial comparison was:

\[
\widetilde\theta_F=\frac{a_F}{R_F},
\qquad
\widetilde\theta_0=\frac{a_0}{R_0^\dagger}.
\]

Raw full precision rose from roughly \(0.83\) at low coverage to above \(1.7\) at high coverage. Raw null precision showed
a similar rising curve. The DEB posterior values stayed near one.

Therefore the DB/DNB residual score was **already generating coverage curvature**. No new probability-coordinate
transformation was required to create it.

## CB.3 The DEB hyperparameters explained the flattening

Across many Pickrell splits, fitted \(\widehat\alpha_0\) hit or approached \(10^9\), with prior mean near one. For a
Gamma prior:

\[
CV(\theta)=\frac{1}{\sqrt{\alpha_0}}.
\]

At \(10^9\), this is only about \(3\times10^{-5}\): effectively a point mass.

The Pickrell failure therefore moved from "missing curvature" to "curvature erased by global empirical-Bayes shrinkage."
[S34]

<a id="cc-residual-component-audit"></a>
# CC. Residual-component audit localized the coverage trend upstream of DEB

The full side was decomposed into \(S_F/a_F\), \(C_F/a_F\), and
\(R_F/a_F=(S_F-C_F)/a_F\). The null side was decomposed analogously.

The corrected residual information was strongly coverage-dependent:

```text
R_F/a_F:
    ~1.20 at low coverage
    -> ~1 near the middle
    -> ~0.58-0.60 at the high end

R_0^dagger/a_0:
    similar trajectory
```

Since:

\[
\widetilde\theta=\frac{a}{R}=\frac{1}{R/a},
\]

this is exactly the source of raw-\(\theta\) curvature.

At very low coverage, \(C/a\) was large and negative, so the first-order normalizer correction did substantial work.
At moderate/high coverage, \(C/a\) was already only a few hundredths while \(R/a\) continued to fall. The high-coverage
trend is therefore not primarily a C-prime interpolation/normalizer artifact.

The support-feasible null adjustment \((S_0-S_0^\dagger)/a_0\) was also small compared with the broad movement in
\(R/a\). It remains necessary for admissibility but does not explain the coverage trend.

The WEB-DP audit showed the first two coverage bins at approximately the Poisson boundary
(\(r_g\sim10^8,\alpha_g\sim10^{-8}\)), followed by a sharp transition to finite NB curvature. That can help explain the
near-zero FPR at the extreme low-coverage end, but the main middle/high-coverage problem remains the second-stage
precision shrinkage. [S34, S35]

<a id="cd-contextual-db-rescaling"></a>
# CD. Contextual DB rescaling experiments were informative failures

Sean proposed that effective proportion might need a universal weighted variation context. Two implementations were
tested before the raw-precision result fully clarified the problem.

The first used:

\[
u_\epsilon(p)=\epsilon+(1-2\epsilon)F_w(p).
\]

Current geometry had combined flatness around \(0.63\); the best weighted-CDF candidate was around \(5.8\) with about
17–18% nested inversions. The mathematical cause was then identified: local KL curvature contains \(u'(p)^2\), and here
\(u'(p)\propto f_w(p)\), so the empirical weighted density entered the metric squared.

The second used weighted quantile anchors and:

\[
\operatorname{logit}u(p)=A+B\operatorname{logit}p.
\]

The best tested candidate was roughly \(\gamma=.5,\tau=.05,\epsilon=.05\), but its objective was still around **1.77**,
far worse than the ~0.63 baseline, and its residual curves were strongly U-shaped.

The weighted anchors were approximately \(p_L\approx0.151\) and \(p_U\approx0.994\), also showing that Pickrell's fitted
effective proportions were not universally trapped near zero.

Both one-dimensional contextual maps are demoted. The immediate problem does not require manufacturing new curvature:
the raw DB/DNB precision already has it. [S36]

<a id="ce-tempered-deb"></a>
# CE. Major development: tempered DEB shrinkage

## CE.1 Exact reciprocal-precision shrinkage identity

For ordinary DEB:

\[
\widehat\theta
=
\frac{a+\alpha_0}{R+\beta_0},
\qquad
m=\frac{\alpha_0}{\beta_0},
\qquad
\widetilde\theta=\frac{a}{R}.
\]

Define:

\[
\boxed{
s=\frac{\alpha_0}{a+\alpha_0}.
}
\]

Then exactly:

\[
\boxed{
\frac{1}{\widehat\theta}
=
(1-s)\frac{1}{\widetilde\theta}
+
s\frac{1}{m}.
}
\]

Thus \(s\) is the shrinkage fraction on the reciprocal-precision scale.

## CE.2 Current Pickrell DEB is effectively at \(s=1\)

For essentially all 30 Pickrell splits:

\[
s_F\approx1,
\qquad
s_0\approx1.
\]

Many displayed values were 1.000000 to six decimals. The current DEB fit is therefore applying near-total shrinkage to
the raw precision curve.

## CE.3 Universal shrinkage-strength experiment

Every upstream M03 component was frozen:

```text
WEB-DP r_g
fitted means
C_F and C_0
global kappa_c
support-feasible S0dag
mandatory exact branch
current fitted DEB prior means m_F and m_0
```

Only one universal \(s\) was varied for both ordinary full and null precision. The response was strongly U-shaped:

```text
s = 0:
    mean raw p<.01 ~1.48%

s ~ .3-.6:
    mean raw p<.01 ~0.9%-1.05%

s = .75:
    mean raw p<.01 ~0.96%

s -> 1/current:
    mean raw p<.01 ~1.32%
```

Neither no shrinkage nor near-total shrinkage was optimal.

## CE.4 Broad raw-p calibration was remarkably good near \(s=.75\)

Approximate observed/nominal ratios at \(s=.75\):

| Threshold | Observed / nominal |
|---:|---:|
| .001 | 0.97 |
| .005 | 0.95 |
| .010 | 0.95 |
| .025 | 0.98 |
| .050 | 0.99 |

Archived current DEB was approximately:

```text
.001: 2.44
.005: 1.55
.010: 1.32
.025: 1.15
.050: 1.06
```

## CE.5 BH behavior showed a broad moderate-shrinkage basin

Approximate complete-null \(P(R>0)\):

```text
s = 0:         0.50
s = .05:       0.30
s = .10:       0.167
s = .15:       0.133
s = .20-.25:   0.067
s = .30-.50:   0.033
s = .55-.70:   0.067
s = .75-.80:   0.10
s = .85:       0.20
s = .90-.925:  0.267
s = .95:       0.433
s = .975:      0.533
s = .99:       0.60
s -> current:  0.667
```

With 30 splits, these values move in steps of \(1/30=0.0333\). The exact minimum is not precise; the broad intermediate
basin is the robust result.

## CE.6 Scientific interpretation

The earlier statement "DEB theta is the leading precision branch" remains historically correct relative to WEB theta in
Baik/MAQC. The new conclusion is narrower and more important:

\[
\boxed{
\text{the DEB prior target can be useful while its fitted concentration is far too strong.}
}
\]

The raw DB/DNB residual precision already contains calibration information. A globally over-concentrated EB prior can
erase it. [S37]

<a id="cf-extreme-tail"></a>
# CF. Extreme-tail audit and the prior-concentration-cap formulation

With roughly \(10^4\) genes, the first BH threshold is near \(5\times10^{-6}\), so calibration at \(p=.001\) is too
coarse to explain BH.

The extended audit pushed down to \(2.5\times10^{-6}\). Archived current DEB had approximate observed/nominal ratios:

```text
2.5e-6:  ~31.6
5e-6:    ~22.5
1e-5:    ~15.8
2.5e-5:  ~10.1
5e-5:    ~8.0
1e-4:    ~5.6
2.5e-4:  ~4.1
5e-4:    ~3.1
1e-3:    ~2.44
5e-3:    ~1.55
1e-2:    ~1.32
```

Moderate shrinkage raised median split minimum p-values from a few \(10^{-6}\) to roughly \(10^{-4}\) in the middle of
the grid. "All genes" and "ordinary first-order genes" minimum-p curves were essentially identical, showing that the
0–1 mandatory exact genes per split were not causing the extreme-tail problem.

The stable extended-tail development objective again favored approximately \(s=.75\). Absolute error of the 30-split BH
FDR from 0.05 favored roughly \(s=.25\), but the resolution is only \(1/30\).

The project therefore should not overfit a decimal-level \(s\) to Pickrell.

A cleaner formulation is a cap on prior concentration. Let:

\[
c=\frac{\alpha_0}{a},
\qquad
s=\frac{c}{1+c}.
\]

Then use:

\[
\boxed{
\alpha_0^*
=
\min(\widehat\alpha_0,\ c_{\max}a)
}
\]

while preserving fitted prior mean \(m=\widehat\alpha_0/\widehat\beta_0\):

\[
\boxed{
\beta_0^*
=
\frac{\alpha_0^*}{m}.
}
\]

The active frozen external-validation candidates are:

```text
c_max = 0.25   <=> s_max = 0.20
c_max = 1      <=> s_max = 0.50
c_max = 3      <=> s_max = 0.75
```

A cap is preferable to forcing one shrinkage fraction everywhere because it leaves an already-diffuse DEB fit unchanged
and only prevents empirical Bayes from becoming effectively infinitely concentrated. [S38]

<a id="cg-updated-architecture"></a>
# CG. Updated M03 architecture and immediate validation plan

The new finding specifically reopens ordinary DEB shrinkage strength. These layers remain supported:

```text
WEB-DP r_g
first-order C'(1) ordinary residual coordinate
exact-discrete-when-feasible / seam-safe C-prime backend
kappa_F = 1 in the first-order coordinate
ASH kappa_c in ordinary null-rich designs
support-feasible null geometry
mandatory exact normalized-DNB theta profile when R_F<=0
T=max(T_raw,0) as the explicit current negative-statistic rule
BH over the declared finite-p universe
```

Two method identities must now be kept separate.

**Archived M03 / MAQC-era DEB** uses the unconstrained fitted DEB concentration. This is the method that produced the
saved MAQC, RNAontheBENCH, and current Pickrell/Li2020 development outputs.

**Tempered/capped-DEB M03** keeps the same fitted DEB prior mean but caps concentration relative to \(a\), preserving part
of the raw \(a/R\) curvature. This is now the leading precision-layer development candidate.

The next experiment should freeze \(c_{\max}=.25,1,3\) and evaluate all three on Li2020 40v40 with global
\(\kappa_c\), no coverage-specific directional correction, the same support-feasible geometry, the same seam-safe 20M
backend, the same mandatory exact fallback, and the same BH universe.

Primary outputs should include:

```text
raw-p calibration by threshold
raw-p calibration by coverage
complete-null P(R>0)
BH discovery distribution
minimum-p distribution
paired split changes relative to archived M03
theta_F / theta_0 coverage curves
R_F<=0 and support-floor frequencies
```

Only after frozen external transfer should one cap be considered for the production method.

## CH.5 Meaning of `max exact width`

The log quantity:

```text
max exact width
```

is the largest single exact NB support window among exact-evaluated C-prime grid nodes in that split. It is not the
number of genes and not total exact work.

Under the 20M policy, the reported maximum is censored by the exact-work threshold, so repeated values near 20M are
expected when the support-width distribution extends beyond the cap. A near-cap maximum alone does not prove that many
nodes are over 20M.

<a id="dated-continuation-oct3-evening"></a>
# Part V — Dated continuation and current interpretation: October 3 evening, 2026

**Update timestamp:** October 3, 2026, 9:11:00 PM EDT (UTC−04:00).

This continuation records the work after the 1:20 PM Part-IV cutoff. It supersedes Part IV where the later experiments
changed the interpretation of the coverage problem. The principal scientific change is that the major Pickrell/Li2020
coverage-dependent calibration defect is no longer attributed primarily to a missing coverage-specific \(\kappa_c\) or
to a missing coverage-specific DEB target. The strongest current evidence places the dominant defect in the
**conditional null reference law of the final M03 statistic**, with DEB shrinkage strength remaining a smaller but real
second-order calibration layer. A one-extra-parameter generalized-Gamma positive component then removed a substantial
part of the remaining Li2020 tail error but did not fully repair the first-BH extreme tail. [S39–S44]

<a id="ci-capped-deb-transfer"></a>
# CI. Why direct capped-DEB transfer was not enough

## CI.1 The first external transfer test used the available Li2020 40v40 splits

After Pickrell showed that the fitted DEB concentration could imply essentially complete shrinkage, the project tested
the simpler hypothesis that limiting that concentration would itself repair Li2020.

The first transfer audit used the completed 40v40 splits available at the time:

```text
AdLC: 15 splits
OC:   18 splits
TNBC: 16 splits
total: 49 splits
```

The principal concentration-cap candidates inherited from the Pickrell development work were:

\[
c_{\max}=0.25,\;1,\;3,
\]

equivalently maximum universal reciprocal-precision shrinkage fractions:

\[
s_{\max}=0.20,\;0.50,\;0.75.
\]

All upstream M03 objects were held fixed and only the ordinary DEB full/null precision shrinkage was changed. [S39]

## CI.2 Strong caps did not transfer cleanly

The direct capped-DEB branch did **not** reproduce the dramatic Pickrell improvement on Li2020.

The qualitative pattern was consistent across the displayed tables:

- aggressive caps could increase Li2020 FPR substantially;
- \(c_{\max}=3\) / \(s_{\max}=.75\) was often closest to current DEB and could be slightly better in some cells, but there
  was no clean three-dataset transfer win;
- the coverage-specific raw-\(p\) hump remained recognizable;
- changing only the amount of DEB shrinkage did not explain why the same statistic changed from near-zero error at the
  lowest and highest coverage to strong anti-conservatism in the middle.

This result materially changed the interpretation of the Pickrell \(s\)-grid. Pickrell had correctly revealed that
near-total DEB shrinkage can be harmful, but **DEB over-shrinkage alone was not the whole Li2020 mechanism**.

## CI.3 The cap formulation remains useful algebra, but is no longer the complete repair

The identity

\[
\frac{1}{\theta_s}
=
(1-s)\frac{1}{\theta_{\rm raw}}
+
s\frac{1}{m}
\]

remains exact and useful. The corresponding concentration ratio remains

\[
\frac{\alpha_0}{a}=\frac{s}{1-s}.
\]

What changed is its role. The project no longer treats a concentration cap as the primary explanation of the
coverage-dependent null distortion. It is now a **secondary precision-shrinkage control to be evaluated after the null
reference law is corrected**. [S39, S42]

<a id="cj-raw-precision-transfer"></a>
# CJ. Cross-dataset raw-precision reproducibility and the failure of a universal target-shape explanation

## CJ.1 Raw precision curves were highly reproducible within datasets

A split-to-split reproducibility audit compared the raw ordinary full and null precision curves

\[
\widetilde\theta_F=\frac{a_F}{R_F},
\qquad
\widetilde\theta_0=\frac{a_0}{R_0^\dagger}
\]

across Pickrell and the available Li2020 40v40 splits.

Within each dataset, the coverage curve was highly reproducible. Median split-to-reference curve correlations were
generally near one. The broad curve was therefore not a random artifact of one split.

## CJ.2 But the curve amplitude was strongly dataset-dependent

The important cross-dataset result was that **Pickrell's raw-precision curvature was much larger than the Li2020
curvature**.

Pickrell moved from clearly sub-one low-coverage raw precision to roughly 1.9 or higher at the high-coverage end. The
three Li2020 datasets showed much smaller departures around one. Between-split noise was also much larger for Pickrell
than for the Li2020 curves.

This ruled out a simple story in which one universal coverage-shaped prior target learned from Pickrell could simply be
ported into every dataset. The raw residual coordinate contains real information, but its amplitude is itself
dataset-dependent. [S40]

## CJ.3 Consequence

The project therefore separated two questions:

```text
How much of the raw precision-coverage shape should DEB retain?
```

from

```text
What is the correct null distribution of the final test statistic conditional on coverage?
```

The first is a shrinkage problem. The second is a reference-law problem. The next experiments showed that the second
was the larger effect.

<a id="ck-target-shape-retention"></a>
# CK. Target-shape retention experiment

A smooth target-shape-retention experiment explicitly varied how strongly the fitted DEB target was allowed to inherit a
coverage-dependent shape while holding the rest of M03 fixed.

The retention parameter was varied from a flat target through the full estimated shape. The resulting precision curves
changed exactly as intended, but the benchmark behavior did not change enough to explain the original problem:

- AdLC and TNBC Li2020 FPR were almost flat across much of the retention path;
- OC moved somewhat more, but no common retention value repaired all three datasets;
- Pickrell calibration changed more visibly, reinforcing that Pickrell's raw precision shape is unusually strong;
- the coverage-dependent raw-\(p\) hump survived.

This was an informative negative result. The error could not be reduced to “DEB needs the right coverage-dependent
target curve.” [S40]

<a id="cl-conditional-null-law"></a>
# CL. Conditional null-law diagnosis

## CL.1 The final M03 statistic was not behaving like one \(\chi_1^2\) law across coverage

The project returned to the actual ordinary first-order test statistic after all upstream M03 geometry and precision
work:

\[
Q_g=\max(T_{{\rm raw},g},0).
\]

The existing production mapping was:

\[
p_g=P(\chi_1^2\ge Q_g).
\]

Coverage-stratified null diagnostics showed that the conditional distribution of \(Q\) changed substantially with
coverage even after the existing scale corrections.

Two features were especially important:

1. clipping negative \(T_{\rm raw}\) at zero creates a genuine point mass at \(Q=0\);
2. among \(Q>0\), both scale and shape changed with coverage.

Thus a single continuous \(\chi_1^2\) reference law was being asked to represent a distribution with an explicit atom at
zero and a coverage-dependent positive component.

## CL.2 Moment/Gamma diagnostics localized the remaining mismatch

Coverage-bin audits estimated:

- the zero fraction \(P(Q=0\mid x)\);
- the mean and variance of \(Q\mid Q>0,x\);
- Gamma-equivalent shape/scale, or equivalently effective null-law degrees of freedom and scale;
- conditional QQ behavior.

These diagnostics showed that the dominant coverage hump could be represented as a change in the **conditional law of
the test statistic itself**, not merely as a scalar directional-scale error.

This explains why directly replacing global \(\kappa_c\) by \(\kappa_c(\text{coverage})\) could improve one part of the
curve while making the high-coverage and BH tails worse: one scalar cannot simultaneously repair a point mass, body
shape, scale, and far tail. [S41]

<a id="cm-hurdle-gamma"></a>
# CM. Major success: cross-fitted hurdle-Gamma

## CM.1 Model

The first successful conditional reference law was:

\[
\boxed{
Q_g\mid x_g
\sim
\pi_0(x_g)\delta_0
+
[1-\pi_0(x_g)]
\Gamma\{k(x_g),\theta_\Gamma(x_g)\},
}
\]

where

\[
x_g=\log\{1+\text{total normalized gene coverage}\}.
\]

The implementation used:

```text
5 deterministic coverage-balanced cross-fitting folds
12 equal-count training coverage bins
half-count-stabilized zero fraction pi_0
positive-Q moment-matched Gamma shape and scale
PCHIP smoothing of logit(pi_0), log(k), and log(theta_Gamma)
held-out scoring only
mandatory exact-profile genes retain their canonical current M03 p-values
```

For \(Q>0\),

\[
p_g
=
[1-\pi_0(x_g)]
P\{\Gamma(k(x_g),\theta_\Gamma(x_g))\ge Q_g\},
\]

and for \(Q=0\),

\[
p_g=1.
\]

This was a **reference-law replacement**, not a redefinition of the upstream M03 statistic. [S41]

## CM.2 The coverage hump nearly disappeared in all four null datasets

The first audit used all 30 Pickrell splits and the same 49 available Li2020 40v40 splits.

The key mean raw-\(p\) results were approximately:

| Dataset | Current raw \(p<.01\) | Hurdle-Gamma raw \(p<.01\) | Current raw \(p<.001\) | Hurdle-Gamma raw \(p<.001\) |
|---|---:|---:|---:|---:|
| Pickrell | 0.013216 | **0.008265** | 0.002441 | **0.001282** |
| AdLC | 0.039939 | **0.008810** | 0.011695 | **0.001435** |
| OC | 0.025024 | **0.010079** | 0.006208 | **0.001363** |
| TNBC | 0.047568 | **0.009059** | 0.013990 | **0.001293** |

The coverage plots were more important than the aggregate means. The very different Pickrell, AdLC, OC, and TNBC
coverage humps collapsed to curves close to the nominal 1% level.

This is the strongest current evidence that:

\[
\boxed{
\text{the dominant coverage defect belongs in the conditional null law of }Q.
}
\]

## CM.3 Li2020 benchmark FPR improved in every selected split

Approximate mean Li2020 benchmark FPR changed from:

| Dataset | Current M03 | Hurdle-Gamma |
|---|---:|---:|
| AdLC | 0.008471 | **0.001175** |
| OC | 0.001783 | **0.000071** |
| TNBC | 0.011087 | **0.000120** |

The paired result was unusually clean:

```text
AdLC: 15 / 15 splits improved
OC:   18 / 18 splits improved
TNBC: 16 / 16 splits improved
```

There were no equal or worsened Li2020 FPR splits in that first comparison. [S41]

## CM.4 What the hurdle-Gamma did not solve

The ordinary tail/body was close to calibrated, but the far left tail remained inflated.

At \(p<.001\), the hurdle-Gamma rates remained approximately 1.28–1.45 times nominal across the four datasets. At the
\(10^{-5}\)-scale thresholds relevant to the first BH rejection with tens of thousands of genes, the residual inflation
was much larger.

Consequently complete-null BH FDR was not repaired by the hurdle-Gamma alone:

```text
Pickrell:
    P(any BH(.05) rejection) remained ~0.667,
    although mean BH discoveries fell from ~2.70 to ~1.33.

Li2020:
    BH discovery counts and benchmark FPR fell dramatically,
    but the first-order-statistic tail remained too heavy.
```

Thus the hurdle-Gamma solved the dominant **coverage-conditioned body/ordinary-tail problem**, not the ultra-extreme
order-statistic tail.

<a id="cn-joint-s-grid"></a>
# CN. Joint post-hurdle search over DEB shrinkage strength

## CN.1 Why the \(s\)-grid had to be repeated

The earlier Pickrell \(s\)-grid optimized DEB shrinkage while retaining the incorrect global \(\chi_1^2\) reference.
Once the downstream reference law changed, the old \(s\approx.75\) optimum could not simply be carried over.

The correct experiment therefore searched the combined system:

\[
s
\longrightarrow
\theta_F(s),\theta_0(s)
\longrightarrow
Q_s
\longrightarrow
\text{new cross-fitted hurdle-Gamma fit}
\longrightarrow
p_s.
\]

For every tested \(s\), the hurdle-Gamma was independently re-fit. The current fitted-DEB + hurdle-Gamma method was
retained as a separate baseline. [S42]

## CN.2 Search grid

The joint audit tested:

\[
s\in
\{
0,.1,.2,.3,.4,.5,.6,.7,.75,.8,.85,.9,.925,.95,.975,.99,.995,.999,1
\}.
\]

The resulting curves remained clearly non-flat. This was important: if DEB shrinkage had merely been compensating for
the wrong reference law, the post-hurdle performance would have become nearly invariant to \(s\).

It did not.

Therefore the project now has evidence for two separable effects:

\[
\boxed{
\text{DEB precision-shrinkage error}
\quad+\quad
\text{conditional reference-law error}.
}
\]

## CN.3 Equal-dataset calibration criterion

To avoid choosing \(s\) by Li2020 benchmark FPR minimization, the project constructed an equal-dataset null-calibration
criterion over Pickrell, AdLC, OC, and TNBC.

For dataset \(d\), shrinkage \(s\), and nominal threshold \(t\),

\[
L_d(s)
=
\operatorname{mean}_{t\in\{10^{-4},10^{-3},10^{-2},.05\}}
\left|
\log_{10}
\frac{\widehat P(p\le t)}{t}
\right|.
\]

The combined criterion was:

\[
L_{\rm combined}(s)
=
\frac14\sum_d L_d(s).
\]

The displayed selection table was:

| Criterion | \(s\) | Mean log10 calibration error | Typical multiplicative error | Worst-dataset log10 error | Worst-dataset multiplicative error |
|---|---:|---:|---:|---:|---:|
| Current fitted DEB + hurdle-Gamma | — | 0.184539 | 1.529464 | 0.216943 | 1.647945 |
| Best equal-dataset mean calibration | **0.60** | **0.140254** | **1.381190** | **0.198889** | **1.580844** |
| Best minimax calibration | **0.60** | **0.140254** | **1.381190** | **0.198889** | **1.580844** |

The 10%-of-best plateau was extremely broad:

\[
\boxed{
s\in\{.1,.2,.3,.4,.5,.6,.7,.75,.8\}.
}
\]

The mean log10 calibration loss fell by roughly 24% relative to current fitted DEB + hurdle-Gamma. [S43]

## CN.4 Interpretation

The important result is not that \(s=.600\) is uniquely optimal. It is that:

```text
near-total shrinkage is systematically worse;
moderate shrinkage is broadly robust;
the optimum is interior and very broad.
```

Both the equal-dataset mean and minimax criteria independently selected \(s=.60\), which lies near the center of the
shared good region.

The current leading development value is therefore:

\[
\boxed{s=.60.}
\]

Equivalently,

\[
\frac{\alpha_0}{a}
=
\frac{s}{1-s}
=
1.5.
\]

This is simpler than allowing the empirical-Bayes concentration to diverge toward effectively infinite shrinkage. The
fitted prior mean can still be retained while the amount of borrowing is controlled.

<a id="co-generalized-gamma"></a>
# CO. Positive-tail refinement: generalized Gamma

## CO.1 Why the Gamma positive component was tested next

After the hurdle-Gamma and \(s=.60\) corrections, the body and ordinary tail were close to calibrated, while residual
inflation became progressively more visible toward the far tail.

That pattern suggested that the positive component might need one additional shape degree of freedom rather than
another coverage-scale correction.

The next candidate was the generalized Gamma:

\[
f(q)
=
\frac{c}{b\Gamma(a)}
\left(\frac qb\right)^{ac-1}
\exp\left[-(q/b)^c\right],
\qquad q>0.
\]

At

\[
c=1,
\]

this nests the ordinary Gamma exactly.

The test fixed \(s=.60\), kept the hurdle zero mass \(\pi_0(x)\), and used one common \(c\) per training fold. The
coverage-specific positive-component \(a(x)\) and \(b(x)\) were reparameterized for each candidate \(c\) to preserve the
observed positive-Q mean and variance in each training coverage bin. Thus \(c\) was asked specifically to explain
higher-order/tail shape rather than merely rescale the distribution. It was estimated by profile likelihood on the
training genes only. [S44]

## CO.2 The three Li2020 datasets independently preferred a heavier-than-Gamma positive tail

The fitted cross-fold \(c\) summaries were approximately:

| Dataset | Median \(c\) | Q25 | Q75 | Median per-gene log-likelihood gain vs Gamma |
|---|---:|---:|---:|---:|
| Pickrell | 0.975342 | 0.913419 | 1.136914 | 0.001175 |
| AdLC | 0.700465 | 0.553096 | 0.790746 | 0.022346 |
| OC | 0.672566 | 0.611756 | 0.707252 | 0.027523 |
| TNBC | 0.701534 | 0.643687 | 0.831319 | 0.022284 |

None of the displayed fits piled up at the lower \(c\) bound.

Under this parameterization:

```text
c = 1:
    ordinary Gamma

c < 1:
    heavier positive upper tail than Gamma
```

Pickrell therefore said approximately “Gamma is adequate,” whereas AdLC, OC, and TNBC independently converged near
\(c\approx.67-.70\).

That convergence across three separate Li2020 matrices is one of the strongest mechanistic results in the current
tail audit.

## CO.3 The correction became stronger as the nominal tail became more extreme

At fixed \(s=.60\):

| Dataset | \(p<10^{-4}\): Gamma → GenGamma | \(p<10^{-3}\): Gamma → GenGamma | \(p<.01\): Gamma → GenGamma |
|---|---:|---:|---:|
| AdLC | 0.000375 → **0.000219** | 0.001351 → **0.000903** | 0.008888 → **0.008537** |
| OC | 0.000223 → **0.000095** | 0.001350 → **0.000777** | 0.010388 → **0.009862** |
| TNBC | 0.000324 → **0.000176** | 0.001235 → **0.000851** | 0.009019 → **0.008601** |

This is exactly the desired qualitative behavior.

The generalized Gamma barely moved the already-good \(p<.01\) body, changed the \(p<.001\) tail more, and changed the
\(10^{-4}\) tail still more. It therefore did **not** behave like a generic conservative rescaling.

Pickrell was almost unchanged:

```text
raw p<1e-4:
    0.000106 -> 0.000118

raw p<.001:
    0.000790 -> 0.000791

raw p<.01:
    0.008170 -> 0.008286
```

which is consistent with its fitted \(c\approx1\).

## CO.4 Li2020 benchmark FPR improved without any worsened split

At fixed \(s=.60\), mean Li2020 benchmark FPR changed approximately:

| Dataset | Hurdle-Gamma | Hurdle-Generalized-Gamma |
|---|---:|---:|
| AdLC | 0.000156 | **0.000038** |
| OC | 0.000063 | **0.000017** |
| TNBC | 0.000095 | **0.000032** |

Paired split counts were:

```text
AdLC:
    14 improved
     1 equal
     0 worsened

OC:
     7 improved
    11 equal
     0 worsened

TNBC:
    13 improved
     3 equal
     0 worsened
```

Thus the positive-tail shape correction improved or tied every selected Li2020 split.

## CO.5 Complete-null BH FDR improved substantially in Li2020 but remained elevated

The complete-null \(P(\mathrm{BH}(.05)\text{ rejects anything})\) comparison was:

| Dataset | Hurdle-Gamma | Hurdle-Generalized-Gamma |
|---|---:|---:|
| Pickrell | 0.166667 | 0.166667 |
| AdLC | 1.000000 | **0.733333** |
| OC | 0.444444 | **0.277778** |
| TNBC | 1.000000 | **0.562500** |

This is a large improvement in the Li2020 extreme tail, but it is not nominal control.

<a id="cp-ultra-tail"></a>
# CP. The remaining ultra-extreme-tail problem

## CP.1 The remaining error is now concentrated beyond the ordinary \(10^{-3}\) tail

After fixed \(s=.60\) plus the generalized-Gamma hurdle law, the Li2020 \(p<.001\) rates were:

```text
AdLC: 0.000903
OC:   0.000777
TNBC: 0.000851
```

These are at or below the nominal 0.001 level.

Yet complete-null BH FDR remained far above 0.05.

Therefore the current unresolved defect is **not** broad \(p<.01\) or even ordinary \(p<.001\) anti-conservatism. It is
concentrated in the much smaller order-statistic region that determines the first BH rejection:

\[
\boxed{
p\sim10^{-6}\text{ to }10^{-5}
}
\]

for these gene counts.

This is now a much narrower problem than the original coverage hump.

## CP.2 The next mechanistic question

The next diagnostic should use held-out generalized-Gamma p-values and determine whether the remaining excess is:

```text
A. a smooth, continuous tail that is still heavier than generalized Gamma;
```

or

```text
B. a small contaminating / outlier component that generates a few
   exceptionally extreme statistics.
```

Those hypotheses imply different final models. Adding another smooth shape parameter is appropriate only for A. A
mixture/contamination model is the more natural response to B.

The project should therefore inspect:

- empirical survival ratios from roughly \(10^{-3}\) through the first-BH threshold;
- split-level minimum-p distributions;
- the identities and coverage/information properties of genes creating the first few order statistics;
- whether the same genes or gene states recur across folds/splits;
- whether the excess is smoothly distributed or concentrated in a tiny number of points.

## CP.3 The negative-\(T_{\rm raw}\) theory is not automatically closed

The hurdle model correctly acknowledges the observed point mass created by:

\[
Q=\max(T_{\rm raw},0),
\]

but it does not derive or prove that clipping is the uniquely correct theoretical test construction.

Therefore the historical negative-\(T_{\rm raw}\) question remains open at the likelihood-theory level. The empirical
hurdle law makes the induced point mass explicit and calibratable under complete-null data; it does not by itself prove
the clipping rule.

## CP.4 Production estimation in the presence of alternatives remains a separate problem

All of the hurdle-Gamma / generalized-Gamma results above were developed on complete-null or within-population-null
benchmarks, where treating the training-gene distribution as null is meaningful.

A real DGE experiment can contain substantial alternatives. A production conditional-null estimator will therefore need
a principled way to protect the fitted \(\pi_0(x)\), positive-component shape, and far-tail parameter from alternative
contamination.

Possible production mechanisms include robust fitting or posterior-null weighting, but no such alternative-containing
implementation has yet been validated. The null-benchmark success should therefore not be described as a finished
general-user empirical-null estimator.

<a id="cq-updated-development-architecture"></a>
# CQ. Updated development architecture and next validation

## CQ.1 Current leading null-calibration development branch

The current strongest two-group DGE development branch is now:

```text
WEB-DP r_g
+
first-order C'(1) ordinary residual coordinate
+
seam-safe exact-when-feasible 20M C-prime backend
+
global kappa_c under the existing ordinary null-rich policy
+
support-feasible null geometry
+
mandatory exact normalized-DNB theta fallback when R_F<=0
+
moderate ordinary full/null DEB shrinkage with fixed s = .60
+
Q = max(T_raw, 0)
+
coverage-conditioned hurdle reference law
+
generalized-Gamma positive component with one common tail parameter c
```

This is a **development branch**, not yet the archived production method.

The statistical roles are now more sharply separated:

\[
\boxed{
r_g
=
\text{NB mean-variance curvature}
}
\]

\[
\boxed{
s
=
\text{amount of borrowing in ordinary full/null residual precision}
}
\]

\[
\boxed{
\kappa_c
=
\text{contrast-direction geometry}
}
\]

\[
\boxed{
\pi_0(x),\,a(x),\,b(x),\,c
=
\text{conditional reference law of the final null statistic}
}
\]

These should not be collapsed into interchangeable scale parameters.

## CQ.2 What is now demoted

The following are not leading fixes for the coverage problem:

```text
coverage-specific kappa_c used directly in the test
weighted-reference-CDF DB remapping
logit-affine contextual DB remapping
a universal Pickrell-derived prior-target shape
DEB concentration capping by itself
forcing the old s=.75 value after changing the reference law
```

Each supplied useful diagnostic evidence, but none matched the cross-dataset success of the conditional hurdle law.

## CQ.3 What has genuinely worked

The sequence of successful corrections is now:

\[
\boxed{
\text{conditional hurdle-Gamma}
}
\]

for the dominant coverage/body problem,

followed by:

\[
\boxed{
s=.60
}
\]

as a moderate, broad-plateau second-order shrinkage choice,

followed by:

\[
\boxed{
\text{generalized-Gamma positive tail}
}
\]

for a substantial part of the residual Li2020 \(10^{-4}\)–\(10^{-3}\) mismatch.

This layered interpretation is supported by ablation:

- hurdle-Gamma produced the large cross-dataset change;
- the post-hurdle \(s\)-grid remained non-flat, showing a distinct shrinkage effect;
- generalized Gamma changed the far tail much more than the \(p<.01\) body, showing a distinct tail-shape effect.

## CQ.4 Immediate next sequence

The recommended order is now:

```text
1. Keep s=.60 frozen for the next tail diagnostic.

2. Keep the hurdle zero-mass and coverage dependence unchanged.

3. Diagnose the held-out generalized-Gamma p-value survival curve specifically
   from p~1e-3 down through the first-BH region.

4. Determine continuous residual tail versus sparse contamination/outliers.

5. Only then decide whether a second smooth tail parameter or a small mixture
   component is warranted.

6. Freeze the resulting null-reference model before testing an
   alternative-containing benchmark.

7. Build and validate a null-protected estimator for real-signal data rather
   than fitting the complete gene distribution as though it were null.

8. Revisit the full Li2020 40v40 and then all-sample-size blocks once the fresh
   20M run is complete.

9. Preserve archived MAQC/RNAontheBENCH results under their original method
   identity; do not silently rescore them as though the new conditional-null
   model had been used historically.
```

The conditional-null audits use deterministic coverage-balanced gene folds so held-out genes do not estimate their own

The exact branch was numerically tiny in the displayed null splits and was not the driver of the broad coverage hump.

Once the reference law changed, \(s=.75\) could no longer be treated as the previously established optimum. The later
joint grid correctly re-estimated the hurdle law separately for every \(s\), and both combined criteria selected

The project should therefore cite the joint-grid result, not the intermediate fixed-\(.75\) run, when describing the

## CR.4 Generalized-Gamma numerical identity check

The generalized-Gamma implementation was explicitly checked at:

\[
c=1,
\]

where it recovers the ordinary Gamma moment map and survival function numerically.

The candidate therefore nests the previous hurdle-Gamma reference exactly. This makes the likelihood gain and fitted
\(c\) values interpretable as a test of an additional positive-tail shape degree of freedom rather than a comparison
between unrelated families.

<a id="cs-ultra-tail-recurrence"></a>
# CS. Ultra-extreme-tail recurrence and the sparse-contamination diagnosis

## CS.1 The generalized-Gamma repair narrowed the problem to the first few order statistics

Part V established that, after fixed \(s=.60\) plus the coverage-conditioned hurdle-generalized-Gamma law, the ordinary
\(10^{-3}\) tail was near nominal or slightly conservative in AdLC, OC, and TNBC, while complete-null BH FDR remained
substantially elevated. The next audit therefore looked specifically below approximately \(10^{-3}\), down through the
first-BH region.

The critical new observation was that the residual excess was **not broad across hundreds of genes**. The inflation was
concentrated in the first few order statistics, and the same or similar genes could recur among the most extreme ranks across
multiple randomized splits.

This favored the second hypothesis posed in Part V:

```text
not:
    one smoothly heavier positive tail affecting the entire conditional distribution

but increasingly:
    a tiny recurrent / leverage-sensitive subset that can generate
    ultra-extreme statistics under particular sample assignments.
```

The generalized-Gamma remains an informative nested tail-shape diagnostic, but the ultra-extreme audit showed that another
global smooth shape parameter is not automatically the right response.

## CS.2 Recurrence was a gene-state diagnostic, not a blacklist rule

The recurrence tables counted whether genes repeatedly appeared among the first few generalized-Gamma p-value order
statistics across frozen null splits. Recurrence was treated as evidence about the **state of the statistic under certain
sample configurations**, not as evidence that those genes should be permanently excluded.

The recurrent genes were enriched for sample-imbalance coordinates that describe how concentrated a gene's normalized
expression is in one or a few physical samples. This motivated directly conditioning the null law on sample dominance rather
than deleting recurrent genes or adding ad hoc p-value floors. [S45]

<a id="ct-sample-dominance"></a>
# CT. Sample dominance as a second conditional-null coordinate

## CT.1 Definition and motivation

The leading additional coordinate was the normalized top-sample contribution, operationally a measure of the largest
sample's share of a gene's normalized total expression. Denote the coverage coordinate by \(x_g\) and sample dominance by
\(d_g\).

The one-dimensional conditional null was already:

\[
Q_g\mid x_g
\sim
\pi_0(x_g)\delta_0+
[1-\pi_0(x_g)]\Gamma\{k(x_g),\theta_\Gamma(x_g)\}.
\]

The two-dimensional diagnostic asked whether the positive null law should instead depend on:

\[
\boxed{Q_g\mid x_g,d_g.}
\]

This was motivated by the recurrent extreme genes, not by the NanoString truth labels.

## CT.2 Two-dimensional conditioning helped Li2020 but could overfit Pickrell

Coverage-plus-dominance conditioning materially reduced the high-dominance extreme-tail excess in the Li2020 40v40
null splits. It also preserved the already solved coverage-body behavior.

However, the unpooled two-dimensional law was not uniformly better. Pickrell, where the dominance coordinate was less
necessary, became noisier and could have worse extreme-tail/BH behavior. Cross-fitting prevents a held-out gene from fitting
its own reference distribution, but it does **not** eliminate variance from estimating a needlessly high-dimensional
conditional law.

The conclusion was therefore:

\[
\boxed{
\text{sample dominance is a real missing coordinate in some nulls,}
\quad
\text{but it should not be forced at full strength in every dataset.}
}
\]

This directly motivated partial pooling. [S46]

<a id="cu-partial-pooling"></a>
# CU. Partial pooling of coverage-only and coverage-plus-dominance null laws

## CU.1 Distribution-level pooling

The safer construction retained both candidate conditional distributions:

```text
F_1D:
    coverage-only hurdle-Gamma

F_2D:
    coverage + sample-dominance hurdle-Gamma
```

and formed a convex distributional pool:

\[
\boxed{
F_{\rm pool}
=
(1-w)F_{1D}
+
wF_{2D},
\qquad 0\le w\le1.
}
\]

The weight \(w\) was not fixed from the final held-out genes. It was learned inside **nested cross-validation** from the
training data. Thus a dataset/split could use the dominance coordinate strongly when it was predictively useful while still
retaining a stabilizing contribution from the coverage-only law.

A key empirical observation was that even when the learned \(w\) was fairly high, a modest 1D component could materially
stabilize the first-BH extreme tail. This explains why the pooled model could outperform the unpooled 2D model even when
the data clearly supported a dominance effect.

## CU.2 ASH posterior-null weighting for alternative-containing data

The complete-null development fits could treat essentially the full ordinary population as null. A real DGE dataset cannot.

For the alternative-containing branch, the training objective was therefore weighted by the already-existing Stage-03 ASH
posterior-null probabilities:

\[
w_g^{ASH}=P(H_{0g}\mid\text{Stage-03 data}).
\]

These weights protect estimation of the conditional null law by reducing the influence of genes that M03 already considers
strongly alternative-like. They are **not** a new contrast-geometry estimator and do not redefine \(\kappa_c\).

This produced the frozen development architecture:

```text
current seam-safe 20M M03 backbone
+
universal ordinary-precision shrinkage s=.60
+
ASH-null-weighted conditional-null training
+
coverage-only hurdle-Gamma
+
coverage + sample-dominance hurdle-Gamma
+
nested-CV distributional partial pooling.
```

The positive family was returned to ordinary Gamma for this frozen branch. The earlier generalized-Gamma work remains
important evidence about Li2020 tail shape, but the partial-pooling experiment showed that much of the practical
ultra-extreme defect could be attacked through the missing dominance coordinate plus regularization rather than by stacking
additional global tail-shape complexity. [S46, S47]

<a id="cv-frozen-rnaotb"></a>
# CV. Frozen alternative-containing branch for RNAontheBENCH

## CV.1 Frozen method identity

The branch carried into RNAontheBENCH was frozen before NanoString truth was loaded:

```text
method:
    WEB-DNB-M03-2026-10-04-s060-ASHnull-
    partialpool-coverage-dominance-hurdleGamma-v2

M03 version:
    2026-10-01-M03-seam-safe-cprime-20m-v1

ordinary precision:
    s = .60

positive conditional family:
    Gamma

conditional coordinates:
    coverage
    + partially pooled sample dominance

null protection:
    Stage-03 ASH posterior-null weights
```

The input contained **15,583 genes and nine samples**, with six WBS and three Dup7 samples.

The truth file was deliberately unavailable to the fitting step. The fitting code wrote the frozen genome-wide results and a
SHA-bearing manifest before scoring was permitted to open NanoString truth. [S48]

## CV.2 Two implementation bugs were fixed without changing the frozen statistical result

The first frozen runner failed after M03 Stage 05 because the new ASH-null-law code checked the Stage-03 ASH weight vector
against the original 15,583-row input universe. M03 Stage 01 had removed all-zero genes, so the ASH weights were aligned to
the post-filter `core["gene_ids"]` universe instead.

The correction:

```text
aligned ASH weights to Stage-01 analyzed genes;
re-expanded frozen results to the original input universe afterward;
marked all-zero excluded genes explicitly;
did not refit the already-completed Stage-01--05 M03 state.
```

The second failure occurred **after the truth-blind freeze had completed**. The scoring code assumed the NanoString truth CSV
contained a literal column named `truth`. The actual columns were:

```text
gene_id
nanostring_p
nanostring_log2FC
truth_positive
```

The scoring-resume version reused the exact SHA-locked pre-truth freeze and reconstructed the historical reference truth from
the already-established rule:

\[
\boxed{\text{NanoString raw }p<.01.}
\]

It required the historical invariant:

```text
150 truth genes
24 positives
126 negatives
```

and refused to refit M03 or the conditional-null model after truth had been opened.

These failures reinforce the project rule:

```text
syntax checked
!=
runtime/schema validated
!=
scientifically validated.
```

They were implementation/alignment failures, not statistical-method failures. [S48]

<a id="cw-rnaotb-frozen-result"></a>
# CW. RNAontheBENCH frozen result

## CW.1 The frozen correction preserved the operating point but did not improve it

All 150 NanoString truth genes were present in the RNA-seq input and all 150 were analyzed by M03. At the historical

| Method | TP | FP | TN | FN | Sensitivity | Specificity | Precision | Reference FDP | MCC | AUC |
|---|---:|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| Current seam-safe 20M M03 | 18 | 11 | 115 | 6 | 0.7500 | 0.9127 | 0.6207 | 0.3793 | 0.6152 | **0.9431** |
| Frozen \(s=.60\) + ASH-null pooled hurdle-Gamma | 18 | 11 | 115 | 6 | 0.7500 | 0.9127 | 0.6207 | 0.3793 | 0.6152 | 0.9398 |

Thus the frozen conditional-null correction produced:

\[
\boxed{\text{no operating-point improvement on RNAontheBENCH.}}
\]

It preserved all 18 current true calls but also all 11 current reference false calls. Ranking changed slightly in the wrong

## CW.2 Sample-dominance conditioning did not erase concentrated true signal

Truth-positive retention by sample-dominance region was unchanged:

| Dominance region | Truth positives | Current called | Frozen called |
|---|---:|---:|---:|

| 90th–98th percentile | 7 | 7 | 7 |
| at/above 98th percentile | 7 | 6 | 6 |

Therefore the feared failure mode:

```text
"dominance conditioning will automatically treat concentrated real biology
as null pathology and erase it"
```

did **not** occur in this first frozen alternative-containing test.

At the same time, the correction did not preferentially eliminate NanoString-reference false positives either.

## CW.3 Interpretation

The null-benchmark discovery remains valid:

\[
\text{sample dominance helps identify an important randomized-null failure mode.}
\]

But the stronger claim:

\[
\text{therefore the same correction improves real DGE discrimination}
\]

failed its first frozen RNAontheBENCH test.

This branch should therefore be retained as a null-calibration / conditional-reference candidate rather than promoted as a
universally superior final p-value law on the basis of the null benchmarks alone. [S48]

<a id="cx-rnaotb-s-grid"></a>
# CX. Development-only RNAontheBENCH universal-\(s\) scans

## CX.1 RNAontheBENCH became development data after the frozen score was observed

After the frozen result was scored, Sean explicitly chose to use RNAontheBENCH to ask what universal reciprocal-precision
shrinkage fraction \(s\) the NanoString benchmark itself prefers.

From that point forward, RNAontheBENCH is **development data for \(s\)**. Any \(s\) selected from these scans requires a
different untouched benchmark for external validation.

For every candidate \(s\), the analysis:

```text
reused the completed M03 Stage 01-05 state;
rebuilt theta_F(s), theta_0(s), and Q_s;
re-fit the ASH-null-weighted nested-CV partial-pooled hurdle-Gamma;
then scored NanoString truth.
```

Thus downstream calibration was re-estimated at every \(s\); the scan did not simply rescale a fixed p-value vector.

## CX.2 Coarse grid: truth-based and truth-blind objectives diverged

The first grid produced:

| Criterion | \(s\) | TP | FP | Sensitivity | Reference FDP | MCC | AUC |
|---|---:|---:|---:|---:|---:|---:|---:|
| Current fitted-DEB M03 | fitted | 18 | 11 | 0.7500 | 0.3793 | 0.6152 | 0.9431 |
| Best development MCC | **0.05** | 17 | **7** | 0.7083 | **0.2917** | **0.6528** | 0.9315 |
| Lowest FDP with sensitivity \(\ge\) current | **0.55** | **18** | **10** | **0.7500** | **0.3571** | **0.6310** | 0.9385 |
| Best development AUC | 0.99 | 18 | 14 | 0.7500 | 0.4375 | 0.5717 | **0.9494** |
| Best truth-blind nested-CV fit | **1.00** | 18 | 14 | 0.7500 | 0.4375 | 0.5717 | 0.9494 |

This divergence is scientifically important.

The conditional-null predictive likelihood increasingly preferred stronger shrinkage and selected approximately \(s=1\),
while the best NanoString operating-point MCC occurred at very weak shrinkage. Therefore:

\[
\boxed{
\text{best empirical-Bayes / null predictive fit}
\ne
\text{best shrinkage for hypothesis-testing discrimination}.
}
\]

This is a direct warning against using the DEB hyperparameter likelihood alone to decide how much gene-specific precision
structure should survive into the test.

## CX.3 Fine grid around \(s=.05\): there was no unique local optimum

A 41-point scan from:

\[
s=0
\quad\text{to}\quad
.10
\]

in increments of \(0.0025\) showed that the apparent \(s=.05\) optimum was actually a broad discrete call plateau.

From:

\[
\boxed{s=0\text{ through }.0725}
\]

the operating point remained:

\[
\boxed{17\ \text{TP},\quad 7\ \text{FP}.}
\]

At \(s=.075\), one reference false positive crossed the raw-\(p=.01\) boundary and the result became 17 TP / 8 FP.

Thus the meaningful information was the **boundary location**, not a unique optimum at \(s=.05\).

## CX.4 Full \(0\)–\(1\) scan at step \(0.0025\)

The 401-point scan exposed a sequence of discrete operating regimes:

| \(s\) interval | TP | FP |
|---|---:|---:|
| 0–0.0725 | 17 | 7 |
| 0.075–0.1075 | 17 | 8 |
| 0.110–0.125 | 17 | 9 |
| 0.1275–0.1575 | 17 | 10 |
| 0.160 | 16 | 10 |
| 0.1625–0.1825 | 17 | 10 |
| 0.185–0.2225 | 17 | 11 |
| 0.225–0.3725 | 18 | 11 |
| 0.375–0.490 | 18 | 12 |
| 0.4925–0.5475 | 18 | 11 |
| 0.550–0.5725 | 18 | **10** |
| 0.575–0.640 | 18 | 11 |
| 0.6425–0.660 | 18 | 12 |
| 0.6625–0.7225 | 18 | 11 |
| 0.725–0.730 | 18 | 12 |
| 0.7325–0.8075 | 18 | 13 |
| 0.810–1.000 | 18 | 14 |

The operating point is therefore highly **piecewise constant** and not monotone in \(s\).

The two most interesting regimes were:

\[
s\lesssim.0725:
\quad
17\text{ TP}/7\text{ FP},
\]

which sacrifices one current true positive to remove four current false positives, and

\[
s\approx.55-.5725:
\quad
18\text{ TP}/10\text{ FP},
\]

which preserves all current true positives while removing one false positive.

Neither region should be treated as a universal tuning result. The scan is development-only and the 150-gene panel creates
discrete threshold plateaus.

The nested 2D-pooling weight generally rose smoothly with \(s\), from roughly the low-.2 range at very small \(s\) toward
nearly one as \(s\to1\). The jagged TP/FP path therefore was not simply an unstable pooling-weight optimizer. Specific genes
were crossing the \(p=.01\) boundary as the precision/test geometry changed. [S49, S50]

<a id="cy-boundary-gene-trajectories"></a>
# CY. Boundary-gene trajectories and the meaning of shrinkage

## CY.1 The boundary analysis identified the actual genes causing the steps

The full \(s\)-grid caches were reused without refitting. Every adjacent-\(s\) call change on the NanoString truth panel was
identified, and for each transition gene the following trajectories were reconstructed over all 401 \(s\) values:

\[
p_g(s),\quad
Q_g(s),\quad
\theta_{F,g}(s),\quad
\theta_{0,g}(s),\quad
\frac{\theta_{0,g}(s)}{\theta_{F,g}(s)}.
\]

The diagnostic also retained coverage, sample dominance, ASH posterior-null weight, NanoString truth status, and the exact
grid points at which a gene entered or left the \(p<.01\) call set.

Examples visible in the completed plots included:

```text
TBP        NanoString positive   transition near s=.160/.1625
GTF2IRD1   NanoString positive   TP gained near s=.1625
CLDN3      NanoString positive   TP gained near s=.225

ROR1       NanoString negative   FP gained near s=.0725
MEIS3      NanoString negative   FP gained near s=.1075
DPPA2      NanoString negative   FP gained near s=.1275
STX1B      NanoString negative   FP gained near s=.1825
PIWIL2     NanoString negative   FP gained near s=.3725
DUSP3      NanoString negative   FP lost near s=.550
TRPC5      NanoString negative   FP gained near s=.5725
```

The aggregate TP/FP changes were exactly reconstructable from the gene-level events.

## CY.2 Low raw precision and high raw precision respond in opposite directions

The plots exposed a simple mechanistic pattern.

For genes whose raw ordinary precision is low relative to the DEB prior target, increasing \(s\) raises their precision toward
the common target. Several such genes became **more significant** as shrinkage increased. The displayed examples included
GTF2IRD1, CLDN3, MEIS3, DPPA2, STX1B, and especially PIWIL2.

For genes whose raw precision is high relative to the target, increasing \(s\) pulls precision downward. Examples such as
TBP, ROR1, and DUSP3 became **less significant** over substantial portions of the path.

This means that DEB shrinkage is doing both:

\[
\boxed{\text{raising low gene-specific precision}}
\]

and

\[
\boxed{\text{lowering high gene-specific precision}}
\]

toward a common prior target. Whether that helps or hurts the benchmark depends on whether the gene-specific raw precision
was informative or merely noisy.

## CY.3 Why \(s\) changes more than estimator variance

For ordinary genes:

\[
\theta_F(s)
=
\frac{a_F}
{(1-s)R_F+s\,a_F/m_F},
\qquad
\theta_0(s)
=
\frac{a_0}
{(1-s)R_0+s\,a_0/m_0}.
\]

At:

\[
s=0,
\]

the raw estimates satisfy:

\[
\theta_FR_F=a_F,
\qquad
\theta_0R_0=a_0.
\]

Thus the residual-magnitude terms are self-normalized by each gene's own residual precision.

At:

\[
s=1,
\]

the precisions approach the common prior means:

\[
\theta_F\approx m_F,
\qquad
\theta_0\approx m_0,
\]

so the gene-specific residual magnitudes re-enter the statistic under nearly common scales.

Therefore moving \(s\) from zero to one is not merely a harmless variance-reduction operation. It continuously changes the
effective geometry of the test from a strongly gene-self-normalized residual comparison toward a common-scale residual
comparison.

## CY.4 The \(s=.160\) one-point anomaly was downstream, not numerical chaos

TBP illustrated a very narrow loss/re-gain around \(s=.160\)–.1625. Its \(Q(s)\), \(\theta_F(s)\), and \(\theta_0(s)\)
trajectories were smooth. The threshold instability came from independently re-estimating the conditional null law at every
\(s\), which could move a gene sitting almost exactly at raw \(p=.01\) across the boundary.

This is another reason not to optimize a decimal-level universal \(s\) against one small external panel.

## CY.5 These RNAontheBENCH transition genes were not mainly a sample-dominance phenomenon

The displayed transition genes had ordinary-to-moderate dominance values, roughly in the 0.13–0.28 range, rather than the
extreme leverage pattern that motivated the Li2020 dominance correction.

Several also had low ASH posterior-null weights, including PIWIL2 near zero. Such genes contribute little to fitting the
ASH-protected conditional null law. Their \(s\)-driven transitions therefore point primarily to the **upstream precision /
\(Q\) layer**, not to the dominance-reference-law fit.

This cleanly separates two problems:

\[
\boxed{
\text{sample-dominance conditioning}
=
\text{reference-law / recurrent-null-tail problem}
}
\]

versus

\[
\boxed{
\text{precision shrinkage strength}
=
\text{how much gene-specific residual scale survives into }Q.
}
\]

[S51]

<a id="cz-updated-position-oct5"></a>
# CZ. Updated scientific position and next step

## CZ.1 What is now supported

The latest evidence supports all of the following simultaneously:

1. **The seam-safe exact-when-feasible 20M M03 count/normalizer/support backbone remains useful.**
2. **The large coverage-dependent null hump is principally a conditional-reference-law problem.**
3. **Sample dominance explains an additional recurrent ultra-extreme-tail failure mode in the Li2020 randomized nulls.**
4. **A fully unpooled 2D coverage-plus-dominance empirical null is too variable to force universally.**
5. **Nested-CV partial pooling of 1D and 2D hurdle-Gamma laws is a safer way to borrow the dominance coordinate.**
6. **Stage-03 ASH posterior-null weighting gives a natural way to protect that conditional-null fit when real alternatives are present.**
7. **The first frozen RNAontheBENCH transfer did not improve the declared 18 TP / 11 FP operating point, but it also did not destroy high-dominance true signal.**
8. **RNAontheBENCH development scans show that fitted-DEB / truth-blind predictive criteria can prefer near-total shrinkage even when the external operating point prefers much weaker or moderate shrinkage.**
9. **The universal-\(s\) objective is broad and discrete; decimal-level tuning is not scientifically meaningful.**
10. **Gene-level trajectories show why: the same global \(s\) can rescue one gene and create a false positive in another because it moves low and high raw precisions toward the same target.**

## CZ.2 Current conceptual decomposition

The two-group DGE development architecture should now be thought of as:

\[
\boxed{
\text{count/mean/curvature model}
\rightarrow
\text{gene-specific raw residual precision}
\rightarrow
\text{reliability-controlled precision borrowing}
\rightarrow
Q
\rightarrow
\text{conditional null law in coverage/dominance space}.
}
\]

The previous fixed-\(s=.60\) rule remains a useful null-calibration development setting and the frozen RNAontheBENCH
branch should remain archived under that identity. It should **not** yet be called the final universal precision rule.

## CZ.3 The next scientific question is precision reliability, not another global \(s\) search

The boundary trajectories strongly suggest that the next precision-layer question should be:

\[
\boxed{
\text{How reliable is gene }g\text{'s raw precision estimate?}
}
\]

A useful future \(s_g\) would need to depend on **pre-test reliability information** about \(R_g\), \(\theta_{\rm raw,g}\),
sample size, coverage/information, or precision-estimation uncertainty—not on whether the gene is significant and not on
external truth labels.

The desired behavior is:

```text
raw theta is noisy / weakly identified:
    borrow more toward the prior target

raw theta is stable / information-rich:
    preserve more of the gene-specific value
```

The current universal DEB concentration does not make that distinction.

## CZ.4 Validation implications

The frozen RNAontheBENCH result remains the valid external result for the October-4 branch:

```text
18 TP / 11 FP
AUC 0.9398
```

The subsequent \(s\)-grid and gene-transition experiments are development analyses and must not be folded back into that
external result.

Any reliability-adaptive precision rule developed from these findings must be frozen before testing on a genuinely untouched
alternative-containing benchmark.

The null-calibration hierarchy also remains important:

```text
coverage-conditioned hurdle law:
    dominant body / ordinary-tail correction

sample-dominance coordinate + partial pooling:
    recurrent ultra-extreme null-tail stabilization

precision reliability / shrinkage:
    controls how much gene-specific residual information enters Q
```

This is the current boundary between what has been demonstrated, what has failed, and what remains to be solved. [S45–S52]

This continuation records the precision-layer work after the 10:12 AM cutoff. It begins exactly where Part VI ended: the
project had established that one universal shrinkage fraction could move different genes in opposite inferential directions
and had therefore reopened the question of **gene-specific precision reliability**.

The sequence after that cutoff was:

```text
pre-detour DEB/WEB shrinkage reappraisal
    ->
mean-coverage-trend EB target
    ->
leave-one-sample-out raw-theta reliability
    ->
target-standardized raw-theta departure Z
    ->
parameter-free reliability-adaptive s_g = 1 - reliability percentile
    ->
Pickrell transfer
    ->
Li2020 / RNAontheBENCH transfer
    ->
decision NOT to adopt the percentile rule as a statistical method.
```

The most important distinction at the new cutoff is:

\[
\boxed{
\text{LOO precision reliability is empirically informative}
}
\]

but

\[
\boxed{
s_g=1-\text{reliability percentile}
\text{ is only a heuristic mapping and is not adopted as the method.}
}

<a id="da-pre-detour-precision-reappraisal"></a>
# DA. Reappraisal of the pre-detour DEB / WEB precision layer

## DA.1 Near-total DEB pooling is now treated as a substantive modeling problem

The reciprocal-precision identity remains:

\[
\frac{1}{\widehat\theta_g}
=
(1-s)\frac{1}{\widetilde\theta_g}
+
s\frac{1}{m},
\qquad
s=\frac{\alpha_0}{a+\alpha_0},
\qquad
\widetilde\theta_g=\frac{a}{R_g}.
\]

The earlier Pickrell work already showed that fitted DEB often has:

\[
s_F\approx1,
\qquad
s_0\approx1,
\]

so the ordinary gene-specific raw residual precision is nearly replaced by the common fitted prior target.

The later RNAontheBENCH and Li2020 development work strengthened the interpretation: the fitted empirical-Bayes objective
can prefer nearly complete pooling even when external operating-point behavior does not. This means that the pre-detour
ordinary DEB precision rule can no longer be treated as a settled, sophisticated gene-adaptive precision layer merely because
it is obtained from an empirical-Bayes likelihood.

The historical MAQC and Baik results remain valid results of the archived fitted-DEB method. What has changed is the claim
about the precision layer's general adequacy.

## DA.2 WEB is not algebraically identical to DEB, but switching names does not solve the pooling problem

WEB and DEB use different parameterizations:

\[
\widehat\theta_g^{WEB}
=
\frac{a_g}
{(1-w_g)S_g+w_gS_{\mathrm{target},g}},
\]

versus the Gamma-Gamma DEB form:

\[
\widehat\theta_g^{DEB}
=
\frac{a_g+\alpha_0}{R_g+\beta_0}.
\]

They can be written in closely related shrinkage form under the corresponding target/weight mapping, but they are not
automatically the same fitted statistical model.

The development record nevertheless already contains regimes where WEB weights are near one and DEB \(s\) is near one.
Therefore simply replacing DEB with WEB does not solve the deeper question. The issue is **how much gene-specific residual
precision survives population borrowing**, not the estimator label.

## DA.3 Raw precision contains useful information but is too noisy to use unregularized

RNAontheBENCH supplied the clearest development example:

```text
current fitted DEB:
    18 TP / 11 FP

raw theta:
    18 TP / 9 FP

fixed s=.60:
    18 TP / 10 FP
```

Raw precision therefore preserved information that heavy DEB pooling erased on that panel.

But the complete-null / within-population-null data showed the other side:

```text
Li2020 40v40 mean benchmark FPR

             current DEB     raw theta
AdLC            0.008471      0.013033
OC              0.001783      0.004203
TNBC            0.011087      0.017462
```

Thus:

\[
\boxed{
\text{raw precision contains useful gene-specific information}
\quad+\quad
\text{raw precision is too noisy to use without regularization}.
}
\]

The scientific problem is therefore not "DEB or no DEB." It is how to regularize residual precision according to how well
each gene's precision is identified. [S53]

<a id="db-mean-trend-eb"></a>
# DB. Mean-coverage-trend empirical Bayes: a principled-looking target change that did not solve the problem

## DB.1 Motivation and model

A natural alternative to a single global target was to let the reciprocal-precision target depend on mean coverage:

\[
\phi_g^{raw}
=
\frac{R_g}{a}
=
\frac{1}{\theta_g^{raw}},
\]

with:

\[
\phi_{\mathrm{target},g}
=
f(\bar\mu_g),
\]

and:

\[
\widehat\theta_g
=
\frac{a+\alpha_0}
{R_g+\alpha_0\phi_{\mathrm{target},g}}.
\]

Equivalently:

\[
\frac1{\widehat\theta_g}
=
(1-s)\frac1{\theta_g^{raw}}
+
s\,\phi_{\mathrm{target},g}.
\]

This preserves a coverage-dependent target even if the fitted concentration again drives \(s\) toward one.

The first implementation used an absolute ASH-weight-sum requirement to decide whether a coverage bin had enough null
information. That was corrected because a common rescaling of the ASH weights leaves the normalized weighted objective
unchanged. The corrected implementation used Kish effective sample size:

\[
ESS
=
\frac{(\sum_g w_g)^2}{\sum_g w_g^2},
\]

which is invariant to common weight scaling.

## DB.2 Mean-trend EB did not improve the null benchmarks

The resulting mean Li2020 40v40 benchmark FPRs were:

| Dataset | Current DEB | Mean-trend EB | Raw theta | Fixed \(s=.60\) |
|---|---:|---:|---:|---:|
| AdLC | 0.008471 | **0.009618** | 0.013033 | 0.009189 |
| OC | 0.001783 | **0.001756** | 0.004203 | 0.002179 |
| TNBC | 0.011087 | **0.015275** | 0.017462 | 0.011999 |

The paired result versus current DEB was:

```text
AdLC:
     0 improved
     0 equal
    15 worsened

OC:
     8 improved
     1 equal
     9 worsened

TNBC:
     0 improved
     0 equal
    16 worsened
```

Pickrell was also not rescued:

```text
mean raw p<.01:
    current DEB       0.013216
    mean-trend EB     0.013327
    raw theta         0.014829
    fixed s=.60       0.009167

complete-null BH FDR:
    current DEB       0.6667
    mean-trend EB     0.8000
    raw theta         0.5000
    fixed s=.60       0.0667
```

RNAontheBENCH likewise returned to the current operating point:

```text
current DEB:      18 TP / 11 FP
mean-trend EB:    18 TP / 11 FP
raw theta:        18 TP /  9 FP
fixed s=.60:      18 TP / 10 FP
```

## DB.3 What the negative result established

The target curve was mechanically preserved, but calibration did not improve. Therefore:

\[
\boxed{
\text{"shrink toward a coverage-dependent curve rather than a constant"}
\text{ is not sufficient.}
}
\]

The fit still often wanted very strong pooling. Moreover, fitting separate full and null target curves can itself introduce a
structured coverage-dependent full/null contrast into the final statistic when shrinkage is strong.

The problem therefore moved from "what should the target curve be?" to:

\[
\boxed{
\text{how much should gene }g\text{'s raw precision estimate be trusted?}
}
\]

[S53]

<a id="dc-loo-reliability"></a>
# DC. Leave-one-sample-out raw-precision reliability

## DC.1 Reliability was defined as stability under case deletion, not residual magnitude

The next audit deliberately separated:

```text
low precision
from
unreliable precision.
```

For each ordinary gene, the analysis held the Stage-01 normalization exposures and WEB-DP \(r_g\) fixed, left out one sample
at a time, refit the two-group NB full-model means on the remaining samples, and recomputed the full-model residual
half-deviance.

The genome-wide reliability proxy used:

\[
\theta_{S,g,-i}
=
\frac{a_{F,-i}}{S_{F,g,-i}},
\]

with jackknife variability of:

\[
\log\theta_{S,g,-i}.
\]

Raw instability was not ranked globally. Genes were compared within similar information regimes defined by coverage and
NB dispersion/curvature. The resulting conditional reliability percentile has the interpretation:

```text
near 1:
    unusually stable raw precision for genes with similar information

near 0:
    unusually fragile raw precision for genes with similar information.
```

For the small RNAontheBENCH transition-gene set, the expensive corrected quantity:

\[
R_{F,g,-i}
=
S_{F,g,-i}
-
C_{F,g,-i}
\]

was also recomputed directly so that the cheap genome-wide \(S\)-based stability could be checked against the corrected
\(R=S-C\) stability.

## DC.2 Pickrell showed strong enrichment of null extremes among unreliable genes

Across the 30 complete-null Pickrell splits, the median raw-\(p<.01\) fraction was approximately:

```text
lowest-reliability 10%:
    0.0291

other 90%:
    0.0122
```

The bottom reliability decile also contributed roughly:

```text
Top-10 smallest p-values:     20%
Top-50 smallest p-values:     22%
Top-100 smallest p-values:    21%
```

despite representing only 10% of genes.

This is direct evidence that precision-estimation instability is disproportionately represented in the extreme null tail.

The overall split-level relationship remained modest, with median Spearman reliability versus \(-\log_{10}p\) around
\(-0.047\). That is consistent with reliability being one tail mechanism rather than a complete explanation of gene ranking.

## DC.3 RNAontheBENCH showed the complementary pattern

The genes harmed by increasing global shrinkage—false positives gained or true positives lost—were more reliable on average
than the genes helped by shrinkage.

Approximate medians were:

| Gene class | Reliability percentile | \(SE_{\log\theta,S}\) | corrected \(SE_{\log\theta,R}\) |
|---|---:|---:|---:|
| Shrinkage harms | **0.509** | **0.624** | **0.657** |
| Shrinkage helps | 0.239 | 0.874 | 0.891 |
| Mixed | 0.134 | ~1.04 | ~1.08 |

The cheap genome-wide \(S\)-based stability ranking and direct corrected \(R=S-C\) stability were almost identical on the
15 transition genes:

\[
\boxed{
\rho_{\mathrm{Spearman}}
\approx0.9964.
}
\]

The saved-vs-direct normalizer discrepancy was also tiny, with median absolute \(C_F\) difference around \(5.8\times10^{-5}\).

Thus the expensive exact normalizer does not need to be recomputed genome-wide merely to rank precision stability.

## DC.4 The first EB conversion of reliability failed

An exploratory measurement-error-style rule was:

\[
s_g
=
\frac{v_g}{v_g+\tau^2},
\]

where \(v_g\) was LOO variance and \(\tau^2\) was estimated by a simple plug-in subtraction of measurement variance from
between-gene variance.

That crude \(\widehat\tau^2\) collapsed to zero, producing:

\[
s_g=1
\]

for essentially every gene.

This did **not** invalidate the reliability diagnostic. It showed that the first plug-in variance decomposition simply
recreated the original complete-pooling problem. [S54]

<a id="dd-departure-support-z"></a>
# DD. Target-standardized raw-\(\theta\) departure \(Z\): mechanistically informative, not a shrinkage-need score

## DD.1 Construction

The next diagnostic asked whether the relevant quantity was not stability alone but evidence that a gene's raw precision
really differs from an information-matched target.

A five-fold gene-cross-fitted, truth-blind and contrast-blind target was estimated from other genes in the same
coverage-by-alpha information stratum, with coverage-only and global fallbacks.

For each gene:

\[
\delta_g
=
\log\theta_g^{raw}
-
\log\theta_g^{target},
\]

and:

\[
\boxed{
Z_g
=
\frac{\delta_g}{SE_g^{LOO}}.
}
\]

Large \(|Z_g|\) means the gene's raw precision is a well-resolved departure from its information-matched target.

## DD.2 Pickrell falsified the idea that low departure support identifies the false extremes

Reliability alone had median AUC approximately:

\[
0.583
\]

for identifying current raw-\(p<.01\) null genes.

Using **low departure support** instead had median AUC only:

\[
0.404.
\]

Across the 30 Pickrell splits:

```text
Z-based AUC better:   0
equal:                0
worse:               30
```

with median AUC change approximately \(-0.175\).

The raw-\(p<.01\) rate generally increased rather than decreased with stronger departure support. Therefore:

\[
\boxed{
|Z_g|
\text{ is not a valid shrinkage-need score.}
}
\]

## DD.3 RNAontheBENCH showed that signed \(Z\) correctly describes what shrinkage will do

For separating shrinkage-harmed from shrinkage-helped transition genes:

```text
AUC from reliability:        ~0.722
AUC from |Z_S|:              ~0.694
AUC from |Z_R|:              ~0.694
```

Reliability therefore remained the better "should I trust the raw precision?" diagnostic.

But signed \(Z\) had a strong mechanistic association with the effect of moving from raw precision toward complete
shrinkage across all 150 truth-panel genes:

\[
\rho\left(
Z_g,
\Delta[-\log_{10}p]
\right)
\approx
-0.4003,
\qquad
p\approx3.9\times10^{-7}.
\]

The absolute effect also increased with \(|Z|\):

\[
\rho\left(
|Z_g|,
|\Delta[-\log_{10}p]|
\right)
\approx
0.2993,
\qquad
p\approx1.98\times10^{-4}.
\]

The cheap \(S\)-based and corrected \(R\)-based signed \(Z\) rankings were identical on the transition genes:

\[
\boxed{
\rho_{\mathrm{Spearman}}(Z_S,Z_R)=1.000.
}
\]

Thus the three quantities are now explicitly separated:

```text
delta_g:
    direction of shrinkage movement relative to target

LOO reliability:
    whether the raw precision estimate should be trusted

Z_g:
    how consequential / well-resolved that target departure is,
    but not whether it should be erased.
```

[S55]

<a id="de-reliability-adaptive-sg"></a>
# DE. Parameter-free reliability-adaptive \(s_g\): successful transfer as a diagnostic, insufficient as a method

## DE.1 The deliberately simple rule

The next experiment used the most transparent possible mapping:

\[
\boxed{
s_g
=
1-\mathrm{ReliabilityPercentile}_g.
}
\]

The same \(s_g\) was used for full and null precision:

\[
\frac1{\widehat\theta_{F,g}}
=
(1-s_g)\frac1{\widetilde\theta_{F,g}}
+
s_g\frac1{m_F},
\]

\[
\frac1{\widehat\theta_{0,g}}
=
(1-s_g)\frac1{\widetilde\theta_{0,g}}
+
s_g\frac1{m_0}.
\]

No truth label, tested effect, or significance measure enters \(s_g\).

The ordinary final law was intentionally kept at the **pre-detour \(\chi_1^2\)** reference so the experiment isolated the
precision layer.

## DE.2 LOO zero-deviance implementation correction

The first implementation failed because an ordinary gene could have:

\[
S_{F,-i}=0
\]

after deleting one sample. That had been treated as a missing LOO precision.

Mathematically:

\[
\theta_{raw,-i}
=
\frac{a_{F,-i}}{S_{F,-i}}
\to\infty,
\]

so the corrected implementation treated this as a boundary precision estimate and capped it at the already-existing
production `THETA_CAP`.

Any ordinary gene still lacking a finite reliability score after that correction received the explicit conservative fallback:

```text
reliability = 0
s_g = 1
```

with the frequency reported rather than hidden.

In the final Pickrell/Li2020/RNAontheBENCH results, the median reliability-fallback fraction was zero. The Pickrell splits
did have a median of approximately **1,107 genes** hitting the LOO precision cap at least once, which remains a useful
future sensitivity diagnostic.

## DE.3 Pickrell: substantial improvement, but fixed \(s=.60\) remained much better

Pickrell complete-null means were:

| Method | raw \(p<.001\) | raw \(p<.01\) | raw \(p<.05\) | BH FDR | Mean BH discoveries |
|---|---:|---:|---:|---:|---:|
| Current DEB M03 | 0.002441 | 0.013216 | 0.053123 | 0.666667 | 2.700 |
| Fixed \(s=.60\) | **0.000784** | **0.009167** | 0.050295 | **0.066667** | **0.0667** |
| Raw \(\theta\) | 0.001926 | 0.014829 | 0.065490 | 0.500000 | 1.833 |
| Reliability-adaptive \(s_g\) | **0.001288** | **0.011202** | 0.054726 | **0.266667** | **0.2667** |

Relative to current DEB, the adaptive rule gave:

```text
raw p<.01:
    28 / 30 improved
     0 equal
     2 worsened
    mean change ~ -0.002015

BH discovery count:
    19 / 30 improved
    11 equal
     0 worsened
    mean change ~ -2.433

complete-null BH FDR:
    0.6667 -> 0.2667
```

The direction is encouraging, especially in the extreme tail, but the fixed universal \(s=.60\) remained dramatically
better on Pickrell.

## DE.4 Li2020 40v40: the rule transferred in the same direction across all three datasets

On the same 49 available 40v40 splits:

| Dataset | Current DEB FPR | Reliability-adaptive FPR | Paired result |
|---|---:|---:|---|
| AdLC | 0.008471 | **0.007323** | **15 improved / 0 equal / 0 worsened** |
| OC | 0.001783 | **0.001555** | 13 improved / 1 equal / 4 worsened |

Mean FPR changes were approximately:

```text
AdLC:  -0.001149
OC:    -0.000228
TNBC:  -0.002272
```

The raw \(p<.01\) rates also improved:

```text
AdLC:
    0.039939 -> 0.037839

OC:
    0.025024 -> 0.024002

TNBC:
    0.047568 -> 0.043471
```

Mean BH discovery counts fell:

```text
AdLC:
    273.13 -> 235.00

OC:
     63.28 -> 55.11

TNBC:
    349.38 -> 276.44
```

However, complete-null BH FDR under the intentionally retained pre-detour \(\chi_1^2\) law remained 1.0 in all three
Li2020 datasets. The precision rule therefore did not solve the reference-law problem.

## DE.5 RNAontheBENCH: modest operating-point gain

On the development-only NanoString panel:

| Method | TP | FP | TN | FN | Reference FDP | MCC | AUC |
|---|---:|---:|---:|---:|---:|---:|---:|

Adaptive shrinkage therefore preserved all 18 current true positives and removed one reference false positive. It did not
match raw \(\theta\)'s 18/9 result, and raw \(\theta\) was already known to be much worse on the null benchmarks.

## DE.6 Comparison with edgeR and DESeq2 at Li2020 40v40

The reliability-adaptive M03 FPRs were much lower than edgeR exact/QL on this particular benchmark, but DESeq2 Wald still

Across these three dataset means:

Thus the adaptive precision rule moved M03 farther into the low-FPR regime, but it did not produce an overwhelming empirical

<a id="df-why-percentile-rule-not-adopted"></a>
# DF. Why the reliability-percentile rule is not adopted

## DF.1 Mechanistic plausibility is not enough

The experiment established a real empirical fact:

\[
\boxed{
\text{LOO instability contains information relevant to null-tail behavior and shrinkage harm.}
}
\]

But the mapping:

\[
\boxed{
s_g
=
1-\mathrm{ReliabilityPercentile}_g
}
\]

has no likelihood, Bayesian posterior, estimating-equation, or decision-theoretic derivation.

There is no statistical argument for why, for example, the 37th reliability percentile should imply exactly \(s_g=.63\).
The percentile transformation is a convenient rank normalization, not statistical glue to the DNB/DEB likelihood.

The empirical improvement was also not large enough to justify accepting that arbitrariness merely because it worked:

```text
Li2020:
    mean FPR reductions of roughly 13-21%

Pickrell:
    substantial improvement over current DEB,
    but much worse than fixed s=.60

RNAontheBENCH:
    one fewer reference false positive,
    no TP gain
```

This is useful development evidence, not enough for a publishable precision estimator.

## DF.2 Current status of the pre-detour DEB precision rule

The project should now describe the pre-detour ordinary DEB precision layer as **reopened and not defensible as the final
gene-specific precision method**.

The problem is not that Gamma-Gamma DEB is mathematically invalid. It is that the fitted concentration can become so large
that the supposedly gene-specific precision layer effectively collapses toward a common value, while the project's own
null and external-reference experiments show that some gene-specific raw precision information matters.

The archived M03 results should retain their original method identity. They should not be rewritten as though the new
reliability work had already been part of those fits.

## DF.3 Reliability remains a finding even though the heuristic rule is rejected

Rejecting the percentile mapping does **not** return the project to the old universal-DEB conclusion.

The positive evidence that survives is:

1. LOO precision instability is enriched in Pickrell null extremes.
2. RNAontheBENCH genes harmed by stronger shrinkage are, on average, more stable than genes helped by it.
3. The cheap \(S\)-based LOO stability ordering agrees almost perfectly with direct corrected \(R=S-C\) stability on the
   transition genes.
4. Raw precision can improve real-signal operating points but becomes too noisy under biological nulls.
5. Mean-dependent targets alone do not solve the problem.
6. Target-standardized \(Z\) predicts the direction and magnitude of shrinkage's inferential effect, but not whether the
   gene should be shrunk.

The remaining question is therefore statistical, not merely empirical:

\[
\boxed{
\text{How should uncertainty in } \theta_g \text{ enter the model itself?}
}
\]

[S54–S58]

<a id="dg-next-principled-precision-model"></a>
# DG. Current scientific position: reliability should enter as uncertainty, not as an imposed percentile shrinkage

## DG.1 Heteroskedastic measurement-error empirical Bayes is the first principled route

A natural next model is to work on:

\[
\eta_g
=
\log\theta_g.
\]

Treat the raw estimate as a noisy measurement:

\[
\widehat\eta_g\mid\eta_g
\approx
N(\eta_g,v_g),
\]

where \(v_g\) is informed by the LOO/case-deletion uncertainty.

Then model the latent precision conditionally on information:

\[
\eta_g\mid x_g
\sim
N\{m(x_g),\tau^2(x_g)\}.
\]

In the simple Gaussian case the posterior mean has an automatically reliability-adaptive form:

\[
E(\eta_g\mid\widehat\eta_g,x_g)
=
(1-s_g)\widehat\eta_g
+
s_gm(x_g),
\]

with:

\[
s_g
=
\frac{v_g}{v_g+\tau^2(x_g)}.
\]

The earlier failure of the crude plug-in \(\widehat\tau^2\) does not falsify this model class. It shows that subtracting one
median measurement variance from an observed between-gene variance is not an adequate heteroskedastic likelihood fit.

A proper likelihood must estimate the latent between-gene variation while acknowledging that different genes have different
\(v_g\).

## DG.2 A deeper route is to integrate precision uncertainty rather than plug in one shrunken \(\theta_g\)

An even more coherent possibility is to stop forcing the second stage to output a single precision point estimate before the
test.

For model \(M\):

\[
p(Y_g\mid M)
=
\int
p(Y_g\mid\theta_g,M)
p(\theta_g\mid x_g)
\,d\theta_g.
\]

The full/null comparison would then marginalize over residual-precision uncertainty directly.

Under this view:

```text
poorly identified theta:
    broad precision posterior
    -> stronger averaging automatically

well identified theta:
    concentrated precision posterior
    -> gene-specific residual information naturally survives
```

No arbitrary percentile-to-\(s\) conversion is required.

## DG.3 Current decomposition of the two-group DGE problem

At the new cutoff, the project should keep the following layers distinct:

\[
\boxed{
\text{WEB-DP / DNB count-curvature backbone}
}
\]

\[
\boxed{
\text{precision uncertainty / borrowing}
}
\]

\[
\boxed{
\text{contrast geometry}
}
\]

\[
\boxed{
\text{conditional reference law of }Q
\text{ in coverage / sample-dominance space}.
}
\]

The October 3–4 conditional-null results remain important. The failure to justify the heuristic reliability mapping does not
invalidate the hurdle/partial-pooling diagnosis, and the hurdle-law work does not resolve the precision-borrowing problem.

## DG.4 Current decision

The parameter-free reliability-adaptive rule is archived as a **diagnostic experiment**, not promoted into M03.

The next precision-layer work should therefore be:

```text
1. retain the LOO reliability result as evidence that precision uncertainty matters;

2. do not tune a nonlinear percentile-to-s_g mapping on the current development datasets;

3. formulate and fit a proper heteroskedastic uncertainty model for log theta,
   or marginalize theta uncertainty directly in the full/null likelihood;

4. keep the conditional-null reference-law problem separate while the precision model is rebuilt;

5. freeze any statistically derived replacement before a new untouched
   alternative-containing benchmark is opened.
```

The strongest current precision-layer statement is:

\[
\boxed{
\text{gene-specific precision reliability appears real;}
\quad
\text{the statistically justified way to use it remains unsolved.}
}
\]

[S58]

**[S58] October 5 afternoon precision-layer synthesis and method decision.**
Conversation interpretation after the mean-trend EB, LOO reliability, departure-\(Z\), and parameter-free \(s_g\)
transfer experiments. Supports the decision that LOO precision reliability is a real and useful diagnostic but that
\(s_g=1-\) reliability percentile lacks sufficient statistical justification for adoption; the pre-detour fitted-DEB
precision layer is therefore reopened rather than treated as final; and the next principled direction is either a proper
heteroskedastic measurement-error empirical-Bayes model for \(\log\theta_g\) using gene-specific uncertainty or direct
marginalization of \(\theta_g\) uncertainty in the full/null likelihood.
